# What Is Classification?

*Understand how classification maps input features to classes and how its performance is evaluated.*

## Supervised Learning and Prediction Targets

- Predicting a fruit's weight from its color and size requires examples that record both the measurements and the actual weight.
- Supervised learning learns relationships from data containing paired inputs and known targets. The trained model predicts outcomes for new inputs.

### Regression and Classification

- Regression predicts a numerical quantity, such as a fruit's weight.
- Classification predicts a distinct category, such as apple or orange. Each category is called a class.

### Types of Classification

- Binary classification selects one of two classes, such as legitimate or spam.
- Multiclass classification selects one of three or more classes, such as apple, orange, or banana.

## Mapping Measurements to Classes

### Input Features and Target Labels

- A feature is an attribute used as an input, such as color or shape. In classification, a label identifies the observed object's actual class.
- Let the fruit's hue and length-to-width ratio be the input $\mathbf{x}$, and its actual type be the target label $y$.

  $\displaystyle \mathbf{x}=(x_1,x_2), \qquad y\in\{\text{Apple},\text{Orange},\text{Banana}\}.$

- $x_1$ is hue, which represents color as an angle. Red, orange, and yellow are near 0, 30, and 60 degrees, respectively.
- $x_2$ is length divided by width. It is close to 1 for round fruit and larger for elongated fruit.

### Training and Prediction

- Training uses features and target labels together to determine the model's parameters.
- Prediction passes a new fruit's features to the trained model to obtain a predicted label $\hat y$. Changing the input leaves the learned parameters unchanged.

### Class Probabilities and Predicted Labels

- A probabilistic classifier estimates the probability $p_k$ of class $k$ for input $\mathbf{x}$. Selecting the class with the highest probability gives:

  $\displaystyle p_k=P(y=k\mid\mathbf{x}), \qquad \hat y=\underset{k}{\arg\max}\;p_k.$

- $\arg\max$ returns the class associated with the largest value, rather than the value itself. Comparing the predicted label $\hat y$ with the actual label $y$ shows whether the classification is correct.

### Generalization Performance

- Generalization means that learned relationships also apply to unseen examples. Correctly separating training fruit alone does not establish performance on new fruit.
- Even a prediction with a high probability can be wrong when features overlap or inputs fall outside the training range. Predictions must therefore be compared with actual labels on separate data.

### Training, Validation, and Test Data

- Training data determines the model's parameters.
- Validation data is used to select settings such as the number of tree splits or a probability threshold. Repeated selection can lead to overfitting to the validation data.
- Test data is excluded from training and setting selection, and is used to evaluate the final model's generalization performance.
- The fruit and messages in these notebooks are synthetic data. Assessing performance in practice requires test data representative of the intended environment.

## Takeaway

Classification maps features to discrete classes. Separate training, validation, and test data help distinguish learning from evaluation.
