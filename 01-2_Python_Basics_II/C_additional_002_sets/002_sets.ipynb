*Use a set when uniqueness and membership matter more than position or repeated values.*

## Method · Unique-Value Sets

Repeated input values appear only once in a set. Do not rely on a set's printed order.

```python
observed_islands = {"Biscoe", "Dream", "Biscoe", "Torgersen"}

print("Unique islands:", sorted(observed_islands))
print("Unique count:", len(observed_islands))
print("Contains Dream:", "Dream" in observed_islands)
```

## Implementation · Set Comparisons

`&` keeps shared values, `|` keeps values from either set, and `-` keeps values found only on the left.

```python
survey_a = {"Biscoe", "Dream", "Torgersen"}
survey_b = {"Biscoe", "Dream", "New Island"}

print("In both:", sorted(survey_a & survey_b))
print("In either:", sorted(survey_a | survey_b))
print("Only in survey A:", sorted(survey_a - survey_b))
```

## Implementation · Set Mutation

Use `discard()` when an absent value is acceptable because it does not raise an error for a missing value.

```python
survey_a.add("New Island")
survey_a.discard("Unknown Island")
print(sorted(survey_a))
```

### Interpretation · Set Mutation

Choose a set to answer “which unique values occur?” or “which values overlap?” Convert to a sorted list only when a predictable display order is needed.
