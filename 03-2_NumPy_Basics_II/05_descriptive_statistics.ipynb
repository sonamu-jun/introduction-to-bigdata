## Center and Range

The mean is the sum divided by the number of values. The median is
the middle of the sorted values; for an even count, it averages the
two middle values. `np.ptp()` gives the range, `maximum - minimum`.

**Question.** The ten sorted scores have 84 and 85 in the two middle
positions. What median will NumPy report?


```python
import numpy as np

scores = np.array([72, 85, 90, 68, 77, 95, 88, 91, 73, 84], dtype=float)

print("Scores:", scores)
print("Sorted scores:", np.sort(scores))
print("Mean:", scores.mean())
print("Median:", np.median(scores))
print("Minimum:", scores.min())
print("Maximum:", scores.max())
print("Range:", np.ptp(scores))
```

The median is `(84 + 85) / 2 = 84.5`, and the mean is 82.3. The
minimum is 68, the maximum is 95, and the range is 27 points. Mean
and median summarize the same data in different ways.


## Variance and Standard Deviation

Variance summarizes squared distances from the mean. Standard
deviation is its square root, so it uses the same units as the data.
By default, `var()` and `std()` divide by `N`, the number of values
(`ddof=0`). This describes the spread of these ten recorded scores.

`ddof=1` uses `N - 1` instead, the usual sample-variance denominator
when estimating population variance from a sample. Choose the
convention according to the question and label the result.


```python
population_variance = scores.var(ddof=0)
population_std = scores.std(ddof=0)
sample_variance = scores.var(ddof=1)
sample_std = scores.std(ddof=1)

print("Variance (ddof=0):", round(population_variance, 2))
print("Standard deviation (ddof=0):", round(population_std, 2))
print("Variance (ddof=1):", round(sample_variance, 2))
print("Standard deviation (ddof=1):", round(sample_std, 2))
```

The variance is 76.41 with `ddof=0` and 84.90 with `ddof=1`. The
corresponding standard deviations are about 8.74 and 9.21 points.
The smaller denominator makes the `ddof=1` result larger.


## Statistics by Student or Subject

Each row in this table is a student and each column is a subject.
Use `axis=0` to summarize students within each subject, or `axis=1`
to summarize subjects within each student.


```python
class_scores = np.array([[72, 85, 90],
                         [68, 77, 95],
                         [88, 91, 73],
                         [84, 79, 87]], dtype=float)

print("Subject means:", class_scores.mean(axis=0))
print("Student means:", class_scores.mean(axis=1))
print("Subject standard deviations (ddof=0):", np.round(class_scores.std(axis=0), 2))
```

Subject means are `[78., 83., 86.25]`, one result per column.
Student means are approximately `[82.33, 80., 84., 83.33]`, one
result per row. An overall mean alone would hide these differences.
