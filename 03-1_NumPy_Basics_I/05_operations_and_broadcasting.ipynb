# Operations and Broadcasting

*Apply elementwise arithmetic and match array shapes when sharing values across rows or columns.*


## Elementwise Arithmetic

For arrays with the same shape, `+`, `-`, `*`, `/`, and `**` calculate
at matching positions. Division returns fractional values when needed.
A scalar is applied to every position.


```python
import numpy as np

values = np.array([1, 2, 3])
factors = np.array([10, 20, 30])

print("Addition:", values + factors)
print("Subtraction:", values - factors)
print("Multiplication:", values * factors)
print("Division:", values / factors)
print("Squares:", values ** 2)
print("Add a scalar:", values + 5)
```

Multiplication gives `[10, 40, 90]`, division gives `[0.1, 0.1, 0.1]`,
and adding 5 gives `[6, 7, 8]`. These are separate results; the inputs
retain their values.


## One Price per Product

Each row below is a branch and each column is a product. Multiplying
unit counts by a price vector applies the same product prices to every
branch. Broadcasting matches dimensions from the right. A pair is
compatible when its lengths are equal or one is 1; missing leading
dimensions are treated as 1.

| Array | Shape | Meaning |
|---|---|---|
| `sales` | `(3, 4)` | Three branches, four products |
| `prices` | `(4,)` | Four product prices |
| `sales * prices` | `(3, 4)` | Revenue for each branch and product |

**Question.** Which price multiplies `sales[1, 2]`, and what revenue results?


```python
sales = np.array([[120, 80, 95, 110],
                  [100, 90, 105, 115],
                  [130, 70, 100, 120]])
prices = np.array([2000, 1500, 1200, 3500])
revenue = sales * prices

print("Sales shape:", sales.shape)
print("Price shape:", prices.shape)
print("Revenue (KRW):")
print(revenue)
print("Branch 1, product 2 revenue:", revenue[1, 2])
```

Column 2 uses the price 1,200, so 105 units produce 126,000 KRW.
The four prices align with the four columns and are reused for all
three rows. The result retains shape `(3, 4)`.


## One Adjustment per Branch

A `(3,)` array aligns with the last axis, whose length is 4 here, so
it cannot supply one adjustment per row. Reshape the three values to
`(3, 1)` to retain the row axis and share each value across four columns.


```python
branch_adjustments = np.array([-5, 0, 5])

try:
    sales + branch_adjustments
except ValueError:
    print("Shapes (3, 4) and (3,) cannot broadcast together.")

adjustment_column = branch_adjustments.reshape(3, 1)
adjusted_sales = sales + adjustment_column

print("Adjustment shape:", adjustment_column.shape)
print("Adjusted unit counts:")
print(adjusted_sales)
```

The first row decreases by 5 at every position, the second row stays
unchanged, and the third row increases by 5. Shape `(3, 1)` expresses
one value per branch, while shape `(4,)` expresses one value per product.
