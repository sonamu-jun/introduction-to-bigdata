## Method · Standard-Library Module Imports

A **module** is one Python file that provides reusable code, while a **package** groups modules in a folder. Python includes standard-library modules; other packages can be installed separately.

`csv`, `statistics`, and `pathlib` come with Python. Each `import` makes its reusable tools available in this notebook.

The repository copy is `input/penguins.csv`. In JSPCV Playground, **Add files** stores it as `uploads/penguins.csv`; a Colab sidebar upload appears as `penguins.csv`. The code checks these locations in that order.

```python
import csv
import statistics
from pathlib import Path

local_path = Path("input/penguins.csv")
playground_path = Path("uploads/penguins.csv")
colab_path = Path("penguins.csv")

if local_path.exists():
    input_path = local_path
elif playground_path.exists():
    input_path = playground_path
else:
    input_path = colab_path
print(input_path.name)
```

## Implementation · Module Tool Use

`csv.DictReader` reads each row as a dictionary. The loop keeps the available body-mass values, and `statistics.mean` calculates their mean.

```python
body_masses_g = []

with input_path.open(mode="r", encoding="utf-8", newline="") as file:
    records = csv.DictReader(file)
    for record in records:
        body_mass_text = record["body_mass_g"]
        has_body_mass = body_mass_text != "" and body_mass_text != "NA"
        if has_body_mass:
            body_masses_g.append(float(body_mass_text))

print("Number of values:", len(body_masses_g))
print("Mean body mass (g):", round(statistics.mean(body_masses_g), 1))
```

### Interpretation · Module Tool Use

Packages such as `pandas` are installed separately and then imported in the same way. Installation is an environment task; `import` is the step used inside Python code.
