# Series and DataFrames

*Distinguish labeled values from a table, and preserve the difference between missing data and zero.*

A pandas **Series** attaches index labels to one sequence of values. A **DataFrame** holds named columns under a shared row index; each column can have its own dtype.


## Labels, Columns, and Missing Values

`.loc[label]` selects by label; `.iloc[position]` selects by integer position. Series arithmetic matches labels even when their positions differ.

`table["score"]` returns a Series. `table[["score"]]` retains a one-column DataFrame, so its shape still has two axes.

`pd.NA` marks missing data. The nullable dtype `"Int64"` holds integers and missing values together. `.isna()` identifies missing entries; `.mean()` skips them by default. A recorded zero still participates in the mean.


**Think before running or viewing the saved output.** With scores `[70, 0, missing]`, what should the mean be? Would replacing the missing score with zero preserve the meaning of the data?

Distinguish a recorded zero from an absent measurement.


```python
import pandas as pd

scores = pd.Series([70, 0, pd.NA], index=["S01", "S02", "S03"], dtype="Int64")
study_table = pd.DataFrame({"minutes": [30, 45, 40], "score": scores})

print(study_table)
print("Table shape:", study_table.shape)
print("Score Series shape:", study_table["score"].shape)
print("One-column table shape:", study_table[["score"]].shape)
print("Minutes for label S02:", study_table.loc["S02", "minutes"])
print("Score at position 1:", scores.iloc[1])
print("Missing scores:", scores.isna().tolist())
print("Mean recorded score:", scores.mean())
```

**Check:** The mean is `(70 + 0) / 2 = 35.0`. Filling the missing score with zero would count an unobserved result. The complete table has shape `(3, 2)`; its score Series has `(3,)`, and the one-column table has `(3, 1)`.
