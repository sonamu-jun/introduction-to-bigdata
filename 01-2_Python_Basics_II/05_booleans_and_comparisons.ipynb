# Booleans and Comparisons

*Create Boolean values with comparison operators and combine them into an explicit decision rule.*

## Value Comparisons

The comparison operators include `<`, `<=`, `>`, `>=`, `==`, and `!=`.


```python
score = 82

print("At least 60:", score >= 60)
print("Exactly 100:", score == 100)
print("Not zero:", score != 0)
print("Between 0 and 100:", 0 <= score <= 100)
```

## Compound Conditions

`and` requires both conditions, `or` requires at least one, and `not` reverses a Boolean value.


```python
attendance_rate = 0.90
submitted_project = True

passed_course = score >= 60 and attendance_rate >= 0.80
eligible_for_review = score < 60 or not submitted_project

print("Passed course:", passed_course)
print("Needs review:", eligible_for_review)
print("Result type:", type(passed_course))
```

## Truth Values of Existing Objects

`bool(value)` reveals how Python interprets a value in a Boolean context. Zero, `None`, and empty strings or collections are false; most other values are true.

```python
print("Zero:", bool(0))
print("Nonzero number:", bool(12))
print("Empty text:", bool(""))
print("Non-empty text:", bool("Python"))
print("Empty list:", bool([]))
print("List with data:", bool([72, 85]))
print("None:", bool(None))
```

Truthiness lets a later conditional ask `if values:` instead of comparing `len(values)` with zero. Explicit comparisons remain better when a specific numeric boundary matters.

## Short-Circuit Evaluation

Python evaluates `and` and `or` from left to right. `and` stops after the first false operand, while `or` stops after the first true operand.

```python
profile = None
has_display_name = profile is not None and "display_name" in profile

entered_name = ""
display_name = entered_name.strip() or "Guest"

print("Has display name:", has_display_name)
print("Display name:", display_name)
```

The membership test never runs when `profile is not None` is false, preventing an invalid lookup in `None`. The `or` expression returns the fallback `Guest` because the cleaned name is empty.

## Use Cases

Booleans commonly record whether data is valid, whether required information exists, and whether several requirements are satisfied together.

### Record a Data-Validation Result

A numeric comparison can produce a reusable validity flag without immediately choosing a code branch.

```python
body_mass_g = 3750
has_plausible_mass = 2500 <= body_mass_g <= 6500
is_suspiciously_round = body_mass_g % 1000 == 0

print("Plausible body mass:", has_plausible_mass)
print("Suspiciously round value:", is_suspiciously_round)
```

The range comparison stores `True` because 3750 lies inside the allowed interval. The second comparison stores `False` because 3750 is not evenly divisible by 1000.

### Check Whether Required Data Exists

Membership tests produce Boolean values. A set relationship can check several required column names at once.

```python
available_columns = {"species", "island", "body_mass_g"}
required_columns = {"species", "body_mass_g"}

has_species = "species" in available_columns
has_all_required_columns = required_columns.issubset(available_columns)
is_missing_required_columns = not has_all_required_columns

print("Has species column:", has_species)
print("Has all required columns:", has_all_required_columns)
print("Missing a required column:", is_missing_required_columns)
```

Both required names occur in `available_columns`, so the subset test returns `True`. Applying `not` creates the opposite missing-data flag.

### Combine Access Requirements

`all()` returns `True` only when every Boolean in a collection is true. `any()` returns `True` when at least one Boolean is true.

```python
is_signed_in = True
has_download_permission = True
account_locked = False

access_checks = [is_signed_in, has_download_permission, not account_locked]
warning_flags = [account_locked, not has_download_permission]

can_download = all(access_checks)
needs_attention = any(warning_flags)

print("Can download:", can_download)
print("Needs attention:", needs_attention)
```

All three access checks are true, so downloading is allowed. Neither warning flag is true, so `needs_attention` is `False`. The next lesson uses flags like these to select a branch.

### Optional Feature Flags

A Boolean variable can record whether an optional feature is enabled, even before the program decides how to act on it.

```python
show_grid = True
save_image = False

display_options = {
    "show_grid": show_grid,
    "save_image": save_image,
}

print("Display options:", display_options)
print("Grid enabled:", display_options["show_grid"])
```

The settings store yes-or-no choices directly. A later conditional can read each flag and enable only the selected behavior.

### Process-State Flags

Comparing progress values creates state flags that can be stored, displayed, or passed to later code.

```python
total_rows = 120
processed_rows = 120
error_count = 0

processing_complete = processed_rows == total_rows
has_errors = error_count > 0

print("Processing complete:", processing_complete)
print("Has errors:", has_errors)
```

Equal row counts produce a true completion flag, while zero errors produce a false error flag. The variable names state exactly what each Boolean means.

Naming a combined condition makes the rule inspectable. The result is a `bool`, which the next lesson can use to choose a branch.
