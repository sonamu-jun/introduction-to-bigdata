# Arrays

*Explain indexed storage and use a typed array to access and update numeric values.*

A **data structure** organizes values and defines how we access or change them.
Python lists, tuples, dictionaries, and sets already give us different rules:
positions, fixed groupings, key-value associations, and unique membership.

An **array** places elements in consecutive storage locations so each element
can be reached by its index. Python's `array` module stores numeric values
with one declared type. A Python list instead stores references and may hold
objects of different types.


## Indexed Storage

`array("i", values)` creates an array of signed integers. Indexing starts at
zero, and a slice such as `[2:4]` includes index 2 and excludes index 4.

```text
Index      0     1     2     3     4
        +-----+-----+-----+-----+-----+
Value   |  1  |  2  |  3  |  4  |  5  |
        +-----+-----+-----+-----+-----+
```

The common element size lets an array locate a position directly. It does
not need to follow links through all earlier elements.


**Question.** What values does `numbers[2:4]` select, and how many elements does it contain?


````python
from array import array

numbers = array("i", [1, 2, 3, 4, 5])

print("Values:", numbers.tolist())
print("First value:", numbers[0])
print("Slice [2:4]:", numbers[2:4].tolist())
print("Slice length:", len(numbers[2:4]))
````

The slice is `[3, 4]`, with two elements. Index 2 selects the third value,
and the stop index 4 is excluded. The original five values are unchanged.


## Updating a Typed Sequence

Indexed assignment replaces a value; `append()` adds one at the end.
This Python array can grow even though each element has a fixed storage type.
`itemsize` reports bytes per element, so `len(numbers) * numbers.itemsize`
gives the bytes occupied by the stored values, excluding object overhead
and any unused capacity.

A direct indexed lookup takes constant time, written `O(1)`: the amount of
work does not grow with the number of elements. Inserting in the middle
may shift later values and takes `O(n)` work for `n` elements. These symbols
describe how work grows, rather than a measured number of seconds.


````python
numbers[1] = 20
numbers.append(6)

print("Updated values:", numbers.tolist())
print("Element count:", len(numbers))
print("Bytes per element:", numbers.itemsize)
print("Stored value bytes:", len(numbers) * numbers.itemsize)

try:
    numbers.append("seven")
except TypeError:
    print("Rejected: a string cannot be stored in this integer array.")
````

The array now holds `[1, 20, 3, 4, 5, 6]`. The failed string insertion leaves
those six integers intact. The printed byte count concerns the numeric
storage; it is not a complete comparison of Python object memory.

The type code and `itemsize` describe the array's storage on the running
platform. [Python 3.12 array documentation](https://docs.python.org/3.12/library/array.html)
