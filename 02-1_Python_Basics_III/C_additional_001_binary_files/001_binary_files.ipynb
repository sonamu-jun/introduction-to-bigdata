## Method · Text Encoding as Bytes

`encode()` converts text to bytes. `decode()` reverses the conversion when the same encoding is used.

```python
from pathlib import Path

lesson_dir = Path.cwd()
output_dir = lesson_dir / "output"
output_dir.mkdir(exist_ok=True)

message = "Penguin observations: 5"
encoded = message.encode("utf-8")
restored_message = encoded.decode("utf-8")

print("Python type:", type(encoded))
print("Encoded bytes:", encoded)
print("Decoded text:", restored_message)
```

## Implementation · Structured-Record Serialization

The `pickle` module preserves Python-specific structure. Use it only for files you created and trust.

```python
import pickle

penguin = {
    "species": "Adelie",
    "island": "Torgersen",
    "body_mass_g": 3750,
}
record_path = output_dir / "penguin_record.pkl"

with record_path.open(mode="wb") as file:
    pickle.dump(penguin, file)

with record_path.open(mode="rb") as file:
    restored_penguin = pickle.load(file)

print("Restored record:", restored_penguin)
print("Exact round trip:", restored_penguin == penguin)
print("Saved:", record_path.relative_to(lesson_dir))
```

### Interpretation · Structured-Record Serialization

The byte sequence needs UTF-8 to recover its text. The `pickle` file records Python object structure, but loading an untrusted `pickle` file is unsafe because it can execute embedded instructions.
