# Comprehensions

*Compress a simple “transform each value” loop only after the expanded loop is understood.*


## Expanded Transformation Loop

This loop creates a new list without changing `numbers`.


```python
numbers = [1, 2, 3, 4, 5]
squares_with_loop = []

for number in numbers:
    squares_with_loop.append(number ** 2)

print(squares_with_loop)
```

## Compact Transformations

The expression `number ** 2` is evaluated once for each `number` in `numbers`.


```python
squares = [number ** 2 for number in numbers]
print(squares)
print("Same result:", squares == squares_with_loop)
```

## Filtered Comprehensions

The condition comes last but is checked before the value is added to the result.


```python
even_squares = [number ** 2 for number in numbers if number % 2 == 0]
print(even_squares)
```

## Dictionary Comprehensions

Placing `key: value` before `for` builds a dictionary instead of a list. Here, each number becomes a key and its square becomes the value.


```python
square_lookup = {number: number ** 2 for number in numbers}
print(square_lookup)
```

Use a comprehension for one short transformation and at most one clear filter.
