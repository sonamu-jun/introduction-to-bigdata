# Boolean Filtering

*Build row conditions and combine them to select matching records from a table.*


## A Boolean Mask Selects Rows

A column comparison produces one Boolean for each row. `table[mask]`
keeps rows marked `True`. Build the mask from the table being filtered
so its index matches the table's rows.


```python
import pandas as pd

sales_df = pd.DataFrame({
    "employee": ["Minji", "Joon", "Hana", "Taeho", "Sora"],
    "team": ["A", "B", "A", "B", "A"],
    "sales": [120, 95, 140, 110, 80],
})
high_sales_mask = sales_df["sales"] >= 100

print("Employee sales:")
print(sales_df)
print("Sales at least 100:")
print(high_sales_mask)
print("Matching employees:")
print(sales_df[high_sales_mask])
```

Minji, Hana, and Taeho meet the threshold. Their original row labels remain 0, 2, and 3; filtering does not renumber them.


## Combine Conditions

Use `&` for AND, `|` for OR, and `~` for NOT on Boolean Series.
Put each comparison in parentheses because these operators have their
own precedence rules. Python's `and` and `or` do not combine Series element by element.


```python
team_a_high_mask = (sales_df["team"] == "A") & (sales_df["sales"] >= 100)

print("Team A employees reaching 100:")
print(sales_df.loc[team_a_high_mask, ["employee", "sales"]])
```

Minji and Hana satisfy both conditions, with sales of 120 and 140. Sora belongs to team A but has only 80 sales.


## Alternative and Complementary Conditions

OR keeps a row if either condition is true. NOT reverses a Boolean mask.
`.loc[mask, columns]` combines row filtering and column selection in one expression.


```python
team_b_or_high_mask = (sales_df["team"] == "B") | (sales_df["sales"] >= 130)

print("Team B or sales at least 130:")
print(sales_df.loc[team_b_or_high_mask, ["employee", "team", "sales"]])
print("Below the 100-sales threshold:")
print(sales_df.loc[~high_sales_mask, ["employee", "sales"]])
```

The OR condition selects Joon, Hana, and Taeho. Reversing the original threshold mask selects Joon and Sora.


Express selection criteria as parenthesized comparisons, combine them with `&` or `|`, and inspect which rows the resulting mask retains.
