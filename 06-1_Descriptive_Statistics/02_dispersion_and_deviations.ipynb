# Dispersion and Deviations

<span style="color: #777777; font-style: italic;">산포와 편차</span>


- `Dispersion` describes how widely values are spread.  
  <span style="color: #777777; font-style: italic;">`산포`(dispersion)는 값들이 얼마나 넓게 퍼져 있는지를 나타냅니다.</span>


## Range

<span style="color: #777777; font-style: italic;">범위</span>

- The `range` is the largest value minus the smallest: $\max(x)-\min(x)$.  
  <span style="color: #777777; font-style: italic;">`범위`(range)는 최댓값에서 최솟값을 뺀 값입니다: </span>${\color{#777777}\max(x)-\min(x)}$<span style="color: #777777; font-style: italic;">.</span>

- `np.ptp()` calculates this difference.  
  <span style="color: #777777; font-style: italic;">`np.ptp()`로 이 차이를 계산합니다.</span>


```python
import numpy as np

values = np.array([1, 3, 5, 7, 9])

print("Range:", np.ptp(values))
```

- The range is 9 − 1 = 8; it uses only the two end values.  
  <span style="color: #777777; font-style: italic;">범위는 9 − 1 = 8이며, 양 끝값 두 개만 사용합니다.</span>


## Deviation

<span style="color: #777777; font-style: italic;">편차</span>

- A `deviation` is a value minus the mean: $d_i=x_i-\bar{x}$.  
  <span style="color: #777777; font-style: italic;">`편차`(deviation)는 각 값에서 평균을 뺀 값입니다: </span>${\color{#777777}d_i=x_i-\bar{x}}$<span style="color: #777777; font-style: italic;">.</span>

- Positive and negative deviations add to zero; squaring makes them nonnegative.  
  <span style="color: #777777; font-style: italic;">양수·음수 편차의 합은 0이며, 제곱하면 모두 0 이상이 됩니다.</span>


```python
deviations = values - values.mean()
squared_deviations = deviations**2

print("Deviations:", deviations)
print("Sum:", deviations.sum())
print("Squared sum:", squared_deviations.sum())
```

- The mean is 5, so the deviations are -4, -2, 0, 2, and 4.  
  <span style="color: #777777; font-style: italic;">평균은 5이므로 편차는 -4, -2, 0, 2, 4입니다.</span>

- The deviations sum to 0; their squares sum to 40.  
  <span style="color: #777777; font-style: italic;">편차의 합은 0이고 편차 제곱합은 40입니다.</span>


## Variance and Standard Deviation

<span style="color: #777777; font-style: italic;">분산과 표준편차</span>

- Treat these five values as the full population: `variance` is the mean of the squared deviations.  
  <span style="color: #777777; font-style: italic;">다섯 값을 모집단 전체로 보면, `분산`(variance)은 편차를 제곱한 값의 평균입니다.</span>

- `np.var(values, ddof=0)` divides by the number of values; `standard deviation` is its square root.  
  <span style="color: #777777; font-style: italic;">`np.var(values, ddof=0)`은 값의 개수로 나누며, `표준편차`(standard deviation)는 그 제곱근입니다.</span>

- If values are in minutes, variance is in minutes squared and standard deviation is in minutes.  
  <span style="color: #777777; font-style: italic;">값의 단위가 분이면 분산은 분의 제곱, 표준편차는 분 단위입니다.</span>


```python
population_variance = np.var(values, ddof=0)
population_std = np.sqrt(population_variance)

print("Variance:", population_variance)
print("Std:", round(population_std, 4))
```

- The `population variance` is 40 / 5 = 8, and the `standard deviation` is about 2.8284.  
  <span style="color: #777777; font-style: italic;">`모분산`(population variance)은 40 / 5 = 8이고 `표준편차`는 약 2.8284입니다.</span>

- This describes `spread`; each value need not be exactly this far from the `mean`.  
  <span style="color: #777777; font-style: italic;">이는 흩어짐의 크기이며, 모든 값이 `평균`에서 정확히 그만큼 떨어져 있다는 뜻은 아닙니다.</span>


## Same Mean, Different Spread

<span style="color: #777777; font-style: italic;">같은 평균과 다른 산포</span>

- The groups [4, 5, 6] and [1, 5, 9] have the same mean; compare their standard deviations.  
  <span style="color: #777777; font-style: italic;">[4, 5, 6]과 [1, 5, 9]는 평균이 같으며, 표준편차로 흩어진 정도를 비교합니다.</span>


- Both groups have mean 5; the values 1, 5, and 9 have more spread.  
  <span style="color: #777777; font-style: italic;">두 집단의 평균은 5이며, 값이 1·5·9인 집단이 더 넓게 퍼져 있습니다.</span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/06-1_Descriptive_Statistics/assets/02_dispersion_and_deviations/concept.webp" width="439" alt="Both groups have mean 5; the values 1, 5, and 9 have more spread.">


```python
compact = np.array([4, 5, 6])
spread = np.array([1, 5, 9])

print("Compact mean:", compact.mean())
print("Spread mean:", spread.mean())
print("Compact std:", round(compact.std(), 4))
print("Spread std:", round(spread.std(), 4))
```

- Both means are 5, but the standard deviations are about 0.8165 and 3.2660.  
  <span style="color: #777777; font-style: italic;">평균은 모두 5이지만 표준편차는 약 0.8165와 3.2660입니다.</span>

- The second group has a larger standard deviation because its values are farther from 5.  
  <span style="color: #777777; font-style: italic;">두 번째 집단의 값이 5에서 더 멀리 떨어져 있어 표준편차가 큽니다.</span>
