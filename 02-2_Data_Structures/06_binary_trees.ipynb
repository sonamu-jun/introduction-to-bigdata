# Binary Trees

*Build a binary search tree and compare inorder, preorder, and postorder traversal.*

A **tree** is connected and has no cycles. Choosing a root gives every
other node one parent. A **binary tree** gives each node at most two
children, called left and right; a node without children is a **leaf**.

A **binary search tree** adds an ordering rule: values in a node's left
subtree are smaller, and values in its right subtree are larger. This
example stores distinct values and ignores duplicate insertions.


## Building a Binary Search Tree

The first value, `27`, becomes the root. Each new value is compared with
the current node and follows a left or right reference until a free child
position is found.

```text
            27
           /  \
         14    35
        /  \  /  \
       10  19 31  42
```

`insert()` calls itself on a smaller subtree. This **recursion** stops
when it creates a child or finds an equal value. Every node has a value,
including zero; `None` is used only for missing child references.


````python
class BinarySearchTree:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

    def insert(self, data):
        if data < self.data:
            if self.left is None:
                self.left = BinarySearchTree(data)
            else:
                self.left.insert(data)
        elif data > self.data:
            if self.right is None:
                self.right = BinarySearchTree(data)
            else:
                self.right.insert(data)


values = [27, 14, 35, 10, 19, 31, 42]
root = BinarySearchTree(values[0])
for value in values[1:]:
    root.insert(value)

print("Root:", root.data)
print("Root's left child:", root.left.data)
print("Root's right child:", root.right.data)
````

The root is `27`, with children `14` and `35`. The root's entire left
subtree contains smaller values, while its entire right subtree contains
larger ones. A general binary tree does not require this ordering rule.


## Traversal Order

Traversal visits every node. The three orders differ in when the current
node is visited relative to its subtrees.

| Traversal | Order |
| --- | --- |
| Inorder | Left subtree, current node, right subtree |
| Preorder | Current node, left subtree, right subtree |
| Postorder | Left subtree, right subtree, current node |

Each function returns an empty list for a missing child. Otherwise it
combines the subtree results with the current value using list `+`.


**Question.** Which traversal produces ascending values in this search tree, and where does `27` appear in preorder and postorder?


````python
def inorder(node):
    if node is None:
        return []
    return inorder(node.left) + [node.data] + inorder(node.right)


def preorder(node):
    if node is None:
        return []
    return [node.data] + preorder(node.left) + preorder(node.right)


def postorder(node):
    if node is None:
        return []
    return postorder(node.left) + postorder(node.right) + [node.data]


print("Inorder:", inorder(root))
print("Preorder:", preorder(root))
print("Postorder:", postorder(root))
````

Inorder gives `[10, 14, 19, 27, 31, 35, 42]` because the search-tree rule
orders the subtrees. Preorder starts with `27`; postorder ends with `27`.
All three results contain the same seven nodes in different orders.

Insertion follows one downward path, so its cost depends on tree height.
A balanced tree has short paths; inserting already sorted values into
this implementation can produce a chain with `O(n)` insertion time.
