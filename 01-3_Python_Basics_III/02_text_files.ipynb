# Text Files

*Learn to load, save, append, and reopen UTF-8 text files with Python's built-in `open()`.*

A text-file workflow starts with a supplied input, writes only to a separate workspace, and reopens the result to confirm what was stored.


## Before You Run

In VSCode, open the repository and run this notebook from either the repository root or the `01-3_Python_Basics_III` directory. The repository already includes `study_plan.txt` in `assets/02_text_files`, so no upload or file copy is needed.

For JupyterLite compatibility, select `study_plan.txt` in **Add files** and enter `assets/02_text_files` for **Save folder**. This makes the file available at `/home/pyodide/assets/02_text_files/study_plan.txt`. The upload destination is resolved from `/home/pyodide`, while a Python relative path is resolved from the directory returned by `os.getcwd()`.


## Setting Input and Output Directories

Start from the local repository or chapter directory, locate the committed input with `os.path`, and keep generated files in `outputs/02_text_files`. JupyterLite follows the same relative layout through its compatibility branch.


```python
import os

chapter_name = "01-3_Python_Basics_III"
launch_directory = os.getcwd()
local_chapter_directory = os.path.join(launch_directory, chapter_name)

if os.path.basename(launch_directory) == chapter_name:
    base_directory = launch_directory
    execution_mode = "Local repository"
elif os.path.isdir(local_chapter_directory):
    base_directory = local_chapter_directory
    execution_mode = "Local repository"
elif os.path.isdir("/home/pyodide"):
    base_directory = "/home/pyodide"
    execution_mode = "JupyterLite"
else:
    raise FileNotFoundError(
        "Run this notebook from the repository root or "
        f"the {chapter_name} directory."
    )

os.chdir(base_directory)

input_directory = os.path.join(base_directory, "assets", "02_text_files")
input_path = os.path.join(input_directory, "study_plan.txt")

assert os.path.isfile(input_path), (
    "Supplied file not found in assets/02_text_files. "
    "Restore the repository asset, or add it there when using JupyterLite."
)

output_directory = os.path.join(base_directory, "outputs", "02_text_files")
os.makedirs(output_directory, exist_ok=True)
output_path = os.path.join(output_directory, "study_plan_updated.txt")

print("Execution mode:", execution_mode)
print("Current directory name:", os.path.basename(os.getcwd()))
print("Input folder:", os.path.relpath(input_directory, base_directory))
print("Input file:", os.path.basename(input_path))
print("Output folder:", os.path.relpath(output_directory, base_directory))
print("Output directory exists:", os.path.isdir(output_directory))
```

The committed input and Python path match in VSCode without an upload step. Output writes go to a separate folder, so a full Run All never overwrites the supplied study plan. JupyterLite uses the same relative input and output paths.


## File Mode Reference

| Mode | Purpose | Existing content |
|---|---|---|
| `"r"` | Read text | Preserved; file must exist |
| `"w"` | Write text | Replaced, or a new file is created |
| `"a"` | Append text | Preserved; new text goes at the end |
| `"x"` | Create text | Fails if the file already exists |

Text examples also specify `encoding="utf-8"`. Binary formats such as pickle use `"rb"` and `"wb"` instead.


## Loading a Text File

Open the supplied file in read mode. The `with` block closes it automatically after `read()` returns the contents as one string.


```python
with open(input_path, mode="r", encoding="utf-8") as source_file:
    loaded_text = source_file.read()

print("Loaded text:")
print(loaded_text)
```

The three lines came from the committed `study_plan.txt`; they were not generated immediately before this load example.


## Saving a Text File

Add one simple line in memory, then open the separate output path in write mode.


```python
text_to_save = loaded_text + "Thursday: Review\n"

with open(output_path, mode="w", encoding="utf-8") as saved_file:
    character_count = saved_file.write(text_to_save)

print("Saved file:", os.path.basename(output_path))
print("Characters written:", character_count)
```

Mode `"w"` creates the output or replaces its previous contents, making repeated Run All deterministic.


## Reopening the Saved File

A successful write call is not the final check. Reopen the new file and compare its contents with the string that was saved.


```python
with open(output_path, mode="r", encoding="utf-8") as saved_file:
    reloaded_text = saved_file.read()

print("Saved content matches:", reloaded_text == text_to_save)
print(reloaded_text)
```

The reopened content matches exactly, confirming both the output path and UTF-8 text write.


## Appending Text

Append mode adds content after the current end of a file without replacing earlier text.


```python
with open(output_path, mode="a", encoding="utf-8") as saved_file:
    saved_file.write("Friday: File I/O practice\n")

with open(output_path, mode="r", encoding="utf-8") as saved_file:
    appended_text = saved_file.read()

print(appended_text)
```

The Friday line appears at the end. Earlier `"w"` mode resets the fixed output file on every full run, so the appended line does not accumulate.


## Handling a Missing Text File

Reading a missing file raises `FileNotFoundError`. Catch the specific exception when the program can report or recover from it.


```python
missing_path = os.path.join(output_directory, "missing_notes.txt")

try:
    with open(missing_path, mode="r", encoding="utf-8") as missing_file:
        missing_file.read()
except FileNotFoundError:
    print("Text file not found:", os.path.basename(missing_path))
```
