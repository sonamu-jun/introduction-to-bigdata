# CSV Files

*Learn to read, write, and verify CSV rows with the `csv` module.*


## Before You Run

In **Add files**, select `scores.csv` and enter `assets/03_csv_files` for **Save folder**. The file will be available at `/home/pyodide/assets/03_csv_files/scores.csv`.


## Locate and Read

Open CSV files with `newline=""`, then let `csv.reader()` parse each row.


```python
import csv
import os


jupyterlite_base_directory = "/home/pyodide"
chapter_name = "01-3_Python_Basics_III"

if os.path.isdir(jupyterlite_base_directory):
    os.chdir(jupyterlite_base_directory)
elif os.path.basename(os.getcwd()) != chapter_name:
    os.chdir(chapter_name)

base_directory = os.getcwd()

input_directory = os.path.join(base_directory, "assets", "03_csv_files")
output_directory = os.path.join(base_directory, "outputs", "03_csv_files")
input_path = os.path.join(input_directory, "scores.csv")
output_path = os.path.join(output_directory, "scores_copy.csv")

os.makedirs(output_directory, exist_ok=True)

with open(input_path, mode="r", encoding="utf-8", newline="") as source_file:
    loaded_rows = list(csv.reader(source_file))

for row in loaded_rows:
    print(row)
```

Each CSV row becomes a list of strings, including a quoted value that contains a comma.


## Write and Verify

Use `csv.writer()` to save the rows, then read the output again.


```python
with open(output_path, mode="w", encoding="utf-8", newline="") as saved_file:
    csv.writer(saved_file).writerows(loaded_rows)

with open(output_path, mode="r", encoding="utf-8", newline="") as saved_file:
    reloaded_rows = list(csv.reader(saved_file))

print("Rows:", len(reloaded_rows))
print("Saved correctly:", reloaded_rows == loaded_rows)
```

## Takeaway

Combine `open()` with `csv.reader()` and `csv.writer()` instead of splitting CSV text by hand.
