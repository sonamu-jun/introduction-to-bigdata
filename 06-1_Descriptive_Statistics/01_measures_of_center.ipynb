# Measures of Center

*Describe a dataset with its mean, median, and modes, and explain their sensitivity to extreme values.*


## Describing the Observed Data

Descriptive statistics summarize the observations available to us. They can
describe a center, spread, shape, or relationship without showing every record.
They do not, by themselves, establish that a sample represents a population or
that one variable causes another.

A population is the complete set of units under discussion; a sample contains
only some of those units. The scope of a conclusion depends on which units
were observed. Here the prepared values are study-session lengths in minutes.


## The Arithmetic Mean

For $n$ observations, the mean is

$$\bar{x}=\frac{1}{n}\sum_{i=1}^{n}x_i.$$

The sum uses every value, so a large observation can move the mean
substantially. `np.mean()` computes this quantity for the array.


```python
import numpy as np
import pandas as pd

minutes = np.array([20, 25, 30, 30, 45], dtype=float)
print("Study minutes:", minutes)
print("Sum:", minutes.sum())
print("Count:", minutes.size)
print("Mean:", minutes.mean())
```

The five sessions total 150 minutes and have mean 30 minutes.
The mean is expressed in the same unit as the observations.


## The Median Uses Ordered Positions

After sorting the observations, the median is the middle value for an odd
count, or the mean of the two middle values for an even count. An even-count
median need not occur in the dataset. `np.median()` performs the sorting and
middle-value calculation internally.


```python
odd_values = np.array([1, 3, 5, 7, 9])
even_values = np.array([1, 3, 5, 7, 9, 10])
print("Odd-count median:", np.median(odd_values))
print("Even-count median:", np.median(even_values))
print("Study-session median:", np.median(minutes))
```

The two lecture examples have medians 5 and 6. For the study sessions,
the sorted middle observation is 30 minutes.


## The Mode Is the Most Frequent Value

A dataset can have more than one mode. `.mode()` returns every value tied for
the greatest frequency. A frequency table shows the evidence behind that result.
For nearly continuous measurements, modes can depend heavily on rounding.


```python
session_series = pd.Series(minutes)
print("Study-minute frequencies:")
print(session_series.value_counts().sort_index())
print("Study-minute modes:", session_series.mode().tolist())
print("Tied modes:", pd.Series([1, 3, 3, 5, 5, 7]).mode().tolist())
```

Thirty minutes occurs twice and is the only mode of the study sessions.
The second example has two equally frequent modes, 3 and 5.


## Compare the Same Data with One Changed Extreme

Replace the longest session, 45, with 180 while retaining the other four values.
The median's resistance comes from its dependence on ordering. It is resistant
to the magnitude of an extreme value, but it is not immune to every change
in the number or placement of observations.


```python
extended_minutes = minutes.copy()
extended_minutes[-1] = 180
comparison = pd.DataFrame(
    {
        "original": [minutes.mean(), np.median(minutes)],
        "with_long_session": [extended_minutes.mean(), np.median(extended_minutes)],
    },
    index=["mean_minutes", "median_minutes"],
)
print(comparison)
```

The mean increases from 30 to 57 minutes, while the median remains 30.
The long session may be genuine; these summaries describe its influence rather
than establish an entry error.
