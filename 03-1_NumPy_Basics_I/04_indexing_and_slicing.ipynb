# Indexing, Slicing, and Copies

*Select values along array axes and distinguish a shared slice from an independent copy.*


## One-Dimensional Selection

Indexing starts at zero; negative indices count from the end.
A slice uses `[start:stop:step]` and excludes the stop position.
Omitting bounds selects the available range in the step's direction.


```python
import numpy as np

measurements = np.array([10, 20, 30, 40, 50])

print("First value:", measurements[0])
print("Third value:", measurements[2])
print("Last value:", measurements[-1])
print("Slice [1:4]:", measurements[1:4])
print("Every second value:", measurements[::2])
print("Reverse order:", measurements[::-1])
print("Backward slice [3:1:-1]:", measurements[3:1:-1])
```

`[1:4]` selects `[20, 30, 40]`. A step of 2 selects `[10, 30, 50]`,
and a step of -1 reverses the direction. `[3:1:-1]` visits positions
3 and 2, producing `[40, 30]`.


## Rows and Columns

In a two-dimensional array, use `array[row, column]`. A colon selects
a range on that axis. An integer index removes that axis from the
result, while a slice retains it.

**Question.** Will `table[:, 1]` and `table[:, 1:2]` have the same shape?


```python
table = np.array([[10, 20, 30],
                  [40, 50, 60],
                  [70, 80, 90]])

print("Row 1, column 2:", table[1, 2])
print("Top-right block:")
print(table[0:2, 1:3])
print("Second row:", table[1, :])
print("Second column:", table[:, 1])
print("Column with integer index:", table[:, 1].shape)
print("Column with slice:", table[:, 1:2].shape)
```

The block is `[[20, 30], [50, 60]]`. The column values are 20, 50, and
80 in both selections, but the shapes differ: `(3,)` for the integer
index and `(3, 1)` for the slice. Retaining an axis matters in later calculations.


## A Basic Slice Shares Data

Basic array slicing returns a **view** that refers to the original
data. Assigning through that view also updates the original array.
This differs from slicing a Python list.


```python
middle_values = measurements[1:4]
middle_values[0] = 200

print("Updated slice:", middle_values)
print("Original array after slice assignment:", measurements)
```

`measurements` becomes `[10, 200, 30, 40, 50]`: the slice and original
array refer to the same storage for the selected positions.


## An Independent Copy

Use `.copy()` when a selected block must be edited independently.
Copying a slice preserves its values in separate storage.


```python
independent_values = measurements[1:4].copy()
independent_values[0] = -1

print("Edited copy:", independent_values)
print("Original after editing the copy:", measurements)
```

Only the copy changes to `[-1, 30, 40]`. The original keeps 200 at
position 1. Re-running the notebook from the beginning recreates the input data.
