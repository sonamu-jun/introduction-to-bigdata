# Structuring

<span style="color: #777777; font-style: italic;">구조화</span>


## Rows and Fields

<span style="color: #777777; font-style: italic;">행과 항목</span>

- A `schema` defines the `fields` (columns) and their data types.  
  <span style="color: #777777; font-style: italic;">`스키마`(schema)는 `필드`(field, 열)와 자료형(data type)을 정합니다.</span>

- Here one row is one student’s score in one subject.  
  <span style="color: #777777; font-style: italic;">여기서는 한 행이 한 학생의 한 과목 점수입니다.</span>

- Names and subjects are text; scores are integers.  
  <span style="color: #777777; font-style: italic;">이름과 과목은 문자열이고 점수는 정수입니다.</span>


## Parse and Combine Records

<span style="color: #777777; font-style: italic;">기록 파싱과 결합</span>

- `csv.reader()` reads comma-separated fields, including commas inside quoted text.  
  <span style="color: #777777; font-style: italic;">`csv.reader()`는 쉼표로 항목을 나누되, 따옴표 안의 쉼표는 같은 항목으로 읽습니다.</span>

- `int()` turns score text into an integer; `parse_scores()` collects the records in a table.  
  <span style="color: #777777; font-style: italic;">`int()`는 점수 문자열을 정수로 바꾸고, `parse_scores()`는 기록을 표로 모읍니다.</span>

- `pd.concat()` stacks the tables; `ignore_index=True` gives the rows new numbers.  
  <span style="color: #777777; font-style: italic;">`pd.concat()`은 표를 아래로 이어 붙이고, `ignore_index=True`는 행 번호를 새로 붙입니다.</span>


- Parsing separates a text record into fields; type conversion makes the score a number.  
  <span style="color: #777777; font-style: italic;">`파싱`(parsing)은 텍스트를 항목으로 나누고, 자료형 변환은 점수를 숫자로 바꿉니다.</span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/05-2_Data_Wrangling/assets/02_structuring/concept.webp" width="467" alt="Parsing separates a text record into fields; type conversion makes the score a number.">


```python
import csv
import pandas as pd

first_lines = [
    "Jiyoung,100,Mathematics",
    "Hojun,90,English",
    "Hana,85,English",
    "Kihoon,95,Mathematics",
]
second_lines = [
    "Jiyoung,78,English",
    "Hojun,95,Mathematics",
    "Hana,77,Mathematics",
    "Kihoon,86,English",
]

def parse_scores(lines):
    records = []
    for name, score, subject in csv.reader(lines):
        records.append({"name": name, "subject": subject, "score": int(score)})
    return pd.DataFrame(records)

first_scores = parse_scores(first_lines)
second_scores = parse_scores(second_lines)
scores_df = pd.concat([first_scores, second_scores], ignore_index=True)

print(scores_df)
```

- The table has eight scores.  
  <span style="color: #777777; font-style: italic;">표에는 점수 8개가 있습니다.</span>

- Each student–subject pair occurs once.  
  <span style="color: #777777; font-style: italic;">학생–과목 쌍은 각각 한 번씩 나옵니다.</span>


## Long and Wide Forms

<span style="color: #777777; font-style: italic;">긴 형식과 넓은 형식</span>

- In `long form`, one row holds one student–subject score.  
  <span style="color: #777777; font-style: italic;">`긴 형식`(long form)은 한 행에 학생–과목 점수 하나를 담습니다.</span>

- `pivot()` makes a `wide form`: names become rows and subjects become columns.  
  <span style="color: #777777; font-style: italic;">`pivot()`은 이름을 행으로, 과목을 열로 배치해 `넓은 형식`(wide form)을 만듭니다.</span>

  - Each name–subject pair must have just one score.  
    <span style="color: #777777; font-style: italic;">이름–과목 쌍마다 점수가 하나씩 있어야 합니다.</span>

- `reset_index()` brings names back as a column; `melt()` puts the subject columns back into rows.  
  <span style="color: #777777; font-style: italic;">`reset_index()`는 이름을 다시 열로 꺼내고, `melt()`는 과목 열을 다시 행으로 펼칩니다.</span>


```python
wide_scores = scores_df.pivot(index="name", columns="subject", values="score")

print(wide_scores)

long_scores = wide_scores.reset_index().melt(
    id_vars="name", var_name="subject", value_name="score"
)
print(long_scores)
```

- The wide table has one row per student and one column per subject.  
  <span style="color: #777777; font-style: italic;">넓은 표는 학생별 행과 과목별 열을 가집니다.</span>

- `melt()` returns all eight scores to `long form`.  
  <span style="color: #777777; font-style: italic;">`melt()`는 점수 8개를 `긴 형식`(long form)으로 되돌립니다.</span>


## Summary

<span style="color: #777777; font-style: italic;">핵심 정리</span>

- Reshaping changes the layout while keeping each student–subject score.  
  <span style="color: #777777; font-style: italic;">형식 변환은 학생–과목별 점수를 유지하면서 배치를 바꿉니다.</span>
