## Method · Working Directory and Path Representations

`Path` comes from Python's standard library, and `Path.cwd()` gives the current working directory. Jupyter, Colab, and Playground may start in different locations, so the printed path provides orientation without requiring a memorized absolute path.

```text
Path.cwd()
└── output
    └── path_demo
        └── observations.txt   ← planned address
```

The tree describes how path parts nest. It does not claim that any folder or file already exists.

````python
from pathlib import Path

working_dir = Path.cwd()
relative_path = Path("output") / "path_demo" / "observations.txt"
resolved_path = (working_dir / relative_path).resolve()

print("Working folder:", working_dir.name)
print("Relative path:", relative_path)
print("Resolved absolute path:", resolved_path)
````

The `/` operator joins path parts when they are `Path` objects. Both variables identify the same planned file: the relative form is portable, while the resolved form shows its current location. Neither form creates a folder or file.

## Implementation · Nonmutating Path Inspection

The `name`, `stem`, and `suffix` properties inspect text in the path. Checking existence before and after assembling the same path confirms that path construction alone has no filesystem effect.

````python
exists_before = resolved_path.exists()
same_path = working_dir / "output" / "path_demo" / "observations.txt"
exists_after = same_path.exists()

print("Name:", same_path.name)
print("Stem:", same_path.stem)
print("Suffix:", same_path.suffix)
print("Exists before joining again:", exists_before)
print("Exists after joining again:", exists_after)
print("Path construction changed the filesystem:", exists_before != exists_after)
````

### Interpretation · Nonmutating Path Inspection

On a clean first run, the planned file does not exist. If that exact target was created during an earlier experiment in this working folder, both checks are true. In either case, joining path parts changes no file.
