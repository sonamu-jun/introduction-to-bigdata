# Sets

*Use a set when uniqueness and membership matter more than order or position.*

A set behaves like a collection of distinct labels. Repeated inputs collapse into one value, and items have no index. Sets are mutable, but their printed order should not be treated as meaningful.

## Creation, Uniqueness, and Membership

Curly braces create a non-empty set. `set()` creates an empty set or removes duplicates from another collection. Empty `{}` creates a dictionary, not a set.

```python
island_entries = ["Biscoe", "Dream", "Biscoe", "Torgersen"]
observed_islands = set(island_entries)
empty_set = set()

print("Original entries:", island_entries)
print("Unique islands:", sorted(observed_islands))
print("Unique count:", len(observed_islands))
print("Contains Dream:", "Dream" in observed_islands)
print("Empty set:", empty_set)
```

The four source entries become three set items because `"Biscoe"` is repeated. `sorted()` is used only to make the displayed order predictable; it returns a list and does not add order to the set.

## Adding One or Several Items

`add(value)` adds one item. `update(iterable)` adds every item from another collection. Adding an existing value leaves the set unchanged.

```python
observed_islands.add("Dream")
observed_islands.add("New Island")
observed_islands.update(["Far Island", "Biscoe"])

print("Expanded islands:", sorted(observed_islands))
```

Only the new labels increase the set's size. Sets enforce uniqueness automatically, so no separate duplicate-removal step is needed.

## Removing Items

`remove(value)` deletes a known item and raises `KeyError` if it is absent. `discard(value)` also deletes an item but quietly does nothing when the value is absent.

```python
observed_islands.remove("Far Island")
observed_islands.discard("Unknown Island")

print("After removal:", sorted(observed_islands))
```

Use `remove()` when absence indicates a mistake. Use `discard()` when an already-absent value is an acceptable state.

## Combining and Comparing Sets

`intersection()` keeps shared values, `union()` keeps values from either set, `difference()` keeps values found only on the left, and `symmetric_difference()` keeps values found in exactly one set.

```python
survey_a = {"Biscoe", "Dream", "Torgersen"}
survey_b = {"Biscoe", "Dream", "New Island"}

print("In both:", sorted(survey_a.intersection(survey_b)))
print("In either:", sorted(survey_a.union(survey_b)))
print("Only in survey A:", sorted(survey_a.difference(survey_b)))
print("In exactly one survey:", sorted(survey_a.symmetric_difference(survey_b)))
```

The operator forms `&`, `|`, `-`, and `^` perform the same four operations. The named methods make the intended comparison easier to read while first learning sets.

## Testing Set Relationships

`issubset()` tests whether every item appears in another set. `issuperset()` tests the reverse relationship, and `isdisjoint()` tests whether two sets share no items.

```python
core_islands = {"Biscoe", "Dream"}
remote_islands = {"Far Island", "Outer Island"}

print("Core is a subset of survey A:", core_islands.issubset(survey_a))
print("Survey A is a superset of core:", survey_a.issuperset(core_islands))
print("Survey A and remote are disjoint:", survey_a.isdisjoint(remote_islands))
```

## Set Method Coverage

The public set methods are grouped by purpose below. The next examples demonstrate the in-place update and cleanup methods not shown earlier.

| Purpose | Methods |
| --- | --- |
| Add items | `add()`, `update()` |
| Remove items | `remove()`, `discard()`, `pop()`, `clear()` |
| Copy | `copy()` |
| Return a compared set | `union()`, `intersection()`, `difference()`, `symmetric_difference()` |
| Change a set by comparison | `intersection_update()`, `difference_update()`, `symmetric_difference_update()` |
| Test relationships | `issubset()`, `issuperset()`, `isdisjoint()` |

### In-Place Set Comparisons

Methods ending in `_update()` change the set on which they are called. Copies keep the source survey sets available for comparison.

```python
shared_islands = survey_a.copy()
shared_islands.intersection_update(survey_b)

only_a_islands = survey_a.copy()
only_a_islands.difference_update(survey_b)

different_islands = survey_a.copy()
different_islands.symmetric_difference_update(survey_b)

print("Shared in place:", sorted(shared_islands))
print("Only A in place:", sorted(only_a_islands))
print("Different in place:", sorted(different_islands))
```

The three copies are replaced by their comparison results. `survey_a` and `survey_b` themselves remain unchanged.

### Removing an Arbitrary Item or Every Item

`pop()` removes and returns an arbitrary set item because sets have no last position. `clear()` removes all items. A disposable copy prevents source-data loss.

```python
temporary_islands = observed_islands.copy()
size_before_pop = len(temporary_islands)
temporary_islands.pop()

print("Size before pop():", size_before_pop)
print("Size after pop():", len(temporary_islands))

temporary_islands.clear()
print("After clear():", temporary_islands)
```

`pop()` reduces the size by one, but code should not predict which item was removed. `clear()` then leaves the temporary set empty.

## Use Cases

Sets are commonly used to remove duplicates, test permissions or categories, and compare membership across groups.

### Remove Duplicate Identifiers

A visitor may appear several times in raw logs, while a unique-visitor count should include each identifier once.

```python
visitor_log = ["u01", "u02", "u01", "u03", "u02", "u04"]
unique_visitors = set(visitor_log)

print("Raw visit records:", len(visitor_log))
print("Unique visitors:", len(unique_visitors))
print("Unique IDs:", sorted(unique_visitors))
```

Converting the log to a set removes repeated identifiers automatically. The original list still preserves every visit event.

### Check Required Permissions

The required permissions form a set that should be a subset of the permissions held by a user.

```python
required_permissions = {"read", "comment"}
user_permissions = {"read", "comment", "download"}

print("Has every required permission:", required_permissions.issubset(user_permissions))
print("Extra permissions:", sorted(user_permissions.difference(required_permissions)))
```

The subset test checks all required permissions at once. The difference identifies permissions the user has beyond the requirement.

### Compare Two Enrollment Groups

Set operations reveal shared students and students found in only one course.

```python
python_students = {"S01", "S02", "S03", "S05"}
statistics_students = {"S02", "S03", "S04"}

print("Taking both:", sorted(python_students.intersection(statistics_students)))
print("Python only:", sorted(python_students.difference(statistics_students)))
print("At least one course:", sorted(python_students.union(statistics_students)))
```

Intersection, difference, and union answer three group-membership questions directly without manually checking every student.

### Category Validation

The difference between observed and allowed categories reveals unexpected labels.

```python
allowed_species = {"Adelie", "Chinstrap", "Gentoo"}
observed_species = {"Adelie", "Gentoo", "Unknown"}
unexpected_species = observed_species.difference(allowed_species)

print("Unexpected categories:", sorted(unexpected_species))
```

Only `Unknown` occurs in the observed set without appearing in the allowed set, so it becomes the validation result.

### Tracking Identifiers Already Seen

A set can record identifiers that have already been processed. Re-adding the same identifier does not create a duplicate.

```python
processed_ids = set()
processed_ids.add("row-001")
processed_ids.add("row-001")
processed_ids.update(["row-002", "row-003"])

print("Processed IDs:", sorted(processed_ids))
print("Processed count:", len(processed_ids))
```

The duplicate addition leaves the size unchanged. This pattern supports quick membership checks when a later loop processes many records.

Choose a set for unique categories, fast membership questions, and overlap comparisons. Choose a list instead when duplicates, insertion order, or position-based access must be preserved.
