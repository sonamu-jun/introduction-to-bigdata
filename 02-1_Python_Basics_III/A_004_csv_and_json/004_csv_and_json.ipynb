# 021-004. CSV and JSON

*Read five small records from CSV, save the same records as CSV and JSON, and compare the restored structures.*

The conceptual foundation is developed in Text and Structured File Round Trips.

## Method · CSV Sample Input

`DictReader` uses the header row as dictionary keys. The loop displays the transformation and selects only three fields.

The repository copy is `input/penguins.csv`. In JSPCV Playground, **Add files** stores it as `uploads/penguins.csv`; a Colab sidebar upload appears as `penguins.csv`. The code checks these locations in that order.

```python
import csv
from pathlib import Path

lesson_dir = Path.cwd()
local_path = Path("input/penguins.csv")
playground_path = Path("uploads/penguins.csv")
colab_path = Path("penguins.csv")

if local_path.exists():
    input_path = local_path
elif playground_path.exists():
    input_path = playground_path
else:
    input_path = colab_path

sample = []
with input_path.open(mode="r", encoding="utf-8", newline="") as file:
    reader = csv.DictReader(file)
    for row in reader:
        sample.append({
            "species": row["species"],
            "island": row["island"],
            "body_mass_g": row["body_mass_g"],
        })
        if len(sample) == 5:
            break

print("Rows read:", len(sample))
for record in sample:
    print(record)
```

## Implementation · CSV Round Trip

CSV stores one flat record per row. Its values are read back as text unless the program converts them.

```python
output_dir = lesson_dir / "output"
output_dir.mkdir(exist_ok=True)
csv_path = output_dir / "penguins_sample.csv"
fieldnames = ["species", "island", "body_mass_g"]

with csv_path.open(mode="w", encoding="utf-8", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames, lineterminator="\n")
    writer.writeheader()
    writer.writerows(sample)

with csv_path.open(mode="r", encoding="utf-8", newline="") as file:
    restored_csv = list(csv.DictReader(file))

print("Saved:", csv_path.relative_to(lesson_dir))
print("First restored CSV row:", restored_csv[0])
```

## Implementation · JSON Round Trip

JSON naturally stores a dictionary containing both metadata and a list of records.

```python
import json

json_path = output_dir / "penguins_sample.json"
payload = {
    "record_count": len(sample),
    "records": sample,
}

with json_path.open(mode="w", encoding="utf-8") as file:
    json.dump(payload, file, ensure_ascii=False, indent=2)

with json_path.open(mode="r", encoding="utf-8") as file:
    restored_json = json.load(file)

print("Saved:", json_path.relative_to(lesson_dir))
print("JSON keys:", list(restored_json))
print("First restored JSON row:", restored_json["records"][0])
```

### Interpretation · JSON Round Trip

CSV is a flat table with a header. JSON can nest lists and dictionaries. Both restored examples contain standard Python strings, lists, and dictionaries.
