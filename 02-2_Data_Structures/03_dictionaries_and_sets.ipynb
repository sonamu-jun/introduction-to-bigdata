## Keys and Membership

A **dictionary** associates each unique key with one current value. `lookup[key]` retrieves by key; assigning that key again replaces its value. Values may repeat. Iteration preserves insertion order, but lookup is not positional.

A **set** keeps unique members without positional order or occurrence counts. `in` checks set membership or dictionary keys. `{}` creates an empty dictionary; `set()` creates an empty set.

For sets, `a | b` is the union, `a & b` is the intersection, and `a - b` is the first-only difference.


**Think before running or viewing the saved output.** Will updating S01 create another dictionary key? Which ID belongs to both groups, and can either set tell how often an ID appeared?

Compare shared, combined, and exclusive membership regions.


```python
scores_by_id = {"S01": 70, "S02": 80}
scores_by_id["S01"] = 75
print("Updated scores:", scores_by_id)
print("Number of keys:", len(scores_by_id))

first_group = {"S01", "S02"}
second_group = {"S02", "S03"}
print("Union:", sorted(first_group | second_group))
print("Intersection:", sorted(first_group & second_group))
print("First group only:", sorted(first_group - second_group))
```

**Check:** The dictionary still has two keys; S01 now stores `75`. S02 belongs to both groups. Sets cannot recover occurrence counts. Sorting the printed members makes the output repeatable without giving the sets positions.


A dictionary keeps a value for each key; a set keeps membership alone. Use strings, numbers, or tuples of suitable items as keys; lists cannot be dictionary keys or set members.
