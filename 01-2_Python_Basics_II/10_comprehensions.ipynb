# Comprehensions

*Compress a simple “transform each value” loop only after the expanded loop is understood.*

## Expanded Transformation Loop

A transformation loop starts with an empty collection, processes each input item, and appends one result at a time. A comprehension expresses this same pattern as one collection expression.


## Compact Transformations

A list comprehension follows the structure `[expression for item in iterable]`. The expression produces one result for each item while preserving iteration order.


## Filtered Comprehensions

A final condition gives the structure `[expression for item in iterable if condition]`. The condition is checked before an item contributes a result.


## Dictionary Comprehensions

A dictionary comprehension uses `{key_expression: value_expression for item in iterable}`. Each iteration produces one key–value pair, and a repeated key replaces its earlier value.


Use a comprehension for one short transformation and at most one clear filter. Keep an ordinary loop when several steps, branches, or side effects would make the compact form harder to read.
