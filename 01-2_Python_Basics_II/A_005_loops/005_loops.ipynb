*Repeat one operation across a sequence and inspect the loop body and state after each iteration.*

The conceptual foundation is developed in Collections and Iteration.

## Method · Value Iteration

On each iteration, `score` receives the next list value. Indentation marks the repeated block.

```python
scores = [72, 85, 91, 68]

for score in scores:
    print("Observed score:", score)
```

## Implementation · Result Accumulation

Initialize an accumulator before the loop, update it once per value, and inspect it after the loop ends.

> **Before you run**
>
> Predict the final value of `total`.

```python
total = 0
for score in scores:
    total = total + score

print("Total:", total)
print("Mean:", total / len(scores))
```

## Implementation · Position Tracking and Paired Iteration

First, `enumerate()` supplies a position for one sequence. Then, in a separate loop, `zip()` pairs values that occupy the same position in two sequences. Each produces a two-value pair. Writing two names after `for` unpacks that pair from left to right: `row_number, name` and `name, score`.

```python
names = ["A", "B", "C", "D"]

for row_number, name in enumerate(names, start=1):
    print(row_number, name)

print()
for name, score in zip(names, scores):
    print(name, score)
```

## Implementation · Condition-Controlled Repetition

`range(1, 4)` produces 1, 2, and 3; its stop value is excluded. A `while` loop needs an update that eventually makes its condition false.

```python
attempt = 1
while attempt <= 3:
    print("Attempt:", attempt)
    attempt = attempt + 1
```

### Interpretation · Condition-Controlled Repetition

Keep the loop body short and name the changing value clearly. If there is already a built-in operation such as `sum(scores)`, prefer it in ordinary analysis and use the explicit loop when learning or implementing custom logic.
