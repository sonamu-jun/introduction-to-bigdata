# Classes

*A class defines a new kind of object by keeping related state and behavior together.*

## Class Definition

A class defines the attributes and methods shared by its instances. `__init__` runs when an instance is created, and `self` refers to that particular instance.


## Instances and Method Calls

A class is a definition, while an instance is one object created from it. A method reads or changes instance state through `self`.


## Independent Instance State

Each instance keeps its own attribute values. Changing an instance attribute does not change the corresponding attribute of another instance.


## Class Attributes, Instance Attributes, and `__str__`

A class attribute is shared as a default across instances. Instance attributes belong to one object. `__str__()` defines the readable text produced by `print(object)`.

Use a class when a domain concept needs both state and reusable behavior. At this stage, understanding instances, attributes, and methods is more important than inheritance or advanced class syntax.
