## Method · Integers and Floating-Point Numbers

An `int` stores a whole number. A `float` stores a number with a fractional part. `type()` lets us inspect which representation Python is using.

```python
student_count = 30
average_score = 82.5

print(student_count, type(student_count))
print(average_score, type(average_score))
```

## Implementation · Arithmetic Operators

The operators `+`, `-`, `*`, `/`, and `**` calculate addition, subtraction, multiplication, division, and powers. Prefix each output line with the operation name.

```python
first_value = 10
second_value = 3

print("Addition:", first_value + second_value)
print("Subtraction:", first_value - second_value)
print("Multiplication:", first_value * second_value)
print("Division:", first_value / second_value)
print("Power:", first_value ** second_value)
```

## Implementation · Quotient, Remainder, and Grouping

`//` returns the whole-number quotient for these positive integers, while `%` returns the remainder. Parentheses make the intended order of calculation explicit.

> **Before you run**
>
> If 10 observations are placed into groups of 3, predict the number of complete groups and the number left over.

```python
observation_count = 10
group_size = 3

print("Complete groups:", observation_count // group_size)
print("Left over:", observation_count % group_size)
print("Grouped calculation:", (2 + 3) * 4)
```

## Implementation · Numeric Conversion and Rounding

`int()` and `float()` create values of another numeric type when conversion is valid. Use `round()` to format displayed values; keep the unrounded value for later calculations.

```python
measured_value = 1.246

print("As an integer:", int(measured_value))
print("As a float:", float(student_count))
print("Rounded for display:", round(measured_value, 2))
print("Scientific notation:", 2.1e3)
```

### Interpretation · Numeric Conversion and Rounding

`/` produces a floating-point result, even when its inputs are integers. Converting with `int()` removes the fractional part; it is not the same operation as rounding.
