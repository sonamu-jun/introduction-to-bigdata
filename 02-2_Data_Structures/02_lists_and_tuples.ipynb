# Lists and Tuples

*Distinguish a sequence that can change from a fixed grouping of items.*

This briefly revisits [lists](../01-2_Python_Basics_II/01_lists.ipynb) and [tuples](../01-2_Python_Basics_II/03_tuples.ipynb) from Basics II. Both preserve order and repeated values.


## Sequence Positions and Mutability

| Property | List | Tuple |
| --- | --- | --- |
| Access | Index and slice | Index and slice |
| Repeated values | Retained | Retained |
| Replace, add, or remove positions | Allowed | Not allowed |

Recall `[0]` for the first position and `[1:3]` for positions 1 and 2. A slice excludes its stop position. List insertion and removal shift later positions.

`append()` grows a list. Tuple unpacking assigns position-ordered values to names; a one-item tuple needs a comma, as in `(30,)`. The handler below catches an expected `TypeError`.


**Think before running or viewing the saved output.** Which existing object can change below? After constructing the revised tuple, what will the original coordinate contain?

Compare editing existing positions with creating a new tuple.


```python
readings = [20, 35, 20]
readings.append(50)
print("List after append:", readings)

coordinate = (2, 4)
row_index, column_index = coordinate
print("Unpacked coordinate:", row_index, column_index)

try:
    coordinate[0] = 3
except TypeError:
    print("Tuple positions cannot be reassigned.")

revised_coordinate = (3, coordinate[1])
print("Original tuple:", coordinate)
print("New tuple:", revised_coordinate)
```

**Check:** The list grows to `[20, 35, 20, 50]`, keeping both readings of `20`. The tuple assignment is rejected. Constructing `(3, 4)` creates a new tuple and leaves `(2, 4)` unchanged.


Use a list for a sequence that must change and a tuple for an agreed fixed grouping. Both preserve positions and repetitions.
