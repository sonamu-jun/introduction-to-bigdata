# Lists

*Use a list when several values belong in an ordered, changeable sequence.*

A list is like a row of labeled storage slots. Each item has a position, the first position is index `0`, and the same value may appear more than once. A list is **mutable**, so its items and length can change after creation.

## Creation, Indexing, and Slicing

Square brackets create a list. An index selects one item, while a slice selects a new list containing a range of items. The slice stop index is excluded.

```python
scores = [72, 85, 91, 68, 88]

print("All scores:", scores)
print("First score:", scores[0])
print("Last score:", scores[-1])
print("Middle scores:", scores[1:4])
print("Number of scores:", len(scores))
```

Positive indexes count from the beginning and negative indexes count from the end. The slice `scores[1:4]` includes indexes 1, 2, and 3, while `len()` reports how many items the list contains.

## Adding Items

The methods `append()`, `insert()`, and `extend()` all make the existing list longer.

- `append(value)` adds one item at the end.
- `insert(index, value)` adds one item at a selected position.
- `extend(iterable)` adds every item from another sequence.

```python
subjects = ["Python", "Statistics"]

subjects.append("SQL")
subjects.insert(1, "Data Ethics")
subjects.extend(["NumPy", "pandas"])

print("Expanded subjects:", subjects)
print("Subject count:", len(subjects))
```

`append()` keeps its input as one item, whereas `extend()` takes the items out of another sequence and adds them separately. `insert(1, ...)` shifts the old items to the right instead of replacing them.

## Removing Items

`pop()` removes an item by position and returns it. With no index, it removes the last item. `remove()` finds and deletes the first matching value.

```python
removed_last = subjects.pop()
subjects.remove("Statistics")

print("After removal:", subjects)
print("Value returned by pop():", removed_last)
```

The returned value makes `pop()` useful when the removed item is still needed. `remove()` expects the value to exist; otherwise, Python raises `ValueError`.

## Finding and Counting Values

`count()` reports how often a value occurs, and `index()` reports the first matching position. The `in` operator answers a membership question with `True` or `False`.

```python
responses = ["yes", "no", "yes", "maybe", "yes"]

print("Number of yes responses:", responses.count("yes"))
print("First maybe position:", responses.index("maybe"))
print("Contains no:", "no" in responses)
```

Repeated values remain in a list, so `count("yes")` returns `3`. Like `remove()`, `index()` expects a match and raises `ValueError` if the value is absent.

## Copying and Reordering

`copy()` creates an independent list. `sort()` rearranges a list in ascending order, and `reverse()` reverses its current order. These methods change the list on which they are called.

```python
original_scores = [72, 85, 91, 70, 88]
ascending_scores = original_scores.copy()
ascending_scores.sort()
descending_scores = ascending_scores.copy()
descending_scores.reverse()

print("Original order:", original_scores)
print("Ascending copy:", ascending_scores)
print("Descending copy:", descending_scores)
```

The copies can be reordered without changing `original_scores`. By contrast, assigning `ascending_scores = original_scores` would give two names to the same mutable list.

## Nested Lists

A list may contain other lists. The first index selects an inner list, and the second index selects an item inside that list.

```python
records = [["A", 72], ["B", 85], ["C", 91]]

print("Second record:", records[1])
print("Second record's score:", records[1][1])
```

## Common List Expressions

Lists also support operators and slice assignment. These expressions are useful when a small program needs to combine, initialize, or replace several items.

### Concatenation and Repetition

`+` creates a new list by joining two lists. `*` creates a new list by repeating the source items.

```python
weekday_topics = ["Python", "Statistics"]
weekend_topics = ["Review"]
all_topics = weekday_topics + weekend_topics
empty_slots = ["pending"] * 3

print("Combined topics:", all_topics)
print("Repeated placeholders:", empty_slots)
print("Original weekdays:", weekday_topics)
```

Concatenation and repetition return new lists, so the source lists remain unchanged. Repetition is safest with immutable values such as strings and numbers.

### Slice Assignment and `del`

Slice assignment replaces a range and may change the list length. The `del` statement removes an item or slice without returning it.

```python
revised_topics = all_topics.copy()
revised_topics[1:2] = ["Data Ethics", "SQL"]
del revised_topics[-1]

print("Source topics:", all_topics)
print("Revised topics:", revised_topics)
```

One selected item is replaced by two items, so the list grows. `del` then removes the last item and does not provide a removed value; use `pop()` when that value is needed.

## List Method Coverage

The public list methods are grouped by their purpose below. The earlier examples demonstrate every method except `clear()`, which appears next.

| Purpose | Methods |
| --- | --- |
| Add items | `append()`, `extend()`, `insert()` |
| Remove items | `remove()`, `pop()`, `clear()` |
| Find values | `index()`, `count()` |
| Reorder items | `sort()`, `reverse()` |
| Make an independent shallow copy | `copy()` |

### Clearing a Disposable List

`clear()` removes every item in place. A copy is used here because clearing the source measurements would discard data that may still be needed.

```python
temporary_scores = original_scores.copy()
temporary_scores.clear()

print("Cleared temporary list:", temporary_scores)
print("Original scores remain:", original_scores)
```

`temporary_scores` becomes an empty list, while `original_scores` remains unchanged. Use `clear()` only when every item in that particular list should be removed.

## Use Cases

Lists are especially useful for ordered plans, sequential measurements, and small collections of rows that need to change.

### Maintain an Ordered Schedule

The order of activities matters, and a late addition can be inserted without rebuilding the whole schedule.

```python
daily_schedule = ["Lecture", "Lunch", "Lab"]
daily_schedule.insert(1, "Office hours")
daily_schedule.append("Review")

print("Updated schedule:", daily_schedule)
print("First activity:", daily_schedule[0])
```

A list preserves the planned order. `insert()` places office hours at a selected position, and `append()` adds review at the end.

### Store Sequential Measurements

Sensor readings may repeat, and their collection order carries time information. A slice can select the most recent observations.

```python
temperature_readings = [21.5, 21.7, 21.7, 22.0]
temperature_readings.append(22.3)

print("All readings:", temperature_readings)
print("Recent three:", temperature_readings[-3:])
print("Repeated 21.7 readings:", temperature_readings.count(21.7))
```

The list keeps both repeated `21.7` readings and their positions. A set would remove the duplicate and lose the sequence information.

### Represent a Small Editable Table

A nested list can hold rows when each row has the same positional structure and only simple access is needed.

```python
attendance_rows = [["Mon", 28], ["Wed", 27], ["Fri", 30]]
attendance_rows[1][1] = 29
attendance_rows.append(["Sat", 18])

print("Attendance rows:", attendance_rows)
print("Wednesday attendance:", attendance_rows[1][1])
```

The outer list preserves row order, and the inner lists keep day and attendance in fixed positions. For many labeled columns, dictionaries or a later DataFrame are clearer.

### A Last-In, First-Out Stack

A small stack keeps the most recently added item at the end. `append()` pushes an item, and `pop()` retrieves the latest one.

```python
page_stack = ["Home", "Courses"]
page_stack.append("Python Basics")
closed_page = page_stack.pop()

print("Closed page:", closed_page)
print("Previous pages:", page_stack)
```

`Python Basics` is removed first because it was added last. Using the end of a list for both operations is a common small-stack pattern.

A list is the natural choice when order, repeated values, and position-based access matter. Its mutability is convenient, but a copy is needed when later changes must not affect the source list.
