# Enriching

<span style="color: #777777;"><em>정보 보강</em></span>


## New Subject Scores

<span style="color: #777777;"><em>새로 확보한 과목 점수</em></span>

- `Enriching` adds newly obtained information to existing data.  
  <span style="color: #777777;"><em>`정보 보강`(enriching)은 기존 데이터에 새로 확보한 정보를 추가합니다.</em></span>

- We have English scores and obtain Math scores for the same students.  
  <span style="color: #777777;"><em>영어 점수가 있는 학생들의 수학 점수를 추가로 확보했습니다.</em></span>

- Student names are the `index`, the labels used to match rows.  
  <span style="color: #777777;"><em>학생 이름을 행을 맞추는 기준인 `인덱스`(index)로 사용합니다.</em></span>


```python
import pandas as pd

english = pd.DataFrame({"English": [85, 90]}, index=["Hana", "Hojun"])
math = pd.DataFrame({"Math": [95, 77]}, index=["Hojun", "Hana"])

print(english)
print(math)
```

## Add the New Scores

<span style="color: #777777;"><em>새 점수 결합</em></span>

- `pd.concat(..., axis=1)` adds columns and matches rows by their index.  
  <span style="color: #777777;"><em>`pd.concat(..., axis=1)`은 열을 추가하며 인덱스가 같은 행끼리 맞춥니다.</em></span>


- English and newly obtained Math scores are matched by student name.  
  <span style="color: #777777;"><em>기존 영어 점수와 새로 확보한 수학 점수를 학생 이름으로 맞춥니다.</em></span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/05-2_Data_Wrangling/assets/04_enriching/concept_concat.webp" width="560" alt="English and newly obtained Math scores are matched by student name.">


```python
enriched = pd.concat([english, math], axis=1)
print(enriched)
```

- Hana has 85 in English and 77 in Math; Hojun has 90 and 95.  
  <span style="color: #777777;"><em>Hana의 영어·수학 점수는 85·77이고, Hojun은 90·95입니다.</em></span>

- The names align correctly even though the two tables list students in different orders.  
  <span style="color: #777777;"><em>두 표의 학생 순서가 달라도 이름을 기준으로 올바르게 연결됩니다.</em></span>
