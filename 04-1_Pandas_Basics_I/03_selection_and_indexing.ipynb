# Selection and Indexing

*Select columns and distinguish label-based selection from position-based selection.*


## One Column or a Table of Columns

`table["score"]` returns a Series. Passing a list of names, such as
`table[["score"]]`, keeps a DataFrame even when the list contains one name.
The outer brackets perform selection; the inner brackets create the list.


```python
import pandas as pd

student_df = pd.DataFrame({
    "name": ["Minji", "Joon", "Hana", "Taeho"],
    "score": [88, 76, 93, 85],
    "department": ["Data", "AI", "Data", "Design"],
}, index=["st_1", "st_2", "st_3", "st_4"])

print("Original table:")
print(student_df)
print("Score Series:")
print(student_df["score"])
print("Series shape:", student_df["score"].shape)
print("One-column DataFrame shape:", student_df[["score"]].shape)
print("Selected columns:")
print(student_df[["name", "score"]])
```

The same four scores have shape `(4,)` as a Series and `(4, 1)` as a DataFrame. Selecting columns leaves the student row labels in place.


## Select Labels with `loc`

`.loc[row_labels, column_labels]` selects by names. A single row label
returns a Series; a list or slice retains a table. In this ordered index,
the label slice `"st_1":"st_3"` includes both endpoints.


```python
print("Student labeled st_2:")
print(student_df.loc["st_2"])
print("Labels st_1 through st_3:")
print(student_df.loc["st_1":"st_3", ["name", "score"]])
```

The label slice contains Minji, Joon, and Hana: three rows. The displayed columns match the requested names and scores.


## Select Positions with `iloc`

`.iloc[row_positions, column_positions]` uses zero-based positions.
Its slices follow Python rules and exclude the stop position.
Labels and positions can identify the same row, but express different requests.


```python
print("Student at position 1:")
print(student_df.iloc[1])
print("Rows 0:2 and columns 0:2:")
print(student_df.iloc[0:2, 0:2])
print("Scalar score for st_3:", student_df.loc["st_3", "score"])
```

The positional block contains Minji and Joon: `0:2` selects two rows. Hana's score is 93 at label `st_3`.
