# Series and DataFrame

*Create labeled one-dimensional and two-dimensional data and recognize label alignment.*


## A Series Has Values and Labels

`pd.Series(values, index=labels, name=...)` creates one labeled sequence.
The index identifies entries; the name describes the whole Series.
`.loc[label]` reads the value attached to a label.


```python
import pandas as pd

temperatures = pd.Series(
    [21, 23, 20, 24],
    index=["Mon", "Tue", "Wed", "Thu"],
    name="temperature",
)

print("Temperatures (degrees Celsius):")
print(temperatures)
print("Tuesday:", temperatures.loc["Tue"])
print("Mean temperature:", temperatures.mean())
print("Index labels:", temperatures.index.tolist())
```

Tuesday is 23 degrees Celsius, and the four-day mean is 22.0. Day names identify the readings without requiring a numeric position.


## Arithmetic Matches Labels

Two Series are added by matching index labels, even when their stored
orders differ. This differs from the positional arithmetic of NumPy arrays.
Here each temperature has a matching adjustment.


```python
adjustments = pd.Series([1, 0, -1, 2], index=["Thu", "Wed", "Tue", "Mon"])
corrected = temperatures + adjustments

print("Adjusted temperatures in weekday order:")
print(corrected.reindex(temperatures.index))
```

Monday receives 2 and becomes 23. Tuesday, Wednesday, and Thursday become 22, 20, and 25. `.reindex()` displays the result in the original weekday order; it does not change how addition matches labels.


## A DataFrame Holds Several Columns

Pass a dictionary of equal-length lists to `pd.DataFrame()`. Dictionary
keys become column names. A default index numbers the rows from zero.
A row combines related values, while each column describes one field.


```python
student_df = pd.DataFrame({
    "name": ["Minji", "Joon", "Hana", "Taeho"],
    "age": [20, 21, 22, 20],
    "major": ["Data Science", "AI", "Economics", "Design"],
})

print("Student table:")
print(student_df)
print("Shape:", student_df.shape)
print("Columns:", student_df.columns.tolist())
print("Column types:")
print(student_df.dtypes)
```

The shape `(4, 3)` means four students and three fields. Age is numeric; names and majors are text. pandas 3 uses a string dtype for these inferred text columns.


## Inspect Before Calculating

`.head(n)` and `.tail(n)` preview the first and last rows without changing
the table. `.info()` prints column types and non-missing counts; it already
prints a report, so call it directly rather than wrapping it in `print()`.


```python
print("First two students:")
print(student_df.head(2))
print("Last student:")
print(student_df.tail(1))
print("Table information:")
student_df.info()
```

The preview starts with Minji and Joon and ends with Taeho. All three columns have four non-missing values.
