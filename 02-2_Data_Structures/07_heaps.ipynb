# Heaps

*Maintain a minimum heap and retrieve the smallest values through repeated removal.*

A **binary heap** is a complete binary tree: levels are full except
possibly the last, which fills from left to right. In a **min-heap**, each
parent is no larger than its children, placing the smallest value at the root.
A max-heap reverses this comparison.


## A Tree Stored in a List

Python's `heapq` maintains a min-heap inside a list. The root is at index
`0`; children of index `i` occupy `2 * i + 1` and `2 * i + 2`, when present.

```text
           1              List: [1, 3, 5, 78, 21, 45]
         /   \
        3     5
       / \   /
      78 21 45
```

This parent-child rule does not sort the whole list. In the diagram,
`78` precedes `21` in storage. `heapify()` rearranges an existing list
in place; `heap[0]` reads the minimum without removing it.


````python
import heapq

heap = [21, 1, 45, 78, 3, 5]
heapq.heapify(heap)

print("Heap storage:", heap)
print("Minimum:", heap[0])
print("Already sorted:", heap == sorted(heap))
````

The minimum is `1`, and the list is not sorted. The tree diagram shows
the same parent-child arrangement as the printed storage.


## Push, Pop, and Replace

`heappush()` inserts a value; `heappop()` removes and returns the minimum.
`heapreplace()` removes the current minimum and inserts a replacement
while keeping the length fixed. Both removal operations require a
nonempty heap. `heapify()` takes `O(n)` time; push and pop take `O(log n)`
as they repair a path through the complete tree.

[Python 3.12 heapq documentation](https://docs.python.org/3.12/library/heapq.html)


````python
heapq.heappush(heap, 8)
print("After pushing 8:", heap)

removed_minimum = heapq.heappop(heap)
print("Popped minimum:", removed_minimum)
print("After popping:", heap)

replaced_minimum = heapq.heapreplace(heap, 6)
print("Replaced minimum:", replaced_minimum)
print("After replacing with 6:", heap)
print("Current minimum:", heap[0])
````

Popping removes `1`. Replacing then removes `3` and adds `6`, leaving
`5` at the root. Only the heap rule is maintained; the remaining list
still need not be fully sorted.


## Repeated Minimum Extraction

A pop exposes the next minimum. Popping `k` times therefore retrieves
the first `k` values in ascending order. The operations remove those
values from the heap, so a copy preserves the original observations.


**Question.** For `[50, 20, 10, 80, 60, 30]`, what is the third smallest value, and how many values remain after three pops?


````python
numbers = [50, 20, 10, 80, 60, 30]
ranking_heap = numbers.copy()
heapq.heapify(ranking_heap)

k = 3
smallest_values = [heapq.heappop(ranking_heap) for _ in range(k)]

print("Extracted values:", smallest_values)
print("Third smallest:", smallest_values[-1])
print("Remaining count:", len(ranking_heap))
print("Original values:", numbers)
````

The extracted values are `[10, 20, 30]`, so the third smallest is `30`.
Three values remain in the heap, and the original list is unchanged.
Repeated extraction orders the removed values even though heap storage
itself is only partially ordered.
