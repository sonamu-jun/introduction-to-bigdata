# Binary Files with Pickle

*Learn to save and reload a trusted Python object as a binary pickle file.*


## Pickle Safety

Never load a pickle from an unknown or untrusted source. Unpickling can execute malicious code. This lesson reloads only the program-owned file created by the same notebook.


## Preparing the Python Object

This dictionary contains the values that will be saved in binary form.


```python
import os
import pickle

if os.path.isdir("01-3_Python_Basics_III"):
    os.chdir("01-3_Python_Basics_III")

study_snapshot = {
    "student": "Amina",
    "week": 3,
    "topics": ["Python", "CSV", "JSON"],
    "minutes": 180,
    "completed": True,
}

print("Original object:", study_snapshot)
```

The dictionary combines strings, numbers, a list, and a Boolean value that pickle will preserve with their Python types.


## Saving a Python Object

Create the output directory, open a `.pkl` file in binary write mode (`"wb"`), and pass the object and file to `pickle.dump()`.


```python
output_directory = os.path.join("outputs", "05_binary_files_and_pickle")
os.makedirs(output_directory, exist_ok=True)
output_path = os.path.join(output_directory, "study_snapshot.pkl")

with open(output_path, mode="wb") as output_file:
    pickle.dump(study_snapshot, output_file)

print("Pickle saved:", os.path.isfile(output_path))
print("Saved file:", output_path)
print("Saved size in bytes:", os.path.getsize(output_path))
```

`pickle.dump()` performs the binary serialization and the `with` block closes the file after writing.


## Loading the Trusted Pickle

Open the same program-owned file in binary read mode (`"rb"`) and restore the object with `pickle.load()`.


```python
# Safe here because this notebook created the pickle itself.
with open(output_path, mode="rb") as input_file:
    reloaded_snapshot = pickle.load(input_file)

print("Reloaded object:", reloaded_snapshot)
print("Python type:", type(reloaded_snapshot).__name__)
print("Saved data matches:", reloaded_snapshot == study_snapshot)
```

The dictionary and its nested Python values are restored with their original types, and the round-trip comparison is `True`.


## Handling a Missing Pickle

A missing pickle raises the same `FileNotFoundError` as other file types. Catching that specific exception keeps the example runnable.


```python
missing_path = os.path.join(output_directory, "missing_snapshot.pkl")

try:
    with open(missing_path, mode="rb") as input_file:
        pickle.load(input_file)
except FileNotFoundError:
    print("Pickle file not found:", os.path.basename(missing_path))
```
