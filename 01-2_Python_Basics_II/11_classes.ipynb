# Classes

*A class defines a new kind of object by keeping related state and behavior together.*

## Small Class Definition

`__init__` runs when an instance is created. `self` refers to that particular instance, and attributes store its state.


```python
class Measurement:
    def __init__(self, feature, value, unit):
        self.feature = feature
        self.value = value
        self.unit = unit

    def describe(self):
        return f"{self.feature}: {self.value} {self.unit}"
```

## Instance Creation and Method Calls

The class is the definition; `bill_length` and `body_mass` are two independent instances. A method reads instance state through `self`.


```python
bill_length = Measurement("bill length", 39.1, "mm")
body_mass = Measurement("body mass", 3750, "g")

print(bill_length.describe())
print(body_mass.describe())
```

## Independent Instance State

Each object keeps its own attribute values.


```python
bill_length.value = 40.3

print("Updated:", bill_length.describe())
print("Unchanged:", body_mass.describe())
```

## Class Attributes, Instance Attributes, and `__str__`

A class attribute is shared as a default across instances. Instance attributes belong to one object. `__str__()` defines the readable text produced by `print(object)`.

```python
class Course:
    platform = "JSPCV"

    def __init__(self, title, enrolled):
        self.title = title
        self.enrolled = enrolled

    def __str__(self):
        return f"{self.title}: {self.enrolled} students"

python_course = Course("Python Basics", 30)
statistics_course = Course("Statistics", 24)

print(python_course)
print(statistics_course)
print("Shared platform:", Course.platform)
print("Is a Course object:", isinstance(python_course, Course))
```

Both objects share the class-level platform but store different titles and enrollment counts. `isinstance()` checks whether an object was created from a class.

## Use Cases

Classes commonly model domain records, manage changing state, and attach repeated analysis behavior to stored data.

### Model a Domain Record

The existing `Measurement` class keeps a feature, value, unit, and display rule together. New measurements follow the same structure.

```python
temperature = Measurement("water temperature", 18.6, "°C")
flipper_length = Measurement("flipper length", 181, "mm")

print(temperature.describe())
print(flipper_length.describe())
```

Each object stores different state but shares the `describe()` behavior defined by the class.

### Manage Changing State

A study-session object can keep its target and current progress together. Calling a method changes only that instance.

```python
class StudySession:
    def __init__(self, subject, target_minutes):
        self.subject = subject
        self.target_minutes = target_minutes
        self.completed_minutes = 0

    def add_minutes(self, minutes):
        self.completed_minutes = self.completed_minutes + minutes

    def remaining_minutes(self):
        remaining = self.target_minutes - self.completed_minutes
        return max(remaining, 0)

python_session = StudySession("Python", 90)
python_session.add_minutes(35)
python_session.add_minutes(25)

print("Completed:", python_session.completed_minutes)
print("Remaining:", python_session.remaining_minutes())
```

The object remembers progress between method calls. `remaining_minutes()` reads the current state and prevents a negative remaining value.

### Keep Data and Analysis Together

A small numeric-series class can protect its own copy of the values and provide the same summary structure for every series.

```python
class NumericSeries:
    def __init__(self, name, values):
        self.name = name
        self.values = values.copy()

    def mean(self):
        return sum(self.values) / len(self.values)

    def summary(self):
        return {
            "name": self.name,
            "count": len(self.values),
            "mean": round(self.mean(), 2),
        }

study_hours = NumericSeries("daily study hours", [1.5, 2.0, 1.0, 2.5])
print(study_hours.summary())
```

The instance owns its copied values and exposes a predictable `summary()` method. This pattern becomes useful when many related datasets need the same operations.

### Configuration with Derived Values

A configuration object can store related dimensions and calculate information derived from its current state.

```python
class ChartCanvas:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def pixel_count(self):
        return self.width * self.height

    def resize(self, scale):
        self.width = round(self.width * scale)
        self.height = round(self.height * scale)

canvas = ChartCanvas(800, 600)
print("Original pixels:", canvas.pixel_count())

canvas.resize(0.5)
print("Resized dimensions:", (canvas.width, canvas.height))
print("Resized pixels:", canvas.pixel_count())
```

The resize operation updates the related dimensions together, and `pixel_count()` always uses the object's current state.

Use a class when a domain concept needs both state and reusable behavior. At this stage, understanding instances, attributes, and methods is more important than inheritance or advanced class syntax.
