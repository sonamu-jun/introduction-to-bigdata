# Linked Lists

*Represent a sequence with linked nodes and insert a value by reconnecting references.*

A **linked list** stores each value in a node alongside a reference to the
next node. `head` refers to the first node; `None` marks the end. The links
determine the order, so nodes do not need consecutive storage locations.


## Nodes and Traversal

`Node` groups `data` and `next`. `LinkedList` keeps the entry point, `head`.
A **traversal** repeatedly follows `next` until it reaches `None`.

```text
head -> [BusStop-A | next] -> [BusStop-B | next] -> ... -> [BusStop-F | None]
```

`append()` walks to the last node and connects a new one. `insert_after()`
first searches for the target, then connects the new node to the old
successor before changing the target's link. This order preserves the
rest of the route. `values()` collects the node data for display.


````python
class Node:
    def __init__(self, data, next_node=None):
        self.data = data
        self.next = next_node


class LinkedList:
    def __init__(self):
        self.head = None

    def append(self, data):
        if self.head is None:
            self.head = Node(data)
            return

        current_node = self.head
        while current_node.next is not None:
            current_node = current_node.next
        current_node.next = Node(data)

    def insert_after(self, target, data):
        current_node = self.head
        while current_node is not None:
            if current_node.data == target:
                new_node = Node(data, current_node.next)
                current_node.next = new_node
                return True
            current_node = current_node.next
        return False

    def values(self):
        result = []
        current_node = self.head
        while current_node is not None:
            result.append(current_node.data)
            current_node = current_node.next
        return result
````

## Inserting Between Nodes

The bus route keeps its stops in travel order. Adding `BusStop-C-2` after
`BusStop-C` changes two references while retaining the following stops.


**Question.** After inserting `BusStop-C-2`, which stop follows it? Do the later stops disappear?


````python
route = LinkedList()
stops = ["BusStop-A", "BusStop-B", "BusStop-C", "BusStop-D", "BusStop-E", "BusStop-F"]
for stop in stops:
    route.append(stop)

print("Original route:", " -> ".join(route.values()))
inserted = route.insert_after("BusStop-C", "BusStop-C-2")
print("Inserted:", inserted)
print("Updated route:", " -> ".join(route.values()))
````

`BusStop-D` follows the new stop, and the route still ends at `BusStop-F`.
The new node keeps the old successor, so the list grows from six stops to
seven without losing a link.

Finding the target takes `O(n)` time in the worst case; reconnecting the
links takes `O(1)` once the target node is known. This implementation also
traverses the list when appending because it stores no reference to the tail.
