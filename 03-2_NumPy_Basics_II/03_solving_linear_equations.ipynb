# Solving Linear Equations

*Represent a linear system as arrays, solve for its unknowns, and substitute the result back into the equations.*


## Representing a System

Consider two equations:

```text
2x + y = 7
 x - y = 2
```

In `coefficients @ solution = constants`, each row of `coefficients`
contains one equation's coefficients. Each column refers to one
unknown, in the same order throughout. `np.linalg.solve()` expects
a square, nonsingular coefficient matrix for a unique solution.

**Question.** What values of `x` and `y` satisfy both equations?


````python
import numpy as np

coefficients = np.array([[2, 1],
                         [1, -1]], dtype=float)
constants = np.array([7, 2], dtype=float)
solution = np.linalg.solve(coefficients, constants)

print("Solution [x, y]:", solution)
````

The solution is `[3., 1.]`, so `x = 3` and `y = 1`.
The column order determines which unknown each output represents.


## Substituting the Answer

Multiply the coefficient matrix by the computed solution to check
the right-hand sides. The residual is `coefficients @ solution - constants`.
It should be zero or sufficiently close to zero for the problem's scale.


````python
reconstructed_constants = coefficients @ solution
residual = reconstructed_constants - constants

print("Reconstructed constants:", reconstructed_constants)
print("Residual:", residual)
print("Equations satisfied:", np.allclose(reconstructed_constants, constants))
````

Substitution returns `[7., 2.]`, the residual is zero, and the check
is `True`. Both original equations are satisfied.


## Three Unknowns

The same representation extends to three equations and three unknowns.
The columns below represent `x`, `y`, and `z`, respectively.

```text
 x +  y +  z = 6
2x -  y +  z = 3
 x + 2y + 3z = 14
```


````python
coefficients_three = np.array([[1, 1, 1],
                               [2, -1, 1],
                               [1, 2, 3]], dtype=float)
constants_three = np.array([6, 3, 14], dtype=float)
solution_three = np.linalg.solve(coefficients_three, constants_three)

print("Solution [x, y, z]:", solution_three)
print("Reconstructed constants:", coefficients_three @ solution_three)
print("Equations satisfied:", np.allclose(coefficients_three @ solution_three, constants_three))
````

The solution is `[1., 2., 3.]`. Substitution gives `[6., 3., 14.]`,
confirming all three equations rather than checking just one.


## When There Is No Unique Solution

Repeated coefficient rows do not provide independent constraints.
A singular system can have no solution or multiple solutions.
Catch `np.linalg.LinAlgError` for the deliberately singular example
so the rest of the notebook can still run.


````python
repeated_coefficients = np.array([[1, 1],
                                  [1, 1]], dtype=float)
inconsistent_constants = np.array([1, 2], dtype=float)

try:
    np.linalg.solve(repeated_coefficients, inconsistent_constants)
except np.linalg.LinAlgError:
    print("The singular coefficient matrix cannot provide a unique solution.")
````

Here the equations demand both `x + y = 1` and `x + y = 2`, which is
impossible. The handled exception reports the failure without
presenting an invalid answer.


Keep the unknown order consistent, use `np.linalg.solve()` for a square nonsingular system, and verify by substitution. [NumPy solve](https://numpy.org/doc/2.3/reference/generated/numpy.linalg.solve.html)
