# CSV Files

*Learn to load, save, and reopen tabular CSV data with Python's `csv` module.*


## Locating the Input File

Build the path to the supplied CSV before loading its rows.


```python
import csv
import os

if os.path.isdir("01-3_Python_Basics_III"):
    os.chdir("01-3_Python_Basics_III")

input_directory = os.path.join("assets", "03_csv_files")
input_path = os.path.join(input_directory, "scores.csv")
```

## Loading a CSV File

Open CSV files with `newline=""` so the `csv` module can handle line endings correctly. `csv.reader()` returns each row as a list of strings.


```python
with open(input_path, mode="r", encoding="utf-8", newline="") as source_file:
    reader = csv.reader(source_file)
    loaded_rows = list(reader)

for row in loaded_rows:
    print(row)
```

The quoted value `Statistics, Intro` remains one field even though it contains a comma.


## Saving a CSV File

Create the output directory, then use `csv.writer()` so CSV quoting rules are applied correctly.


```python
output_directory = os.path.join("outputs", "03_csv_files")
os.makedirs(output_directory, exist_ok=True)
output_path = os.path.join(output_directory, "scores_copy.csv")

with open(output_path, mode="w", encoding="utf-8", newline="") as saved_file:
    writer = csv.writer(saved_file)
    writer.writerows(loaded_rows)

print("Saved file:", output_path)
print("Saved size in bytes:", os.path.getsize(output_path))
```

The copied CSV contains the header and three data records.


## Reopening the Saved CSV

Read the output through the same CSV parser and compare the resulting rows.


```python
with open(output_path, mode="r", encoding="utf-8", newline="") as saved_file:
    reloaded_rows = list(csv.reader(saved_file))

print("Saved rows match:", reloaded_rows == loaded_rows)
print("Row count including header:", len(reloaded_rows))
```

Four reopened rows—including the header—match the loaded source rows exactly.
