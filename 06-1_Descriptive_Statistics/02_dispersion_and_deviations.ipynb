# Dispersion and Deviations

*Connect deviations, variance, and standard deviation to the spread and units of observations.*


## Range Describes the Full Observed Span

The range is $\max(x)-\min(x)$. It is easy to interpret, but two endpoints
cannot show how the other observations are distributed. `np.ptp()` computes
this maximum-to-minimum difference.


```python
import numpy as np
import pandas as pd

values = np.array([1, 3, 5, 7, 9], dtype=float)
with_extreme = np.array([1, 3, 5, 7, 1000], dtype=float)
print("Original range:", np.ptp(values))
print("Range with an extreme endpoint:", np.ptp(with_extreme))
```

The range increases from 8 to 999 even though four of the five observations
are unchanged. The reported span is correct, but it gives little information
about where most values lie.


## Signed Deviations Cancel

The deviation of observation $i$ is $d_i=x_i-\bar{x}$. Negative values lie
below the mean; positive values lie above it. Since
$\sum_i(x_i-\bar{x})=\sum_i x_i-n\bar{x}=0$, the sum of signed deviations
cannot measure the dataset's overall spread.


```python
mean_value = values.mean()
deviations = values - mean_value
deviation_table = pd.DataFrame(
    {"value": values, "deviation": deviations, "squared_deviation": deviations**2}
)
print(deviation_table.to_string(index=False))
print("Mean:", mean_value)
print("Sum of deviations:", deviations.sum())
print("Sum of squared deviations:", np.sum(deviations**2))
```

The mean is 5 and the deviations are -4, -2, 0, 2, and 4. They sum to zero,
while their squares sum to 40. Squaring prevents positive and negative
deviations from cancelling.


## Variance and Standard Deviation Have Different Units

When describing these five values as the complete population of interest,
use $\sigma^2=\sum_i(x_i-\mu)^2/N$ and $\sigma=\sqrt{\sigma^2}$.
`ddof=0` divides by the observation count. If values are lengths in meters,
variance has units of square meters and standard deviation has units of meters.


```python
population_variance = np.var(values, ddof=0)
population_std = np.std(values, ddof=0)
print("Population variance:", population_variance)
print(f"Population standard deviation: {population_std:.4f}")
print("Square root matches:", np.isclose(np.sqrt(population_variance), population_std))
```

The population variance is 40 / 5 = 8 and the standard deviation is about
2.8284. Standard deviation is a distance scale in the original units; it does
not say that every observation is that far from the mean.


## A Shared Center Can Hide Different Spreads

Compare two groups with the same mean. A center alone cannot distinguish a
tightly clustered set from a more dispersed set.


```python
groups = {"compact": np.array([4, 5, 6]), "spread": np.array([1, 5, 9])}
for group_name, group_values in groups.items():
    print(
        group_name,
        "mean:",
        group_values.mean(),
        "population std:",
        round(group_values.std(ddof=0), 4),
    )
```

Both means are 5. Their population standard deviations are about 0.8165
and 3.2660, showing that the second group is four times as spread out on
this scale.


Report a measure of spread with the center. Squared deviations lead to variance, while taking the square root restores the original unit.
