# Graphs

*Build an undirected graph and read direct connections from adjacency lists.*

A **graph** represents items as **vertices** (nodes) and connections as
**edges**. It can have cycles, several routes between two vertices, or
disconnected parts. A tree is a connected graph with no cycles.

An **undirected** edge connects its two endpoints equally. A directed
edge instead has a source and a destination. The city network below uses
undirected edges and stores no distances or other edge weights.


## Adjacency Lists

An **adjacency list** stores each vertex's direct neighbors. A dictionary
associates city names with neighbor lists; each undirected edge appears
in both endpoints' lists.

```text
Suwon -- Seoul -- Pohang -- Busan
                     |
                    Jeju
```

`add_vertex()` adds a city with an empty neighbor list. `add_edge()` adds
both endpoints if needed, then records both directions. This example
represents a simple graph: self-edges are rejected, and adding the same
edge again does not duplicate it.


````python
class Graph:
    def __init__(self):
        self.neighbors = {}

    def add_vertex(self, vertex):
        if vertex not in self.neighbors:
            self.neighbors[vertex] = []

    def add_edge(self, first, second):
        if first == second:
            raise ValueError("An edge must connect two different cities.")
        self.add_vertex(first)
        self.add_vertex(second)
        if second not in self.neighbors[first]:
            self.neighbors[first].append(second)
            self.neighbors[second].append(first)

    def show(self):
        for vertex in sorted(self.neighbors):
            print(f"{vertex} -> {sorted(self.neighbors[vertex])}")


cities = Graph()
routes = [
    ("Seoul", "Pohang"),
    ("Pohang", "Busan"),
    ("Seoul", "Suwon"),
    ("Pohang", "Jeju"),
]
for first_city, second_city in routes:
    cities.add_edge(first_city, second_city)

cities.show()
````

Pohang's neighbors are Busan, Jeju, and Seoul. Every edge is recorded
twice: Seoul lists Pohang, and Pohang lists Seoul. Sorting only fixes
the display order; neighbor order has no meaning in this network.


## Degree and Paths

A vertex's **degree** is its number of direct neighbors. Summing all
degrees counts each undirected edge twice, so dividing by two gives
the edge count.

A **path** follows edges through vertices. A missing direct edge does
not rule out a path through intermediate cities. Set intersection `&`
finds shared neighbors, each of which supplies a two-edge path.


**Question.** Are Seoul and Busan directly connected? Which city connects them in two edges, and what is that city’s degree?


````python
shared_neighbors = set(cities.neighbors["Seoul"]) & set(cities.neighbors["Busan"])
edge_count = sum(len(neighbors) for neighbors in cities.neighbors.values()) // 2

print("Seoul to Busan direct:", "Busan" in cities.neighbors["Seoul"])
print("Shared neighbors:", sorted(shared_neighbors))
print("Pohang's degree:", len(cities.neighbors["Pohang"]))
print("Vertex count:", len(cities.neighbors))
print("Edge count:", edge_count)
````

Seoul and Busan share Pohang, giving the path `Seoul -> Pohang -> Busan`.
They have no direct edge. Pohang has degree three; the network has five
vertices and four edges.

This particular graph is connected and has no cycle, so it also forms
a tree. The graph representation can hold more general connections: an
additional Seoul–Busan edge would create a cycle. The diagram records
connections only; it does not establish travel distances or shortest routes.
