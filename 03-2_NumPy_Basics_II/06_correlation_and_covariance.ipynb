# Correlation and Covariance

*Measure relationships between columns while keeping variable order, units, and interpretation explicit.*

The prepared table contains eight student records. Each row is one student; the four columns are study hours, attendance percentage, gaming hours, and exam score.


## Observations and Variables

A relationship compares two variables across the same observations.
Keep student rows aligned and keep the feature names in the same
order as the columns. Column index 3 contains exam scores.


```python
import numpy as np

feature_names = ["study_hours", "attendance_rate", "gaming_hours", "exam_score"]
student_data = np.array([[1, 65, 9, 60],
                         [2, 90, 4, 68],
                         [3, 72, 8, 66],
                         [4, 88, 3, 80],
                         [5, 70, 7, 72],
                         [6, 95, 2, 88],
                         [7, 68, 8, 70],
                         [8, 85, 4, 84]], dtype=float)

print("Feature order:", feature_names)
print("Data shape:", student_data.shape)
print("Student data:")
print(student_data)
```

Shape `(8, 4)` means eight observations of four variables. A
variable-by-variable relationship matrix should therefore have shape
`(4, 4)`, not `(8, 8)`.


## Covariance and Units

`np.cov(data, rowvar=False)` treats columns as variables. Positive
covariance means that deviations from the two means tend to have
the same sign; negative covariance means their signs tend to differ.
We use `ddof=1`, so the denominator is `N - 1`.

Covariance depends on the variables' units. Its diagonal entries are
sample variances, while off-diagonal entries compare different variables.


```python
covariance = np.cov(student_data, rowvar=False, ddof=1)

print("Covariance matrix:")
print(np.round(covariance, 2))
print("Study hours and score covariance:", round(covariance[0, 3], 2))
print("Exam-score sample variance:", round(covariance[3, 3], 2))
```

Study hours and score have positive covariance, about 16.86
hour-points. The exam-score diagonal is about 92.29 points squared.
Covariances for attendance percentages and hours cannot be compared
directly as if they had the same scale.


## A Standardized Linear Relationship

`np.corrcoef(data, rowvar=False)` calculates Pearson correlations.
It standardizes covariance by the variables' standard deviations,
giving a unitless measure between -1 and 1, apart from numerical
rounding. Values near either endpoint indicate a strong linear
relationship; values near zero indicate a weak linear relationship.

**Question.** Which of the first three variables has the largest
absolute correlation with exam score in these eight records?


```python
correlation = np.corrcoef(student_data, rowvar=False)

print("Correlation matrix:")
print(np.round(correlation, 4))
for index, name in enumerate(feature_names[:3]):
    print(f"{name} vs exam_score: {correlation[index, 3]:.4f}")
```

Gaming hours has the largest absolute correlation with score:
about **-0.8440**. Study hours is about 0.7164 and attendance is about
0.7523. The negative sign describes opposite directions; it does not
make the relationship weaker than a smaller positive value.

These are associations within eight prepared records. Correlation
alone does not show a cause or establish a pattern for all students.
A constant variable has zero standard deviation, so its Pearson
correlation is undefined.


Check `rowvar`, feature order, and units before interpreting a relationship matrix. Use covariance for joint variation in the original units and correlation for standardized linear association. [NumPy covariance](https://numpy.org/doc/2.3/reference/generated/numpy.cov.html), [NumPy correlation](https://numpy.org/doc/2.3/reference/generated/numpy.corrcoef.html)
