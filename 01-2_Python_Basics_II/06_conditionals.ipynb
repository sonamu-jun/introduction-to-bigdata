# Conditionals

*Use Boolean conditions to select exactly the code branch that matches the current observation.*

## Two-Way Branching

The condition after `if` is evaluated first. `else` handles every remaining case.


```python
score = 82

if score >= 60:
    result = "pass"
else:
    result = "fail"

print(result)
```

## Ordered Multiway Branching

The most specific high threshold belongs first. Once a true branch runs, Python skips the rest of the chain.


```python
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D or F"

print("Grade:", grade)
```

## Validation Before Classification

Validate the input range before applying the classification rule.


```python
if not 0 <= score <= 100:
    print("Invalid score")
elif score >= 60:
    print("Valid passing score")
else:
    print("Valid non-passing score")
```

## Truthiness in Conditions

A condition may use a Boolean expression directly. Empty strings and collections are false; non-empty ones are true.

```python
selected_rows = []

if selected_rows:
    selection_message = "Rows are ready"
else:
    selection_message = "No rows selected"

print(selection_message)
```

The empty list is false in a Boolean context, so the `else` branch runs. This form is clearer than `if len(selected_rows) > 0:` for a simple presence check.

## Conditional Expressions

A conditional expression chooses between two short values inside one expression. Its order is `value_if_true if condition else value_if_false`.

```python
row_count = 120
dataset_status = "ready" if row_count > 0 else "empty"

print("Dataset status:", dataset_status)
```

The condition is true, so `dataset_status` receives `ready`. Use an ordinary `if` statement when either branch needs several operations.

Branch order is part of the rule. Validate the allowed range before interpreting a value, and use `==` only when testing equality.
