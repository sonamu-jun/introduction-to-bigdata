# Comprehensions

*Compress a simple “transform each value” loop only after the expanded loop is understood.*

## Expanded Transformation Loop

This loop creates a new list without changing `numbers`.


```python
numbers = [1, 2, 3, 4, 5]
squares_with_loop = []

for number in numbers:
    squares_with_loop.append(number ** 2)

print(squares_with_loop)
```

## Compact Transformations

The expression `number ** 2` is evaluated once for each `number` in `numbers`.


```python
squares = [number ** 2 for number in numbers]
print(squares)
print("Same result:", squares == squares_with_loop)
```

## Filtered Comprehensions

The condition comes last but is checked before the value is added to the result.


```python
even_squares = [number ** 2 for number in numbers if number % 2 == 0]
print(even_squares)
```

## Dictionary Comprehensions

Placing `key: value` before `for` builds a dictionary instead of a list. Here, each number becomes a key and its square becomes the value.


```python
square_lookup = {number: number ** 2 for number in numbers}
print(square_lookup)
```

## Use Cases

Comprehensions are commonly used for direct transformations, filtering, lookup construction, flattening, unique-value normalization, and compact conditional labels.

### Measurement-Unit Transformation

Each Celsius value receives the same short calculation, so a list comprehension clearly expresses one input-to-one-output transformation.

```python
temperatures_c = [18.0, 20.5, 23.0, 25.5]
temperatures_f = [round(value * 9 / 5 + 32, 1) for value in temperatures_c]

print("Celsius:", temperatures_c)
print("Fahrenheit:", temperatures_f)
```

The result contains one converted value for every source value, in the same order. The original Celsius list remains unchanged.

### Record Filtering and Field Extraction

The expression before `for` selects what to store, and the final `if` decides which records enter the result.

```python
penguins = [
    {"species": "Adelie", "body_mass_g": 3750},
    {"species": "Gentoo", "body_mass_g": 5000},
    {"species": "Chinstrap", "body_mass_g": 3700},
    {"species": "Gentoo", "body_mass_g": 5200},
]

heavy_species = [
    penguin["species"]
    for penguin in penguins
    if penguin["body_mass_g"] >= 5000
]

print("Species at or above 5000 g:", heavy_species)
```

Only the two records meeting the mass condition contribute a species name. The comprehension performs one clear selection and one clear extraction.

### Lookup-Dictionary Construction

A dictionary comprehension can convert a list of labeled records into a mapping for direct key-based retrieval.

```python
measurements = [
    {"feature": "bill_length", "unit": "mm"},
    {"feature": "body_mass", "unit": "g"},
    {"feature": "temperature", "unit": "°C"},
]

unit_lookup = {
    measurement["feature"]: measurement["unit"]
    for measurement in measurements
}

print("Unit lookup:", unit_lookup)
print("Body-mass unit:", unit_lookup["body_mass"])
```

Each record supplies one key and one value. The resulting dictionary supports direct retrieval without searching the source list again.

### One-Level List Flattening

Two `for` clauses can collect values from a simple list of lists. Read them from left to right: first select a day, then select an hour from that day.

```python
weekly_hours = [[1.5, 2.0], [1.0], [2.5, 1.5]]
all_hours = [hour for day_hours in weekly_hours for hour in day_hours]

print("Nested hours:", weekly_hours)
print("Flattened hours:", all_hours)
```

The nested lists become one flat list while preserving left-to-right order. A regular loop is clearer when flattening requires additional branches or several processing steps.

### Unique Normalized Categories

A set comprehension applies one normalization expression and removes repeated results.

```python
raw_species = [" adelie", "Adelie ", "GENTOO", "gentoo"]
normalized_species = {name.strip().title() for name in raw_species}

print("Normalized species:", sorted(normalized_species))
```

Whitespace and capitalization differences disappear, and the set keeps one value for each normalized species.

### Conditional Labels During Transformation

A conditional expression chooses one of two output values for every input. Unlike a final comprehension `if`, it does not remove items.

```python
exam_scores = [72, 55, 91, 60]
result_labels = ["pass" if score >= 60 else "review" for score in exam_scores]

print("Scores:", exam_scores)
print("Labels:", result_labels)
```

Every score produces one label, so the output length matches the input length. Keep this form to one short condition; use an ordinary loop for more complex decisions.

Use a comprehension for one short transformation and at most one clear filter. Keep an ordinary loop when several steps, branches, or side effects would make the compact form harder to read.
