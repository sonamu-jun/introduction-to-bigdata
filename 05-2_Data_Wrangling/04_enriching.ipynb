## Add a Related Field

<span style="color: #777777; font-style: italic;">관련 항목 추가</span>

- `Enriching` adds related information to a record.  
  <span style="color: #777777; font-style: italic;">`정보 보강`(enriching)은 기록에 관련 정보를 더하는 작업입니다.</span>

- `merge(..., on="subject", how="left")` performs a `left join`: match subjects and keep all rows from the left table.  
  <span style="color: #777777; font-style: italic;">`merge(..., on="subject", how="left")`는 과목으로 연결하고 왼쪽 표의 모든 행을 유지하는 `왼쪽 조인`(left join)입니다.</span>

  - Each subject appears once in this department table, so each score gets at most one match.  
    <span style="color: #777777; font-style: italic;">소속 분야 표에는 과목이 한 번씩만 나오므로 각 점수에 최대 하나의 행이 연결됩니다.</span>


- Match each subject to its department while keeping the existing scores.  
  <span style="color: #777777; font-style: italic;">과목을 기준으로 연결해 기존 기록에 소속 분야를 추가합니다.</span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/05-2_Data_Wrangling/assets/04_enriching/concept.webp" width="469" alt="Match each subject to its department while keeping the existing scores.">


```python
import pandas as pd

scores = pd.DataFrame({
    "name": ["Jiyoung", "Hojun", "Hana"],
    "subject": ["Mathematics", "English", "Science"],
    "score": [98, 90, 85],
    "maximum": [100, 100, 100],
})
departments = pd.DataFrame({
    "subject": ["Mathematics", "English", "Science"],
    "department": ["STEM", "Languages", "STEM"],
})
enriched = scores.merge(departments, on="subject", how="left")

print(enriched)
```

- All three rows now have a department.  
  <span style="color: #777777; font-style: italic;">세 행 모두 소속 분야가 추가됩니다.</span>

- The scores stay the same.  
  <span style="color: #777777; font-style: italic;">점수는 그대로입니다.</span>


## Calculate a New Field

<span style="color: #777777; font-style: italic;">새 항목 계산</span>

- A `derived variable` is calculated from existing fields; divide score by maximum to get a `proportion`.  
  <span style="color: #777777; font-style: italic;">`파생 변수`(derived variable)은 기존 항목으로 계산한 값이며, 점수를 만점으로 나누면 `비율`(proportion)이 됩니다.</span>

  - The maximum scores here are all positive.  
    <span style="color: #777777; font-style: italic;">여기의 만점은 모두 양수입니다.</span>


```python
enriched["proportion"] = enriched["score"] / enriched["maximum"]

print(enriched[["name", "proportion"]])
```

- The proportions are 0.98, 0.90, and 0.85, or 98%, 90%, and 85% of the maximum.  
  <span style="color: #777777; font-style: italic;">비율은 0.98, 0.90, 0.85이며, 만점의 98%, 90%, 85%를 뜻합니다.</span>
