# Classes

*A class defines a new kind of object by keeping related state and behavior together.*

## Small Class Definition

`__init__` runs when an instance is created. `self` refers to that particular instance, and attributes store its state.


```python
class Measurement:
    def __init__(self, feature, value, unit):
        self.feature = feature
        self.value = value
        self.unit = unit

    def describe(self):
        return f"{self.feature}: {self.value} {self.unit}"
```

## Instance Creation and Method Calls

The class is the definition; `bill_length` and `body_mass` are two independent instances. A method reads instance state through `self`.


```python
bill_length = Measurement("bill length", 39.1, "mm")
body_mass = Measurement("body mass", 3750, "g")

print(bill_length.describe())
print(body_mass.describe())
```

## Independent Instance State

Each object keeps its own attribute values.


```python
bill_length.value = 40.3

print("Updated:", bill_length.describe())
print("Unchanged:", body_mass.describe())
```

## Class Attributes, Instance Attributes, and `__str__`

A class attribute is shared as a default across instances. Instance attributes belong to one object. `__str__()` defines the readable text produced by `print(object)`.

```python
class Course:
    platform = "JSPCV"

    def __init__(self, title, enrolled):
        self.title = title
        self.enrolled = enrolled

    def __str__(self):
        return f"{self.title}: {self.enrolled} students"

python_course = Course("Python Basics", 30)
statistics_course = Course("Statistics", 24)

print(python_course)
print(statistics_course)
print("Shared platform:", Course.platform)
print("Is a Course object:", isinstance(python_course, Course))
```

Both objects share the class-level platform but store different titles and enrollment counts. `isinstance()` checks whether an object was created from a class.

Use a class when a domain concept needs both state and reusable behavior. At this stage, understanding instances, attributes, and methods is more important than inheritance or advanced class syntax.
