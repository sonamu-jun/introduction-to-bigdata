# 012-002. Lists

*Use a list when several values belong in an ordered, changeable sequence.*

The conceptual foundation is developed in Collections and Iteration.

## Method · List Creation and Selection

Square brackets create a list. Indexes select one value; slices select a new list. The stop index is excluded.

```python
scores = [72, 85, 91, 68, 88]

print("All scores:", scores)
print("First score:", scores[0])
print("Last score:", scores[-1])
print("Middle scores:", scores[1:4])
print("Number of scores:", len(scores))
```

## Implementation · Sequence Updates

Lists are mutable. Assignment changes a selected position, `append()` adds one value, and `pop()` removes and returns a value.

```python
scores[3] = 70
scores.append(95)
removed_score = scores.pop()

print("Updated scores:", scores)
print("Removed score:", removed_score)
print("Contains 91:", 91 in scores)
```

## Implementation · Independent List Copies

Two variable names can refer to the same mutable list. `copy()` creates a separate list when later changes should be independent.

> **Before you run**
>
> Predict whether sorting `sorted_scores` will also sort `scores`.

```python
sorted_scores = scores.copy()
sorted_scores.sort()

print("Original order:", scores)
print("Sorted copy:", sorted_scores)
```

## Implementation · Nested-List Indexing

A nested list can represent rows before we introduce arrays and DataFrames.

```python
records = [["A", 72], ["B", 85], ["C", 91]]

print("Second record:", records[1])
print("Second record's score:", records[1][1])
```

### Interpretation · Nested-List Indexing

A list preserves insertion order and repeated values. Use a copy when an operation such as sorting should not change the original sequence.
