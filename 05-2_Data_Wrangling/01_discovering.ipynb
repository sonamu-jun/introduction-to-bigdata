## Inspect the Input

<span style="color: #777777;"><em>입력 데이터 확인</em></span>

- `Discovering` checks the data before changing it.  
  <span style="color: #777777;"><em>`현황 파악`(discovering)은 수정 전에 데이터 상태를 살피는 단계입니다.</em></span>

- A `DataFrame` stores a table; `.isna().sum()` counts missing values in each column.  
  <span style="color: #777777;"><em>`데이터프레임`(DataFrame)은 표를 저장하며, `.isna().sum()`은 열별 결측값 개수를 셉니다.</em></span>


```python
import pandas as pd

data = pd.DataFrame({
    "age": [25, 30, None, 22, 400],
    "score": ["85", 90, 78, "eighty", 88],
})

print(data)
print(data.isna().sum())
```

- One age is missing; `eighty` is text, and age 400 is outside the allowed range of 0–120.  
  <span style="color: #777777;"><em>나이 하나가 빠져 있으며, `eighty`는 문자열이고 나이 400은 허용 범위 0–120을 벗어납니다.</em></span>

- A missing-value check does not detect every data problem.  
  <span style="color: #777777;"><em>결측값 검사만으로 모든 데이터 문제를 찾을 수는 없습니다.</em></span>


- Missing values, wrong formats, and range errors need different checks.  
  <span style="color: #777777;"><em>결측값, 형식 오류, 범위 오류는 서로 다른 검사가 필요합니다.</em></span>
