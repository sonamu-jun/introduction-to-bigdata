# Missing Values

*Locate missing values and compare dropping incomplete records with an explicit filling rule.*


## Detect Missing Cells

`None` in these prepared records represents an unknown value. pandas may
display missing values as `NaN`, `NaT`, or `<NA>` depending on the column type.
Use `.isna()` to detect them; equality with a missing marker is not a reliable test.
Summing the Boolean result counts missing cells in each column.


```python
import pandas as pd

student_df = pd.DataFrame({
    "name": ["Minji", "Joon", "Hana", "Taeho"],
    "score": [88, None, 95, 76],
    "city": ["Seoul", "Busan", None, "Incheon"],
})

print("Original records:")
print(student_df)
print("Missing cells:")
print(student_df.isna())
print("Missing count per column:")
print(student_df.isna().sum())
print("Non-missing scores:", student_df["score"].count())
print("Total rows:", len(student_df))
```

Score and city each have one missing value. There are four rows but only three observed scores, so a score count is different from a row count.


## Drop Rows Only for Required Fields

`.dropna(subset=["score"])` requires a known score and leaves unrelated
missing fields alone. With no `subset`, `.dropna()` drops a row if any
column is missing.


```python
known_scores_df = student_df.dropna(subset=["score"])

print("Rows with an observed score:")
print(known_scores_df)
print("Rows with every field present:")
print(student_df.dropna())
```

Hana has a score of 95, so the score-only selection keeps her missing city. Requiring every field instead keeps only Minji and Taeho.


## Fill According to a Stated Assumption

`.fillna(value)` returns a Series with missing entries replaced. Here
the example fills the unknown score with the observed mean and labels
the unknown city `Unknown`. The mean skips missing scores by default.
An imputed score is an assumption, not an observed result or a grading rule.


```python
observed_mean = student_df["score"].mean()
filled_df = student_df.copy()
filled_df["score"] = filled_df["score"].fillna(observed_mean)
filled_df["city"] = filled_df["city"].fillna("Unknown")

print(f"Observed score mean: {observed_mean:.2f}")
print("Filled records:")
print(filled_df.round({"score": 2}))
print("Remaining missing counts:")
print(filled_df.isna().sum())
print("Original missing score count:", student_df["score"].isna().sum())
```

Joon receives approximately 86.33 from the three observed scores; Hana's city becomes `Unknown`. The filled table has no missing cells, while the original still has its missing score.
