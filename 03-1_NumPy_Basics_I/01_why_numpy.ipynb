# Why NumPy?

*Distinguish list operations from array calculations and apply one calculation to many values.*

NumPy provides the `ndarray`, an array with numbered axes and a common element type. It is useful when the same numerical calculation must be applied to many values.


## Lists and Arrays

`import numpy as np` imports NumPy with the conventional short name `np`.
`np.array()` builds an array from a Python sequence. The meaning of `+`
depends on the objects: lists concatenate, while numeric arrays add
matching positions.

**Question.** How many values will `list_a + list_b` contain, and how
many will `array_a + array_b` contain?


```python
import numpy as np

list_a = [1, 2, 3]
list_b = [4, 5, 6]
array_a = np.array(list_a)
array_b = np.array(list_b)

print("List addition:", list_a + list_b)
print("Array addition:", array_a + array_b)
print("Array object type:", type(array_a).__name__)
```

List addition produces six values, `[1, 2, 3, 4, 5, 6]`. Array addition
produces three values, `[5, 7, 9]`, because each output combines one
pair of numbers. The array object is an `ndarray`.


## Calculating at Matching Positions

Array multiplication with `*` also operates element by element.
A list comprehension can express the same calculation with `zip()`;
the array expression states the operation directly.


```python
products_with_list = [left * right for left, right in zip(list_a, list_b)]
products_with_array = array_a * array_b

print("List comprehension:", products_with_list)
print("Array multiplication:", products_with_array)
print("Same values:", np.array_equal(products_with_array, products_with_list))
```

Both calculations return `[4, 10, 18]`. `np.array_equal()` checks that
the shapes and values match. NumPy handles the repeated arithmetic
inside its array operations instead of requiring a Python loop here.


## Applying One Number to an Array

A single number is a **scalar**. Adding a scalar to an array applies
it to every element; this is the simplest form of broadcasting.
The expression creates a result without assigning new values to the input.


```python
adjusted_values = array_a + 10

print("Adjusted values:", adjusted_values)
print("Original values:", array_a)
print("List repetition:", list_a * 2)
print("Array scaling:", array_a * 2)
```

The adjusted values are `[11, 12, 13]`, while the original array remains
`[1, 2, 3]`. Multiplying a list by 2 repeats its items; multiplying a
numeric array by 2 doubles its values.
