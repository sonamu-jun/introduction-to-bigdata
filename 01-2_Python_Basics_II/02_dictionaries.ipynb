# Dictionaries

*Use a dictionary when each value should be retrieved by a meaningful label rather than a numeric position.*


## Creation and Key-Based Retrieval

Curly braces create a dictionary, a colon connects each key to its value, and square brackets retrieve a value by key.

```python
penguin = {
    "species": "Adelie",
    "bill_length_mm": 39.1,
    "body_mass_g": 3750,
}

print("Complete record:", penguin)
print("Species:", penguin["species"])
print("Body mass:", penguin["body_mass_g"])
print("Number of fields:", len(penguin))
```

The labels make each value's meaning visible.


## Membership and Safe Retrieval

The `in` operator checks dictionary keys. Square-bracket lookup requires the key to exist, while `get()` may return a fallback value for a missing key.

```python
print("Has species key:", "species" in penguin)
print("Species with get():", penguin.get("species"))
print("Island with a fallback:", penguin.get("island", "not recorded"))
```

`penguin["island"]` would raise `KeyError` because that key is absent.


## Adding and Updating Fields

Assignment adds a new key or replaces the value of an existing key. `update()` applies several key–value pairs at once.

```python
penguin["island"] = "Torgersen"
penguin["body_mass_g"] = 3800
penguin.update({"sex": "female", "year": 2007})

print("Updated record:", penguin)
```

Adding `"island"` creates a field, while assigning `"body_mass_g"` replaces its earlier value.


## Inspecting Keys, Values, and Pairs

`keys()` exposes the field names, `values()` exposes the stored values, and `items()` exposes `(key, value)` pairs. Converting these views to lists makes their current contents easy to display.

```python
print("Keys:", list(penguin.keys()))
print("Values:", list(penguin.values()))
print("Key-value pairs:", list(penguin.items()))
```

These methods provide three views of the same dictionary.


## Copying and Removing a Field

`copy()` creates an independent dictionary. `pop(key)` removes a selected field and returns its value.

```python
analysis_record = penguin.copy()
removed_year = analysis_record.pop("year")

print("Original still has year:", "year" in penguin)
print("Analysis record:", analysis_record)
print("Removed year:", removed_year)
```

Only `analysis_record` loses the year because it is a copy.


## A List of Dictionary Records

A common table-like pattern is a list whose items are dictionaries. The list preserves row order, and each dictionary labels the fields in one row.

```python
observations = [
    penguin,
    {"species": "Gentoo", "bill_length_mm": 46.1, "body_mass_g": 5000},
]

print("First row species:", observations[0]["species"])
print("Second row body mass:", observations[1]["body_mass_g"])
```

## Dictionary Merge and Deletion Expressions

The merge operator creates a combined dictionary, while `del` removes a selected key without returning its value.

```python
default_options = {"color": "navy", "width": 640, "show_grid": True}
user_options = {"color": "orange", "width": 800}

merged_options = default_options | user_options
temporary_options = merged_options.copy()
del temporary_options["show_grid"]

print("Merged options:", merged_options)
print("After del:", temporary_options)
print("Defaults remain:", default_options)
```

When both dictionaries contain the same key, the value on the right wins.


## Dictionary Method Coverage

The table groups dictionary methods by purpose.

| Purpose | Methods |
| --- | --- |
| Retrieve safely | `get()` |
| Add a missing default | `setdefault()` |
| Inspect contents | `keys()`, `values()`, `items()` |
| Add or replace fields | `update()` |
| Remove contents | `pop()`, `popitem()`, `clear()` |
| Copy or construct | `copy()`, `fromkeys()` |


### Creating Defaults

`setdefault(key, default)` adds the default only when the key is absent and then returns the stored value. `fromkeys(keys, value)` creates a new dictionary with the same initial value for several keys.

```python
hours_by_subject = {}
hours_by_subject.setdefault("Python", 0)
hours_by_subject["Python"] = hours_by_subject["Python"] + 1.5
hours_by_subject.setdefault("Python", 99)

day_status = dict.fromkeys(["Mon", "Tue", "Wed"], "pending")

print("Hours by subject:", hours_by_subject)
print("Initial day status:", day_status)
```

The second `setdefault()` does not replace the existing Python value.


### Removing the Last Pair or Every Pair

`popitem()` removes and returns the most recently inserted key–value pair. `clear()` removes all pairs in place. Both are demonstrated on a copy.

```python
temporary_record = penguin.copy()
last_pair = temporary_record.popitem()
print("Removed last pair:", last_pair)
print("After popitem():", temporary_record)

temporary_record.clear()
print("After clear():", temporary_record)
print("Original field count:", len(penguin))
```

The temporary dictionary first loses its last pair and then becomes empty.


Choose a dictionary when labeled fields matter more than numeric positions.
