# Measures of Center

<span style="color: #777777; font-style: italic;">대푯값</span>


## Describing the Observed Data

<span style="color: #777777; font-style: italic;">관측 데이터 설명</span>

- `Descriptive statistics` summarize the data we have: center, `spread`, shape, and relationships.  
  <span style="color: #777777; font-style: italic;">`기술통계`(descriptive statistics)는 관측 자료의 중심, `산포`(spread), 모양, 관계를 요약합니다.</span>

- A `measure of center` gives a typical value: mean, median, or mode.  
  <span style="color: #777777; font-style: italic;">`대푯값`(measure of center)은 자료를 대표하는 값이며, 평균·중앙값·최빈값이 있습니다.</span>

- The five study times are 20, 25, 30, 30, and 45 minutes.  
  <span style="color: #777777; font-style: italic;">학습 시간 다섯 개는 20분, 25분, 30분, 30분, 45분입니다.</span>


- Mean, median, and mode are all 30 here, but each uses a different rule.  
  <span style="color: #777777; font-style: italic;">평균·중앙값·최빈값은 모두 30이지만 계산 기준은 서로 다릅니다.</span>

## Mean

<span style="color: #777777; font-style: italic;">평균</span>

- The `arithmetic mean` (mean) is the total divided by the number of values.  
  <span style="color: #777777; font-style: italic;">`산술평균`(arithmetic mean)은 모든 값의 합을 개수로 나눈 값이며, 여기서는 `평균`이라 부릅니다.</span>

- In the formula, $n$ is the count and $\sum$ means add the values: $\bar{x}=\frac{1}{n}\sum_i x_i$.  
  <span style="color: #777777; font-style: italic;">식에서 </span>${\color{#777777}n}$<span style="color: #777777; font-style: italic;">은 개수, </span>${\color{#777777}\sum}$<span style="color: #777777; font-style: italic;">는 값을 더한다는 뜻입니다: </span>${\color{#777777}\bar{x}=\frac{1}{n}\sum_i x_i}$<span style="color: #777777; font-style: italic;">.</span>

- `minutes.mean()` calculates the mean; a large value can raise it.  
  <span style="color: #777777; font-style: italic;">`minutes.mean()`으로 평균을 구하며, 큰 값이 들어오면 평균이 올라갈 수 있습니다.</span>


```python
import numpy as np
import pandas as pd

minutes = np.array([20, 25, 30, 30, 45])

print("Study minutes:", minutes)
print("Sum:", minutes.sum())
print("Count:", minutes.size)
print("Mean:", minutes.mean())
```

- The five sessions total 150 minutes and have `mean` 30 minutes.  
  <span style="color: #777777; font-style: italic;">학습 시간 5개의 합은 150분이고 `평균`은 30분입니다.</span>

- The mean is expressed in the same unit as the observations.  
  <span style="color: #777777; font-style: italic;">평균의 단위는 원래 관측값과 같은 분입니다.</span>


## Median

<span style="color: #777777; font-style: italic;">중앙값</span>

- The `median` is the middle value after sorting.  
  <span style="color: #777777; font-style: italic;">`중앙값`(median)은 정렬했을 때 가운데 값입니다.</span>

- For an even count, use the mean of the two middle values; `np.median()` handles both cases.  
  <span style="color: #777777; font-style: italic;">개수가 짝수이면 가운데 두 값의 평균을 쓰며, `np.median()`이 두 경우를 모두 처리합니다.</span>


```python
even_minutes = np.append(minutes, 60)

print("Median (5 values):", np.median(minutes))
print("Median (6 values):", np.median(even_minutes))
```

- The five study times have middle value 30.  
  <span style="color: #777777; font-style: italic;">학습 시간 5개의 가운데 값은 30입니다.</span>

- After adding 60, the two middle values are 30 and 30, so the median stays 30.  
  <span style="color: #777777; font-style: italic;">60을 추가하면 가운데 두 값이 30과 30이므로 중앙값은 30입니다.</span>


## Mode

<span style="color: #777777; font-style: italic;">최빈값</span>

- The `mode` is the most common value; more than one value can share the highest count.  
  <span style="color: #777777; font-style: italic;">`최빈값`(mode)은 가장 자주 나오는 값이며, 같은 최고 빈도의 값이 여러 개일 수 있습니다.</span>

- `value_counts()` counts each value, and `.mode()` returns the most common values.  
  <span style="color: #777777; font-style: italic;">`value_counts()`는 값별 개수를 세고, `.mode()`는 가장 자주 나오는 값을 반환합니다.</span>


```python
session_series = pd.Series(minutes)

print(session_series.value_counts().sort_index())
print("Mode:", session_series.mode().tolist())
```

- 30 appears twice; every other value appears once, so the mode is 30.  
  <span style="color: #777777; font-style: italic;">30은 두 번, 다른 값은 한 번씩 나와 최빈값은 30입니다.</span>


## Effect of a Large Value

<span style="color: #777777; font-style: italic;">큰 값의 영향</span>

- Change the longest time from 45 to 180; keep the other values.  
  <span style="color: #777777; font-style: italic;">가장 긴 시간 45를 180으로 바꾸고 나머지 값은 유지합니다.</span>

- Compare the mean and median to see how the large value affects each one.  
  <span style="color: #777777; font-style: italic;">큰 값이 평균과 중앙값에 미치는 영향을 비교합니다.</span>


```python
extended_minutes = minutes.copy()
extended_minutes[-1] = 180

print("Original mean:", minutes.mean())
print("New mean:", extended_minutes.mean())
print("Original median:", np.median(minutes))
print("New median:", np.median(extended_minutes))
```

- The mean rises from 30 to 57, while the median stays 30.  
  <span style="color: #777777; font-style: italic;">평균은 30에서 57로 올라가지만 중앙값은 30입니다.</span>

- The mean changes more here; the long study time is not necessarily an error.  
  <span style="color: #777777; font-style: italic;">여기서는 평균이 더 크게 변하며, 긴 학습 시간이 반드시 오류인 것은 아닙니다.</span>


## Summary

<span style="color: #777777; font-style: italic;">핵심 정리</span>

- Mean uses every value; median uses the middle position; mode uses frequency.  
  <span style="color: #777777; font-style: italic;">평균은 모든 값, 중앙값은 가운데 위치, 최빈값은 빈도를 기준으로 합니다.</span>
