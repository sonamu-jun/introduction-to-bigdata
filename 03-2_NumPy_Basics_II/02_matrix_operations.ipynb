# Matrix Operations

*Multiply matrices with compatible shapes, transpose their axes, and verify an inverse numerically.*

A two-dimensional NumPy array can represent a matrix. Elementwise arithmetic still uses the usual operators; matrix multiplication uses `@`.


## Arithmetic at Matching Positions

Addition, subtraction, and `*` combine matching entries in these
equal-shaped matrices. `*` does not add across rows and columns.


```python
import numpy as np

matrix_a = np.array([[1, 2],
                     [3, 4]])
matrix_b = np.array([[10, 20],
                     [30, 40]])

print("Addition:")
print(matrix_a + matrix_b)
print("Subtraction:")
print(matrix_a - matrix_b)
print("Elementwise multiplication:")
print(matrix_a * matrix_b)
```

Elementwise multiplication returns `[[10, 40], [90, 160]]`. Each
output comes from one pair of positions, just as in array arithmetic.


## Rows Multiplied by Columns

For matrix multiplication, `(m, k) @ (k, n)` produces shape `(m, n)`.
The inner lengths must match. Each result is the dot product of a
row from the left matrix and a column from the right matrix.

**Question.** What is the top-left value of `matrix_a @ matrix_b`?


```python
matrix_product = matrix_a @ matrix_b

print("Matrix product:")
print(matrix_product)
print("Top-left calculation:", 1 * 10 + 2 * 30)
print("Reversed multiplication:")
print(matrix_b @ matrix_a)
```

The answer is 70. The full product is `[[70, 100], [150, 220]]`.
These particular inputs give the same reversed product because
`matrix_b` is ten times `matrix_a`; matrix multiplication does not
generally have that property.


## Matrix-Vector Shapes

Multiplying a `(2, 3)` matrix by a `(3,)` vector gives one value per
row, with shape `(2,)`. Giving the same values an explicit column
shape `(3, 1)` makes the output a column with shape `(2, 1)`.


```python
coefficients = np.array([[2, 1, 0],
                         [1, 3, 2]])
values = np.array([10, 20, 30])
value_column = values.reshape(3, 1)

vector_result = coefficients @ values
column_result = coefficients @ value_column

print("Coefficient shape:", coefficients.shape)
print("Vector result:", vector_result)
print("Vector result shape:", vector_result.shape)
print("Column result:")
print(column_result)
print("Column result shape:", column_result.shape)
```

Both forms calculate 40 and 130. The first result is a one-dimensional
vector; the second is a two-dimensional column. Shape affects how
the result can be combined with other arrays.


## Transposing Axes

For a two-dimensional array, `.T` exchanges the row and column axes.
A `(2, 3)` matrix becomes `(3, 2)`. A one-dimensional vector has no
separate row and column axes, so `.T` alone does not turn it into a column.


```python
print("Transposed coefficients:")
print(coefficients.T)
print("Transposed shape:", coefficients.T.shape)
print("One-dimensional transpose shape:", values.T.shape)
```

The transpose has rows `[2, 1]`, `[1, 3]`, and `[0, 2]`. The vector
still has shape `(3,)`; `reshape(3, 1)` was needed for the explicit column.


## Identity and Inverse

`np.eye(n)` creates an identity matrix: ones on the diagonal and
zeros elsewhere. Multiplication by a matching identity preserves a
matrix. A nonsingular square matrix has an inverse, calculated with
`np.linalg.inv()`.

Floating-point arithmetic can leave small numerical differences.
`np.allclose()` compares results within a tolerance instead of
requiring exact equality. An inverse check is useful here, but
`np.linalg.solve()` is the direct tool for solving equations.


```python
identity = np.eye(2)
inverse = np.linalg.inv(matrix_a)

print("Identity:")
print(identity)
print("Matrix times identity:")
print(matrix_a @ identity)
print("Inverse:")
print(np.round(inverse, 6))
print("Product is close to identity:", np.allclose(matrix_a @ inverse, identity))
```

The inverse is approximately `[[-2, 1], [1.5, -0.5]]`, and the
identity check is `True`. Singular matrices have no inverse; a nearly
singular matrix can also give inaccurate results, so this example
does not establish that every inverse is numerically reliable.


Track the inner dimensions when using `@`, distinguish a vector from a column, and verify floating-point results with an appropriate tolerance. [NumPy inverse](https://numpy.org/doc/2.3/reference/generated/numpy.linalg.inv.html)
