# Validating

<span style="color: #777777;"><em>검증</em></span>


## Records to Check

<span style="color: #777777;"><em>검사할 기록</em></span>

- `Validating` checks whether data follow the required rules.  
  <span style="color: #777777;"><em>`검증`(validating)은 데이터가 필요한 규칙에 맞는지 확인합니다.</em></span>

- Here scores must be from 0 to 100, and each subject needs a department.  
  <span style="color: #777777;"><em>여기서는 점수가 0–100 범위에 있어야 하고, 과목마다 소속 분야가 필요합니다.</em></span>


```python
import pandas as pd

records = pd.DataFrame({
    "subject": ["Mathematics", "English", "Science"],
    "score": [98, 90, 85],
    "department": ["STEM", "Languages", None],
})
print(records)
```

## Range and Missing Values

<span style="color: #777777;"><em>범위와 결측값</em></span>

- `.between(0, 100)` checks the allowed range; `.notna()` checks that a value is present.  
  <span style="color: #777777;"><em>`.between(0, 100)`은 허용 범위, `.notna()`는 값의 존재 여부를 확인합니다.</em></span>

- `.all()` returns `True` only when every row passes.  
  <span style="color: #777777;"><em>`.all()`은 모든 행이 통과할 때만 `True`를 반환합니다.</em></span>


```python
print("Scores in range:", records["score"].between(0, 100).all())
print("Departments present:", records["department"].notna().all())
```

- All scores are in 0–100, but Science has no department.  
  <span style="color: #777777;"><em>점수는 모두 0–100 범위에 있지만, Science의 소속 분야가 빠져 있습니다.</em></span>
