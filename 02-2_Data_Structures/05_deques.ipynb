# Deques

*Add and remove items at both ends of a double-ended queue.*

A **deque**, pronounced “deck,” is a double-ended queue. It supports
insertion and removal at either end. Restricting its operations can give
it stack behavior or queue behavior.


## Operations at Both Ends

`collections.deque` provides four end operations:

| End | Add an item | Remove and return an item |
| --- | --- | --- |
| Left | `appendleft(item)` | `popleft()` |
| Right | `append(item)` | `pop()` |

```text
              left                            right
appendleft -> [Grace] [Olivia] ... [Potter] [William] <- append
popleft    <-                                        -> pop
```

Each operation takes approximately `O(1)` time. Reaching positions near
the middle takes more work; a list is better suited to frequent arbitrary
indexed access. [Python 3.12 deque documentation](https://docs.python.org/3.12/library/collections.html#collections.deque)


````python
from collections import deque

waiting = deque()
waiting.append("Daniel")
waiting.appendleft("James")
waiting.appendleft("Benjamin")
waiting.append("Sophia")
waiting.appendleft("Olivia")
waiting.appendleft("Grace")
waiting.append("Potter")
waiting.append("William")

print("Left to right:", list(waiting))
````

The leftmost person is Grace, and the rightmost person is William.
New left-side arrivals appear before earlier left-side arrivals; right-side
arrivals are added after the existing rightmost person.


## Removing from Opposite Ends

The removal method determines which end supplies the next item.
Consecutive `popleft()` calls keep advancing the left end, while `pop()`
advances the right end.


**Question.** Who leaves after two `popleft()` calls and one `pop()` call, in that order?


````python
print("Removed from left:", waiting.popleft())
print("Removed from left:", waiting.popleft())
print("Removed from right:", waiting.pop())
print("Remaining:", list(waiting))
````

Grace, Olivia, and William leave in that order. The remaining sequence is
Benjamin, James, Daniel, Sophia, and Potter. Both ends changed while the
order of the middle entries stayed intact.
