# Printing Strings with Variables

*Begin with three ideas: a value is data, a variable gives that value a name, and `print()` displays a result.*


## Literal and Named Values

Text inside quotation marks is a **string literal**. In `course_name = "Introduction to Big Data"`, the equals sign performs assignment; it does not ask whether two values are equal.


```python
print("Hello, world!")

course_name = "Introduction to Big Data"
print(course_name)
```

The second `print()` reads the value currently stored in `course_name`. Quotation marks identify text in the code, but they are not part of the printed result.


## Value Combination

`print()` can receive several values separated by commas.


```python
language_name = "Python"
week_number = 1

print("Course:", course_name)
print("Language:", language_name)
print("Week:", week_number)
```

## Reassignment and Execution Order

A variable can receive a new value. Notebook cells therefore have a state: rerunning an earlier or later cell can change the value that the next cell sees.


```python
current_topic = "Python"
print("Before:", current_topic)

current_topic = "NumPy"
print("After:", current_topic)
```

The name stays the same while its stored value changes.
