# JSON Files

*Learn to load, save, and reopen structured JSON data with Python's `json` module.*


## Before You Run

Input: `assets/04_json_files/study_profile.json`

Course site: select `study_profile.json` in **Add files** and enter `assets/04_json_files` for **Save folder**. The saved path is `/home/pyodide/assets/04_json_files/study_profile.json`.


## Locating the Input File

Build the path to the supplied JSON before loading it.


```python
import json
import os

if os.path.isdir("01-3_Python_Basics_III"):
    os.chdir("01-3_Python_Basics_III")

input_directory = os.path.join("assets", "04_json_files")
input_path = os.path.join(input_directory, "study_profile.json")

print("Input path:", input_path)
```

`input_path` identifies the supplied JSON file.


## Loading a JSON File

`json.load()` reads JSON text from an open file and converts it to Python values.


```python
with open(input_path, mode="r", encoding="utf-8") as source_file:
    loaded_profile = json.load(source_file)

print("Loaded profile:", loaded_profile)
print("Python type:", type(loaded_profile).__name__)
```

The JSON object becomes a dictionary, its array becomes a list, `true` becomes `True`, and `null` becomes `None`.


## Saving a JSON File

Create the output directory, then use `json.dump()` to write the loaded value as readable JSON.


```python
output_directory = os.path.join("outputs", "04_json_files")
os.makedirs(output_directory, exist_ok=True)
output_path = os.path.join(output_directory, "study_profile_copy.json")

with open(output_path, mode="w", encoding="utf-8") as saved_file:
    json.dump(
        loaded_profile,
        saved_file,
        indent=2,
        ensure_ascii=False,
    )

print("Saved file:", output_path)
print("Saved size in bytes:", os.path.getsize(output_path))
```

The dictionary is stored as a new UTF-8 JSON file, and the supplied source remains unchanged.


## Reopening the Saved JSON

Load the saved file again and compare the restored value with the in-memory source value.


```python
with open(output_path, mode="r", encoding="utf-8") as saved_file:
    reloaded_profile = json.load(saved_file)

print("Saved data matches:", reloaded_profile == loaded_profile)
print("Student:", reloaded_profile["student"])
```

The reopened dictionary matches the value passed to `json.dump()`, confirming the complete save-and-load round trip.
