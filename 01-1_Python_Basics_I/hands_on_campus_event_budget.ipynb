# Campus Event Budget

*Calculate income, cost, and balance from named inputs and print an itemized report.*


## Problem Definition

### Question

Calculate ticket income, refreshment cost, total cost, and the remaining budget.

### Data

| Variable | Value |
|---|---:|
| Attendees | 120 |
| Fee per attendee | 7,200 KRW |
| Refreshment per attendee | 2,600 KRW |
| Room cost | 500,000 KRW |
| Budget | 1,000,000 KRW |


## Case Data


## Named Budget Inputs


```python
event_name = "Data Night"
attendees = 48
fee_per_attendee_krw = 18_000
room_cost_krw = 500_000
refreshment_per_person_krw = 6_500

print("Event:", event_name)
print("Expected attendees:", attendees)
print("Fee per attendee (KRW):", fee_per_attendee_krw)
```

## Income and Cost Arithmetic


```python
ticket_income_krw = attendees * fee_per_attendee_krw
refreshment_cost_krw = attendees * refreshment_per_person_krw
total_cost_krw = room_cost_krw + refreshment_cost_krw
remaining_budget_krw = ticket_income_krw - total_cost_krw

print("Ticket income (KRW):", ticket_income_krw)
print("Refreshment cost (KRW):", refreshment_cost_krw)
print("Total cost (KRW):", total_cost_krw)
print("Remaining budget (KRW):", remaining_budget_krw)
```

## Labeled Budget Balance


```python
report = (
    f"{event_name} | attendees={attendees} | "
    f"income_krw={ticket_income_krw} | "
    f"cost_krw={total_cost_krw} | "
    f"balance_krw={remaining_budget_krw}"
)
print(report)
```

## Result Analysis

The event earns 864,000 KRW, costs 812,000 KRW, and leaves 188,000 KRW.
