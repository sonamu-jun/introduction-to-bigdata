# Text Files

*Learn to read, write, and verify UTF-8 text with `open()`.*


## Before You Run

In **Add files**, select `study_plan.txt` and enter `assets/02_text_files` for **Save folder**. The file will be available at `/home/pyodide/assets/02_text_files/study_plan.txt`.


## Locate and Read

Open the supplied file with mode `"r"`. A `with` block closes it automatically.


```python
import os


jupyterlite_base_directory = "/home/pyodide"
chapter_name = "01-3_Python_Basics_III"

if os.path.isdir(jupyterlite_base_directory):
    os.chdir(jupyterlite_base_directory)
elif os.path.basename(os.getcwd()) != chapter_name:
    os.chdir(chapter_name)

base_directory = os.getcwd()

input_directory = os.path.join(base_directory, "assets", "02_text_files")
output_directory = os.path.join(base_directory, "outputs", "02_text_files")
input_path = os.path.join(input_directory, "study_plan.txt")
output_path = os.path.join(output_directory, "study_plan_updated.txt")

os.makedirs(output_directory, exist_ok=True)

with open(input_path, mode="r", encoding="utf-8") as source_file:
    loaded_text = source_file.read()

print(loaded_text)
```

`read()` returns the file contents as one string.


## Write and Verify

Mode `"w"` creates or replaces the output. Reopen it with `"r"` to check the result.


```python
updated_text = loaded_text + "Thursday: Review\n"

with open(output_path, mode="w", encoding="utf-8") as saved_file:
    saved_file.write(updated_text)

with open(output_path, mode="r", encoding="utf-8") as saved_file:
    reloaded_text = saved_file.read()

print(reloaded_text)
print("Saved correctly:", reloaded_text == updated_text)
```

## Takeaway

Use `open()`, the correct mode, UTF-8 encoding, and a `with` block for text files.
