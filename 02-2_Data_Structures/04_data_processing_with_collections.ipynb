# Data Processing with Collections

*Use nested collections to retain records, identify observations, and group values.*

A **record** groups fields for one observation. A list of dictionaries preserves row order and provides field-name access. Here, each row represents one student on one day.

Select a row first, then a named field.


```python
study_records = [
    {"student_id": "S01", "day": "Mon", "minutes": 30, "tasks": 2, "score": 70},
    {"student_id": "S02", "day": "Mon", "minutes": 45, "tasks": 3, "score": 80},
    {"student_id": "S01", "day": "Tue", "minutes": 40, "tasks": 4, "score": 75},
]

print("Record count:", len(study_records))
print("First record:", study_records[0])
print("Minutes in the third record:", study_records[2]["minutes"])
```

## Keys and Groups

A dictionary keeps one current value per key. Student ID alone repeats here; the tuple `(student_id, day)` distinguishes all three observations. If several sessions per student-day were allowed, we would need an additional identifier or a list per key.

A dictionary of lists groups several durations under one student. A comprehension builds the record lookup; the loop initializes a list for a new student and appends each duration. `sum()` totals a group's values.


**Think before running or viewing the saved output.** Why would student ID alone lose a record in a lookup? What durations and total should S01 retain after grouping?


```python
records_by_key = {
    (record["student_id"], record["day"]): record for record in study_records
}
print("Distinct student-day keys:", len(records_by_key))
print("S01 Monday score:", records_by_key[("S01", "Mon")]["score"])

minutes_by_student = {}
for record in study_records:
    student_id = record["student_id"]
    if student_id not in minutes_by_student:
        minutes_by_student[student_id] = []
    minutes_by_student[student_id].append(record["minutes"])

for student_id in sorted(minutes_by_student):
    values = minutes_by_student[student_id]
    print(student_id, "durations:", values, "total:", sum(values))
```

**Check:** An ID-only lookup would replace S01's earlier record. The tuple keys retain all three. S01 keeps `[30, 40]`, totaling `70`; S02 keeps `[45]`, totaling `45`. The inner lists retain every contribution.


Identify what one observation represents before choosing its key. Keep a list per group when multiple values must contribute to a result.
