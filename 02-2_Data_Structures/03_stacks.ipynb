# Stacks

*Use last-in, first-out access to reverse the order of stored items.*

A **stack** adds and removes items at one end, called the **top**.
**LIFO**, or last in, first out, means the most recently added item is
removed first. An undo history follows the same rule.


## Push and Pop

`push()` puts an item on top; `pop()` removes and returns the top item.
The class uses the end of a Python list as the top: `append()` and `pop()`
already provide these operations. A push takes `O(1)` time on average
over many pushes; removing the last item takes `O(1)`.

```text
          push      pop
            |        ^
            v        |
        +---------------+
top ->  |       4       |
        +---------------+
        |       3       |
        +---------------+
        |       2       |
        +---------------+
        |       1       |
        +---------------+
```


````python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()


stack = Stack()
for number in [1, 2, 3, 4]:
    stack.push(number)

print("Bottom to top:", stack.items)
print("Removed top:", stack.pop())
print("Remaining:", stack.items)
````

`4` leaves first even though `1` arrived first. The remaining stack is
`[1, 2, 3]`, with `3` now on top.


## Reversing a Sequence

Pushing characters from left to right puts the final character on top.
Repeated pops then retrieve the characters in reverse order. The `while`
condition stops when the stack is empty, and `"".join()` combines the
characters into one string.


**Question.** If `Have a nice day` is pushed character by character, what string will repeated pops produce?


````python
text = "Have a nice day"
character_stack = Stack()
for character in text:
    character_stack.push(character)

reversed_characters = []
while character_stack.items:
    reversed_characters.append(character_stack.pop())

print("Original:", text)
print("Reversed:", "".join(reversed_characters))
print("Remaining characters:", len(character_stack.items))
````

The result is `yad ecin a evaH`, including the reversed positions of spaces.
Every character has been removed, leaving an empty stack. Calling `pop()`
again would raise `IndexError`; the loop avoids removing from an empty stack.
