# Modules and Program Structure

*Reuse library code through imports and give a small script one clear entry point.*

A module is a Python file that provides names such as functions, classes, and constants. Imports prevent every program from reimplementing common tools.

## Import Styles

`import module` keeps the module name visible. `from module import name` imports selected names directly, and `as` creates a short local alias.

```python
import math
import statistics as stats
from pathlib import Path

print("Square root:", math.sqrt(81))
print("Mean:", stats.mean([72, 85, 91]))
print("File suffix:", Path("survey.csv").suffix)
```

The qualified forms `math.sqrt()` and `stats.mean()` show where each function came from. The selected `Path` class is available directly.

## Helper Functions and `main()`

Helper functions handle separate jobs. A `main()` function states the program's top-level workflow instead of leaving many unrelated statements at file scope.

```python
def summarize_scores(scores):
    return {
        "count": len(scores),
        "mean": round(stats.mean(scores), 1),
        "minimum": min(scores),
        "maximum": max(scores),
    }

def main():
    scores = [72, 85, 91, 68]
    report = summarize_scores(scores)
    print("Program report:", report)

main()
```

`main()` coordinates the program while `summarize_scores()` owns the calculation. This separation makes each part easier to reuse and test.

## The Main Guard

In a `.py` file, `__name__` equals `"__main__"` when the file is run directly. When another file imports it, `__name__` contains the module name instead.

```python
def guarded_main():
    print("The guarded program is running")

if __name__ == "__main__":
    guarded_main()
```

The notebook kernel uses `__main__`, so the function runs here. In a reusable script, this guard prevents the main workflow from running merely because the file was imported.

## Use Cases

Imports and program structure commonly support mathematical calculations, descriptive summaries, portable path handling, and reusable script entry points.

### Mathematical Utilities

The `math` module supplies tested mathematical functions and constants.

```python
radius = 3
circle_area = math.pi * radius ** 2
required_boxes = math.ceil(47 / 10)

print("Circle area:", round(circle_area, 2))
print("Boxes required:", required_boxes)
```

`math.pi` supplies the constant, and `math.ceil()` rounds upward when a partial box still requires a whole box.

### Descriptive Statistics

The standard `statistics` module provides readable summary functions for small numeric collections.

```python
study_hours = [1.5, 2.0, 1.0, 2.5, 2.0]

print("Mean hours:", stats.mean(study_hours))
print("Median hours:", stats.median(study_hours))
```

The mean uses every value, while the median selects the middle of the ordered values. Importing the module avoids rewriting these calculations.

### Portable Path Inspection

`Path` represents a file-system path without manual slash handling.

```python
data_path = Path("data") / "survey.csv"

print("Path:", data_path)
print("Filename:", data_path.name)
print("Stem:", data_path.stem)
print("Suffix:", data_path.suffix)
```

The `/` operator joins path parts, and the properties expose the filename components. Opening and saving files belongs in the later file-handling lesson.

### Reusable Script Modules

A script can expose helper functions to imports while reserving its full workflow for direct execution.

```python
def format_summary(label, values):
    return f"{label}: n={len(values)}, mean={stats.mean(values):.1f}"

def report_main():
    print(format_summary("Scores", [72, 85, 91]))

if __name__ == "__main__":
    report_main()
```

Another module could import and call `format_summary()` without intentionally running `report_main()`. The main guard separates library behavior from script behavior.

Prefer standard-library tools over reimplementing common behavior, keep imports near the top of a script, divide work into focused functions, and use a main guard when one file should support both importing and direct execution.
