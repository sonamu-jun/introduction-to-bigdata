# Conditionals

*Use Boolean conditions to select exactly the code branch that matches the current observation.*

## Two-Way Branching

The condition after `if` is evaluated first. `else` handles every remaining case.


```python
score = 82

if score >= 60:
    result = "pass"
else:
    result = "fail"

print(result)
```

## Ordered Multiway Branching

The most specific high threshold belongs first. Once a true branch runs, Python skips the rest of the chain.


```python
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D or F"

print("Grade:", grade)
```

## Validation Before Classification

Validate the input range before applying the classification rule.


```python
if not 0 <= score <= 100:
    print("Invalid score")
elif score >= 60:
    print("Valid passing score")
else:
    print("Valid non-passing score")
```

## Truthiness in Conditions

A condition may use a Boolean expression directly. Empty strings and collections are false; non-empty ones are true.

```python
selected_rows = []

if selected_rows:
    selection_message = "Rows are ready"
else:
    selection_message = "No rows selected"

print(selection_message)
```

The empty list is false in a Boolean context, so the `else` branch runs. This form is clearer than `if len(selected_rows) > 0:` for a simple presence check.

## Conditional Expressions

A conditional expression chooses between two short values inside one expression. Its order is `value_if_true if condition else value_if_false`.

```python
row_count = 120
dataset_status = "ready" if row_count > 0 else "empty"

print("Dataset status:", dataset_status)
```

The condition is true, so `dataset_status` receives `ready`. Use an ordinary `if` statement when either branch needs several operations.

## Use Cases

Conditionals commonly validate data, handle missing values, and select behavior from a setting.

### Validate a Data Range

A measurement outside a physically reasonable range should be marked before it is analyzed.

```python
temperature_c = 150

if temperature_c < -90 or temperature_c > 60:
    temperature_status = "invalid"
else:
    temperature_status = "valid"

print("Temperature status:", temperature_status)
```

The `or` condition marks the value invalid when either boundary is violated. Valid values reach the `else` branch.

### Handle a Missing Value

`None` represents the absence of a value. `is None` checks that state before arithmetic or formatting is attempted.

```python
body_mass_g = None

if body_mass_g is None:
    body_mass_display = "not recorded"
else:
    body_mass_display = f"{body_mass_g} g"

print("Body mass:", body_mass_display)
```

The missing value takes the first branch, so the program does not try to use `None` as a numeric measurement.

### Select Behavior from a Setting

A configuration value can choose one of several output behaviors.

```python
export_format = "csv"

if export_format == "csv":
    export_message = "Create a comma-separated file"
elif export_format == "json":
    export_message = "Create a JSON file"
else:
    export_message = "Unsupported export format"

print(export_message)
```

Exactly one branch runs. Equality uses `==`; a single `=` would assign a value rather than compare it.

### Access Control

Ordered branches can explain why access is denied instead of returning only one combined Boolean.

```python
is_signed_in = True
has_download_permission = False

if not is_signed_in:
    access_message = "Sign in first"
elif not has_download_permission:
    access_message = "Download permission required"
else:
    access_message = "Download allowed"

print(access_message)
```

The first condition is false, so Python checks the permission branch next. That branch explains the specific unmet requirement.

### Fallback Display Values

A conditional can replace empty text with a readable fallback before displaying it.

```python
raw_title = "   "
cleaned_title = raw_title.strip()

if cleaned_title == "":
    display_title = "Untitled"
else:
    display_title = cleaned_title

print("Display title:", display_title)
```

Removing spaces leaves an empty string, so the first branch supplies `Untitled`. Non-empty text would be preserved by the `else` branch.

Branch order is part of the rule. Validate the allowed range before interpreting a value, and use `==` only when testing equality.
