# Loops

*Repeat one operation across a sequence and inspect the loop body and state after each iteration.*

## Value Iteration

On each iteration, `score` receives the next list value. Indentation marks the repeated block.


```python
scores = [72, 85, 91, 68]

for score in scores:
    print("Observed score:", score)
```

## Result Accumulation and Augmented Assignment

Initialize an accumulator before the loop and update it once per value. `total += score` is the common short form of `total = total + score`.

```python
total = 0
for score in scores:
    total += score

print("Total:", total)
print("Mean:", total / len(scores))
```

## Position Tracking and Paired Iteration

First, `enumerate()` supplies a position for one sequence. Then, in a separate loop, `zip()` pairs values that occupy the same position in two sequences. Each produces a two-value pair. Writing two names after `for` unpacks that pair from left to right: `row_number, name` and `name, score`.


```python
names = ["A", "B", "C", "D"]

for row_number, name in enumerate(names, start=1):
    print(row_number, name)

print()
for name, score in zip(names, scores):
    print(name, score)
```

## Counting with `range()`

`range(start, stop)` produces integers from the start value up to, but not including, the stop value. A `for` loop is appropriate when the number of repetitions is known.

```python
for attempt in range(1, 4):
    print("Attempt:", attempt)
```

## Condition-Controlled Repetition with `while`

A `while` loop repeats while its condition remains true. The loop body must change a value that eventually makes the condition false.

```python
remaining_attempts = 3

while remaining_attempts > 0:
    print("Attempts remaining:", remaining_attempts)
    remaining_attempts -= 1

print("Attempts finished")
```

`-=` decreases the counter on every iteration. Without that update, the condition would remain true and the loop would not stop.

## Iterating over Dictionary Pairs

`items()` supplies each dictionary key and value as a pair. Two loop variables unpack that pair.

```python
hours_by_subject = {"Python": 4.5, "Statistics": 2.0, "Data": 2.5}

for subject, hours in hours_by_subject.items():
    print(subject, "->", hours, "hours")
```

The loop reads the labeled key and its value together. Use `keys()` or direct dictionary iteration when only keys are needed, and `values()` when only values are needed.

## Nested Loops

A nested loop processes items inside grouped collections. The inner loop completes for each outer-loop item.

```python
weekly_scores = [[72, 85], [91, 68]]

for week_number, week_scores in enumerate(weekly_scores, start=1):
    print("Week", week_number)
    for score in week_scores:
        print("  Score:", score)
```

The outer loop selects one week, and the inner loop prints every score in that week. Keep nesting shallow so the control flow remains readable.

## Use Cases

Loops commonly transform every value, filter unwanted values, and search until a match is found.

### Transform Every Value

The output list begins empty. Each loop iteration calculates one tax-adjusted price and appends it.

```python
prices = [12000, 18000, 25000]
prices_with_tax = []

for price in prices:
    adjusted_price = round(price * 1.1)
    prices_with_tax.append(adjusted_price)

print("Original prices:", prices)
print("Prices with tax:", prices_with_tax)
```

The loop produces one output for every input while leaving the source list unchanged. A later comprehension can express this simple pattern more compactly.

### Filter Invalid Values

`continue` skips the remainder of the current iteration and moves to the next value.

```python
raw_scores = [72, -1, 85, 101, 91]
valid_scores = []

for score in raw_scores:
    if score < 0 or score > 100:
        continue
    valid_scores.append(score)

print("Raw scores:", raw_scores)
print("Valid scores:", valid_scores)
```

The invalid values `-1` and `101` never reach `append()`. The loop continues processing the remaining observations.

### Stop After the First Match

`break` exits the loop immediately after the required record is found.

```python
study_records = [
    {"day": "Mon", "completed": True},
    {"day": "Tue", "completed": True},
    {"day": "Wed", "completed": False},
    {"day": "Thu", "completed": False},
]

first_incomplete_day = None
for record in study_records:
    if not record["completed"]:
        first_incomplete_day = record["day"]
        break

print("First incomplete day:", first_incomplete_day)
```

The search stops at Wednesday instead of inspecting later records. `break` is useful when only the first match is needed.

### Category Frequency Counting

A loop can update one dictionary count for every observed category.

```python
species_observations = ["Adelie", "Gentoo", "Adelie", "Chinstrap", "Adelie"]
species_counts = {}

for species in species_observations:
    species_counts[species] = species_counts.get(species, 0) + 1

print("Species counts:", species_counts)
```

`get(species, 0)` supplies zero for a first observation and the stored count afterward. Each iteration increases exactly one category.

### Grouping Values by Category

A loop can build a dictionary whose values are lists of observations belonging to each category.

```python
study_log = [
    {"subject": "Python", "hours": 1.5},
    {"subject": "Statistics", "hours": 2.0},
    {"subject": "Python", "hours": 1.0},
]
hours_by_subject = {}

for record in study_log:
    subject = record["subject"]
    hours_by_subject.setdefault(subject, [])
    hours_by_subject[subject].append(record["hours"])

print("Grouped hours:", hours_by_subject)
```

`setdefault()` creates an empty list only for a new subject. Each record's hours are then appended to the correct group.

Keep the loop body short and name the changing value clearly. If there is already a built-in operation such as `sum(scores)`, prefer it in ordinary analysis and use the explicit loop when learning or implementing custom logic.
