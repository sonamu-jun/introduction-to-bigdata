## Keys and Missing Matches

<span style="color: #777777; font-style: italic;">키와 연결 누락</span>

- `Validating` checks whether the result follows the rules.  
  <span style="color: #777777; font-style: italic;">`검증`(validating)은 결과가 규칙에 맞는지 확인하는 작업입니다.</span>

- `.duplicated()` marks repeated subjects; `.any()` checks whether at least one is repeated.  
  <span style="color: #777777; font-style: italic;">`.duplicated()`는 반복된 과목을 표시하고, `.any()`는 그런 값이 하나라도 있는지 확인합니다.</span>

- A `lookup table` holds information to match by key; no duplicate subjects does not mean every needed subject is present.  
  <span style="color: #777777; font-style: italic;">`조회 표`(lookup table)는 키로 연결할 정보를 담으며, 과목 중복이 없어도 필요한 과목이 빠질 수 있습니다.</span>


```python
import pandas as pd

scores_df = pd.DataFrame({
    "name": ["Jiyoung", "Hojun", "Hana"],
    "subject": ["Mathematics", "English", "Science"],
    "score": [98, 90, 85],
    "maximum_score": [100, 100, 100],
})
lookup_df = pd.DataFrame({
    "subject": ["Mathematics", "English"],
    "department": ["STEM", "Languages"],
})

unique_subjects = not lookup_df["subject"].duplicated().any()
enriched_df = scores_df.merge(lookup_df, on="subject", how="left")

print("Unique subjects:", unique_subjects)
print(enriched_df)
```

- `Unique subjects` is `True`; all three score rows remain.  
  <span style="color: #777777; font-style: italic;">`Unique subjects`는 `True`이며 점수 행 3개가 모두 남습니다.</span>

- Science has no department, so Hana’s department is `NaN`.  
  <span style="color: #777777; font-style: italic;">Science의 소속 분야가 없어 Hana의 소속 분야가 `NaN`으로 표시됩니다.</span>


## Score Rules

<span style="color: #777777; font-style: italic;">점수 규칙</span>

- A score must be at least zero and no greater than its maximum.  
  <span style="color: #777777; font-style: italic;">점수는 0 이상이고 해당 만점 이하여야 합니다.</span>

- Use `&` to combine the two conditions, and `.all()` to check all rows.  
  <span style="color: #777777; font-style: italic;">`&`로 두 조건을 묶고 `.all()`로 모든 행이 통과하는지 확인합니다.</span>

- `.notna()` checks whether each department is present.  
  <span style="color: #777777; font-style: italic;">`.notna()`는 각 소속 분야가 채워져 있는지 확인합니다.</span>


```python
scores = enriched_df["score"]
maximums = enriched_df["maximum_score"]
valid_scores = (scores >= 0) & (scores <= maximums)
matched = enriched_df["department"].notna()

print("Positive maximums:", (maximums > 0).all())
print("Valid scores:", valid_scores.all())
print("Complete lookup:", matched.all())
```

- The maximums are positive and all scores are in `range`.  
  <span style="color: #777777; font-style: italic;">만점은 양수이고 점수는 모두 허용 `범위` 안에 있습니다.</span>

- Complete lookup is `False` because Science has no department.  
  <span style="color: #777777; font-style: italic;">Science의 소속 분야가 없어 전체 연결 여부는 `False`입니다.</span>


- A check can pass while another fails.  
  <span style="color: #777777; font-style: italic;">한 검사를 통과해도 다른 검사는 실패할 수 있습니다.</span>

- Here the scores are valid, but Science still needs a department.  
  <span style="color: #777777; font-style: italic;">여기서는 점수 규칙은 맞지만 Science의 소속 분야가 필요합니다.</span>
