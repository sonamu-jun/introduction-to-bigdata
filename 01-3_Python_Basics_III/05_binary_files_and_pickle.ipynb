# Binary Files with Pickle

*Learn to save and reload one trusted Python object with `pickle`.*

Only load pickle files that your program created and controls. An untrusted pickle can run harmful code.


## Prepare the Output

Pickle uses binary modes: `"wb"` for writing and `"rb"` for reading.


```python
import os
import pickle


jupyterlite_base_directory = "/home/pyodide"
chapter_name = "01-3_Python_Basics_III"

if os.path.isdir(jupyterlite_base_directory):
    os.chdir(jupyterlite_base_directory)
elif os.path.basename(os.getcwd()) != chapter_name:
    os.chdir(chapter_name)

base_directory = os.getcwd()

output_directory = os.path.join(
    base_directory,
    "outputs",
    "05_binary_files_and_pickle",
)
output_path = os.path.join(output_directory, "study_snapshot.pkl")
os.makedirs(output_directory, exist_ok=True)

study_snapshot = {"student": "Amina", "minutes": 60}
print(study_snapshot)
```

## Save and Reload

`pickle.dump()` writes the object, and `pickle.load()` restores it.


```python
with open(output_path, mode="wb") as output_file:
    pickle.dump(study_snapshot, output_file)

# Safe because this notebook created the file.
with open(output_path, mode="rb") as input_file:
    reloaded_snapshot = pickle.load(input_file)

print("Reloaded:", reloaded_snapshot)
print("Saved correctly:", reloaded_snapshot == study_snapshot)
```

## Takeaway

Use `pickle` only for trusted, program-owned Python data.
