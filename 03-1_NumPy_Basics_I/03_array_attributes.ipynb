# Array Attributes and Data Types

*Read an array's shape, dimension count, element count, and data type before calculating with it.*


## Shape, Dimensions, and Size

An **axis** is one numbered direction of an array. A two-dimensional
table uses axis 0 for row positions and axis 1 for column positions.

| Attribute | Meaning |
|---|---|
| `shape` | Length of every axis, as a tuple |
| `ndim` | Number of axes |
| `size` | Total number of elements |
| `dtype` | Type used to store each element |

**Question.** For a table with shape `(2, 3)`, what are `ndim` and `size`?


```python
import numpy as np

measurements = np.array([10, 20, 30, 40])
table = np.array([[1, 2, 3],
                  [4, 5, 6]])

print("Measurements shape:", measurements.shape)
print("Measurements dimensions:", measurements.ndim)
print("Measurements size:", measurements.size)
print("Table shape:", table.shape)
print("Table dimensions:", table.ndim)
print("Table size:", table.size)
print("Table data type:", table.dtype)
```

The table has `ndim = 2` and `size = 6`: two axes with lengths 2 and 3.
The vector's shape is `(4,)`, a tuple with one axis length; it is not
a four-row, one-column table.


## One Common Numeric Type

A numeric array uses one storage type for every element. NumPy infers
a compatible type from the input values, or you can request a type
with `dtype`. `np.int64` names a 64-bit integer type explicitly.


```python
integer_values = np.array([10, 20, 30, 40], dtype=np.int64)
mixed_numbers = np.array([10, 20.5, 30, 40])

print("Explicit integer type:", integer_values.dtype)
print("Mixed numeric values:", mixed_numbers)
print("Inferred common type:", mixed_numbers.dtype)
```

`integer_values` has type `int64`. Including 20.5 makes the second
array use a floating-point type for all four values, including 10,
30, and 40.


## Explicit Type Conversion

`astype()` converts values into a new array. Convert integer data to
floating point when fractional results must be stored. Converting
floating-point values to integers discards their fractional parts;
it does not round to the nearest integer.


```python
decimal_values = integer_values.astype(float)
decimal_values[0] = 10.5
whole_values = decimal_values.astype(np.int64)

print("Decimal values:", decimal_values)
print("Converted back to integers:", whole_values)
print("Original integer values:", integer_values)
```

The floating-point copy stores 10.5, while conversion back to integers
gives 10. The original integer array still contains `[10, 20, 30, 40]`.


Read `shape` and `dtype` before choosing an operation. `ndim` counts axes; `size` counts values. Choose a type that preserves the information you need.
