# Binary Files with Pickle

*Learn to save and reload a trusted Python object as a binary pickle file.*

Pickle preserves Python data types in a binary format. Unlike text, CSV, or JSON, a pickle is intended for Python programs rather than direct human reading.


## Before You Run

No uploaded file is required. This notebook creates its own trusted pickle in `outputs/05_binary_files_and_pickle` and reopens the saved file for verification.


## Pickle Safety

Never load a pickle from an unknown or untrusted source. Unpickling can execute malicious code. This lesson reloads only the program-owned file created by the same notebook.


## Setting the Base and Output Directories

Set `/home/pyodide` as the course-site working directory. The pickle uses a visible, program-owned output folder because the notebook creates both the original Python object and the saved file.


```python
import os
import pickle

jupyterlite_base_directory = "/home/pyodide"
chapter_name = "01-3_Python_Basics_III"
launch_directory = os.getcwd()

if os.path.isdir(jupyterlite_base_directory):
    base_directory = jupyterlite_base_directory
elif os.path.basename(launch_directory) == chapter_name:
    base_directory = launch_directory
else:
    base_directory = os.path.join(launch_directory, chapter_name)

if not os.path.isdir(base_directory):
    raise FileNotFoundError(f"Base directory not found: {base_directory}")

os.chdir(base_directory)

output_directory = os.path.join(
    base_directory,
    "outputs",
    "05_binary_files_and_pickle",
)
os.makedirs(output_directory, exist_ok=True)
output_path = os.path.join(output_directory, "study_snapshot.pkl")

study_snapshot = {
    "student": "Amina",
    "week": 3,
    "topics": ["Python", "CSV", "JSON"],
    "minutes": 180,
    "completed": True,
}

print("JupyterLite base:", jupyterlite_base_directory)
print("Current directory name:", os.path.basename(os.getcwd()))
print("Output folder:", os.path.relpath(output_directory, base_directory))
print("Output directory exists:", os.path.isdir(output_directory))
print("Pickle file name:", os.path.basename(output_path))
```

The output path ends in `.pkl`, a common pickle extension. The file remains visible in the lesson output folder and is replaced on the next full run.


## Saving a Python Object

Open the output in binary write mode (`"wb"`) and pass both the object and the open file to `pickle.dump()`.


```python
with open(output_path, mode="wb") as output_file:
    pickle.dump(study_snapshot, output_file)

print("Pickle saved:", os.path.isfile(output_path))
print("Saved size in bytes:", os.path.getsize(output_path))
```

`pickle.dump()` performs the binary serialization and the `with` block closes the file after writing.


## Loading the Trusted Pickle

Open the same program-owned file in binary read mode (`"rb"`) and restore the object with `pickle.load()`.


```python
# Safe here because this notebook created the pickle itself.
with open(output_path, mode="rb") as input_file:
    reloaded_snapshot = pickle.load(input_file)

print("Reloaded object:", reloaded_snapshot)
print("Python type:", type(reloaded_snapshot).__name__)
print("Saved data matches:", reloaded_snapshot == study_snapshot)
```

The dictionary and its nested Python values are restored with their original types, and the round-trip comparison is `True`.


## Handling a Missing Pickle

A missing pickle raises the same `FileNotFoundError` as other file types. Catching that specific exception keeps the example runnable.


```python
missing_path = os.path.join(output_directory, "missing_snapshot.pkl")

try:
    with open(missing_path, mode="rb") as input_file:
        pickle.load(input_file)
except FileNotFoundError:
    print("Pickle file not found:", os.path.basename(missing_path))
```

## Use Cases

Pickle can store trusted Python-only snapshots, cached intermediate results, or objects whose Python types would be inconvenient to represent in JSON.


Create a program-owned output path, save with `pickle.dump()` in `"wb"` mode, reload with `pickle.load()` in `"rb"` mode, verify equality, and never unpickle untrusted data.
