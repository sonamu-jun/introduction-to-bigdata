## Inspect the Input

<span style="color: #777777; font-style: italic;">입력 데이터 확인</span>

- `Discovering` means checking what arrived before changing it.  
  <span style="color: #777777; font-style: italic;">`현황 파악`(discovering)은 수정 전에 어떤 자료가 들어왔는지 살피는 단계입니다.</span>

- A `DataFrame` is a table with rows and columns; check its fields and source labels.  
  <span style="color: #777777; font-style: italic;">`데이터프레임`(DataFrame)은 행과 열로 된 표이며, 항목과 출처를 살펴봅니다.</span>

- `.isna()` marks missing cells; `.sum()` counts them in each column.  
  <span style="color: #777777; font-style: italic;">`.isna()`는 결측값을 표시하고, `.sum()`은 열마다 그 개수를 셉니다.</span>


```python
import pandas as pd

records = [
    {"age": 25, "score": "85", "source": "db"},
    {"age": 30, "score": 90, "source": "api"},
    {"age": None, "score": 78, "source": "csv"},
    {"age": 22, "score": "eighty", "source": "web"},
    {"age": 400, "score": 88, "source": "csv"},
]
data = pd.DataFrame(records)

print(data)
print("Missing:")
print(data.isna().sum())
```

- There are five records from four sources; one age is missing.  
  <span style="color: #777777; font-style: italic;">네 출처에서 온 기록 5개 중 나이가 빠진 기록은 1개입니다.</span>

- `eighty` is text and age 400 is outside the example’s allowed range of 0–120.  
  <span style="color: #777777; font-style: italic;">`eighty`는 문자열이고, 나이 400은 이 예제의 허용 범위 0–120을 벗어납니다.</span>

- These two values are present, so a missing-value count does not flag them.  
  <span style="color: #777777; font-style: italic;">두 값은 빈칸이 아니므로 결측값 개수만으로는 찾을 수 없습니다.</span>


- Missing values, wrong formats, and range errors need different checks.  
  <span style="color: #777777; font-style: italic;">결측값, 형식 오류, 범위 오류는 서로 다른 검사가 필요합니다.</span>
