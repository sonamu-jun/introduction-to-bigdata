# Tuples

*Use a tuple for an ordered collection whose structure should remain fixed.*

A tuple resembles a list because it supports positions, duplicates, indexing, and slicing. Its defining property is **immutability**: items cannot be added, removed, or reassigned after the tuple is created.

## Creation, Indexing, and Slicing

Commas create tuple items, and parentheses make the boundary easy to see. A one-item tuple needs a trailing comma; without it, Python treats the parentheses as ordinary grouping.

```python
observation = ("Adelie", 39.1, 3750)
one_item = ("Adelie",)

print("Complete observation:", observation)
print("Species:", observation[0])
print("Measurements:", observation[1:])
print("One-item tuple:", one_item)
print("Number of fields:", len(observation))
```

Tuple indexing and slicing follow the same position rules as lists. The slice `observation[1:]` returns a new tuple containing the measurement fields.

## Tuple Methods

Because tuples cannot change, they have only two commonly used search methods. `count(value)` reports the number of matches, and `index(value)` reports the first matching position.

```python
temperatures = (21.5, 23.0, 21.5, 24.2)

print("21.5 count:", temperatures.count(21.5))
print("First 23.0 position:", temperatures.index(23.0))
```

Repeated values are preserved, so `count(21.5)` returns `2`. `index()` returns only the first matching position and raises `ValueError` when no match exists.

## Positional Unpacking

Unpacking assigns tuple items to names from left to right. The number of names must match the number of tuple items.

```python
species, bill_length_mm, body_mass_g = observation

print("Species:", species)
print("Bill length (mm):", bill_length_mm)
print("Body mass (g):", body_mass_g)
```

The three names document the meaning of the three fixed positions and remove repeated index notation from later code.

## Replacing a Tuple with a New Tuple

An expression can build a revised tuple, but it cannot edit the original one. Here, slicing keeps the unchanged fields and a one-item tuple supplies the replacement value.

```python
revised_observation = observation[:1] + (40.3,) + observation[2:]

print("Original observation:", observation)
print("Revised observation:", revised_observation)
```

`observation` remains unchanged because tuple operations produce new tuples. This fixed structure is useful for coordinates, compact records, and multiple values returned together.

## Converting for Several Changes

For several edits, `list()` creates a mutable list and `tuple()` converts the finished result back to a tuple.

```python
editable_observation = list(observation)
editable_observation[1] = 40.3
editable_observation[2] = 3800
converted_observation = tuple(editable_observation)

print("Converted observation:", converted_observation)
print("Type:", type(converted_observation))
```

## Starred Unpacking

When the number of middle items may vary, a starred name collects them in a list while the surrounding names receive fixed positions.

```python
pipeline_steps = ("load", "clean", "summarize", "save")
first_step, *middle_steps, last_step = pipeline_steps

print("First step:", first_step)
print("Middle steps:", middle_steps)
print("Last step:", last_step)
```

The first and last positions are assigned directly, and `middle_steps` receives every value between them. Exactly one assignment target may use `*`.

## Tuple Method Coverage

Tuples have only two public methods because they are immutable.

| Purpose | Method |
| --- | --- |
| Count matching values | `count()` |
| Find the first matching position | `index()` |

Creation, indexing, slicing, unpacking, concatenation, and conversion are tuple operations rather than tuple methods.

Choose a tuple when order matters but the number and meaning of positions should stay fixed. Choose a list when items need to be added, removed, sorted, or edited frequently.
