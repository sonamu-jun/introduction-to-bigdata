# Structuring

<span style="color: #777777;"><em>구조화</em></span>


## Rows and Fields

<span style="color: #777777;"><em>행과 항목</em></span>

- `Structuring` puts records into rows and columns.  
  <span style="color: #777777;"><em>`구조화`(structuring)는 기록을 행과 열로 정리합니다.</em></span>

- Here one row holds one student’s score in one subject.  
  <span style="color: #777777;"><em>여기서는 한 행에 한 학생의 한 과목 점수를 담습니다.</em></span>


## Parse the Records

<span style="color: #777777;"><em>기록 파싱</em></span>

- `split(",")` separates these simple records at commas.  
  <span style="color: #777777;"><em>`split(",")`은 이 단순한 기록을 쉼표 기준으로 나누는 `파싱`(parsing)입니다.</em></span>

- `int()` changes score text to an integer; `DataFrame` gives the fields column names.  
  <span style="color: #777777;"><em>`int()`는 점수 문자열을 정수로 바꾸고, `DataFrame`은 각 항목에 열 이름을 붙입니다.</em></span>


- Parsing separates a text record into fields; type conversion makes the score a number.  
  <span style="color: #777777;"><em>`파싱`(parsing)은 텍스트를 항목으로 나누고, 자료형 변환은 점수를 숫자로 바꿉니다.</em></span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/05-2_Data_Wrangling/assets/02_structuring/concept.webp" width="467" alt="Parsing separates a text record into fields; type conversion makes the score a number.">


```python
import pandas as pd

records = ["Hana,85,English", "Hana,77,Math", "Hojun,90,English", "Hojun,95,Math"]
rows = []
for record in records:
    name, score, subject = record.split(",")
    rows.append([name, subject, int(score)])

scores = pd.DataFrame(rows, columns=["name", "subject", "score"])
print(scores)
```

- Four text records become four rows with name, subject, and score columns.  
  <span style="color: #777777;"><em>텍스트 기록 4개가 이름·과목·점수 열을 가진 행 4개로 바뀝니다.</em></span>


## Long and Wide Forms

<span style="color: #777777;"><em>긴 형식과 넓은 형식</em></span>

- This `long form` has one row per student–subject score.  
  <span style="color: #777777;"><em>현재 `긴 형식`(long form)은 학생–과목 점수마다 행이 하나입니다.</em></span>

- `pivot()` makes a `wide form` with one subject per column; each student–subject pair needs one score.  
  <span style="color: #777777;"><em>`pivot()`은 과목별 열을 가진 `넓은 형식`(wide form)을 만들며, 학생–과목 쌍마다 점수가 하나여야 합니다.</em></span>


```python
wide_scores = scores.pivot(index="name", columns="subject", values="score")
print(wide_scores)
```

- The same four scores now fill two student rows and two subject columns.  
  <span style="color: #777777;"><em>같은 점수 4개가 학생별 행 2개와 과목별 열 2개에 배치됩니다.</em></span>
