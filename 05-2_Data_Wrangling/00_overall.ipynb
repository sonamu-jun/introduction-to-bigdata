## Purpose

<span style="color: #777777; font-style: italic;">목적</span>

- `Data wrangling` prepares raw data for analysis.  
  <span style="color: #777777; font-style: italic;">`데이터 랭글링`(data wrangling)은 원본 자료를 분석에 맞게 다듬는 과정입니다.</span>

- The input may contain mixed formats, missing values, or repeated records.  
  <span style="color: #777777; font-style: italic;">입력에는 서로 다른 형식, `결측값`(missing value), 중복 기록이 있을 수 있습니다.</span>

- Keep the raw data to check what changed.  
  <span style="color: #777777; font-style: italic;">무엇을 바꿨는지 확인할 수 있도록 원본을 보관합니다.</span>


## Five Common Steps

<span style="color: #777777; font-style: italic;">주요 다섯 단계</span>

- `Discovering` finds what the data contains.  
  <span style="color: #777777; font-style: italic;">`현황 파악`(discovering)은 자료에 무엇이 들어 있는지 살펴봅니다.</span>

- `Structuring` puts it into rows and columns.  
  <span style="color: #777777; font-style: italic;">`구조화`(structuring)는 자료를 행과 열로 정리합니다.</span>

- `Cleaning` handles missing or wrong values.  
  <span style="color: #777777; font-style: italic;">`정제`(cleaning)는 누락되거나 잘못된 값을 처리합니다.</span>

- `Enriching` adds useful fields.  
  <span style="color: #777777; font-style: italic;">`정보 보강`(enriching)은 유용한 항목을 더합니다.</span>

- `Validating` checks the result.  
  <span style="color: #777777; font-style: italic;">`검증`(validating)은 결과를 확인합니다.</span>


## A Score Record

<span style="color: #777777; font-style: italic;">점수 기록</span>

- In `"Hana,85,English"`, Hana is the name, 85 is the score, and English is the subject.  
  <span style="color: #777777; font-style: italic;">`"Hana,85,English"`에서 Hana는 이름, 85는 점수, English는 과목입니다.</span>

- Turn these parts into fields, add a department, and check the score range.  
  <span style="color: #777777; font-style: italic;">각 부분을 항목으로 나누고, 소속 분야를 추가한 뒤 점수 범위를 확인합니다.</span>


- Choose each change based on the data and the analysis goal.  
  <span style="color: #777777; font-style: italic;">데이터 상태와 분석 목적에 맞춰 변경 방법을 선택합니다.</span>
