# Queues

*Use first-in, first-out access to process items in arrival order.*

A **queue** adds items at the rear and removes them from the front.
**FIFO**, or first in, first out, means the earliest waiting item is
processed next. A printer's document queue follows this rule.


## Enqueue and Dequeue

`enqueue()` adds an item; `dequeue()` removes and returns the oldest one.
The small class below stores the front at index zero and the rear at the
end of a list. `pop(0)` returns the first item and shifts the rest left.

```text
dequeue <- front                                rear <- enqueue
           [Document-1] [Document-2] [Document-3]
```


````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        return self.items.pop(0)


documents = Queue()
for document in ["Document-1", "Document-2", "Document-3"]:
    documents.enqueue(document)

print("Waiting:", documents.items)
````

## Preserving Arrival Order

A document that arrives later joins the rear. Completing a document
changes the front but preserves the relative order of those still waiting.


**Question.** After processing `Document-1` and adding `Document-4`, which document is processed next?


````python
print("Processed:", documents.dequeue())
documents.enqueue("Document-4")
print("Waiting after arrival:", documents.items)

print("Processed next:", documents.dequeue())
documents.enqueue("Document-5")
print("Still waiting:", documents.items)
````

`Document-2` is processed next. The queue then contains `Document-3`,
`Document-4`, and `Document-5` in arrival order. The new documents cannot
overtake those already waiting.

`pop(0)` takes `O(n)` time because later list entries shift. For frequent
queue operations, `collections.deque` provides `append()` and `popleft()`
with approximately `O(1)` time at either end. [Python 3.12 deque documentation](https://docs.python.org/3.12/library/collections.html#collections.deque)
