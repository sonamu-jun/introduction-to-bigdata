# Random Numbers and Sampling

*Use a seeded generator to create reproducible values and distinguish sampling with and without replacement.*


## A Reproducible Generator

`np.random.default_rng(seed)` creates a generator with its own state.
A fixed seed reproduces the sequence when the same calls are made
in the same environment. Each call advances the state, so run the
notebook from the beginning when reproducing its complete output.

`integers(low, high, size)` includes `low` and excludes `high`.
A six-sided die therefore uses the bounds 1 and 7.


```python
import numpy as np

rng = np.random.default_rng(42)
dice_rolls = rng.integers(1, 7, size=10)
replay_rng = np.random.default_rng(42)
replay_rolls = replay_rng.integers(1, 7, size=10)

print("Seed:", 42)
print("Dice rolls:", dice_rolls)
print("Recreated generator gives the same rolls:", np.array_equal(dice_rolls, replay_rolls))
```

The ten rolls are `[1, 5, 4, 3, 3, 6, 1, 5, 2, 1]`. Recreating a
generator with seed 42 and making the same first call gives the
same sequence, so the check is `True`.


## Uniform and Normal Values

`random(size)` samples uniformly from `[0, 1)`. `normal(loc, scale,
size)` draws from a normal distribution, where `loc` is the population
mean and `scale` is the population standard deviation. A small
generated sample need not have exactly that mean or spread.


```python
uniform_samples = rng.random(5)
normal_scores = rng.normal(loc=75, scale=10, size=8)

print("Uniform samples:", np.round(uniform_samples, 4))
print("Normal samples:", np.round(normal_scores, 2))
print("Generated sample mean:", round(normal_scores.mean(), 2))
```

The five uniform values lie between 0 and 1. The normal values are
centered around 75 but do not all equal 75. These are simulated
numbers, not student records; a normal distribution is not bounded
to the score interval from 0 to 100.


## Sampling Without Replacement

`rng.choice(values, size, replace=False)` selects distinct input
positions. For the unique student IDs below, that means no student
can appear twice in one sample. The requested count cannot exceed
the available population when sampling without replacement.

**Question.** Can the same student ID occur twice in this four-student sample?


```python
student_ids = np.arange(1, 11)
selected_students = rng.choice(student_ids, size=4, replace=False)

print("Available student IDs:", student_ids)
print("Selected student IDs:", selected_students)
print("Distinct selected IDs:", np.unique(selected_students).size)
```

No: the four selected IDs are distinct, and the distinct count is 4.
A different selection order is possible in another draw, but a
student cannot repeat within this draw.


## Sampling With Replacement

`replace=True` lets an input position be chosen again. A sample can
then be longer than the available data. This is the resampling idea
used by a bootstrap; this one sample alone is not an uncertainty estimate.


```python
resampled_scores = rng.choice(normal_scores, size=10, replace=True)

print("Original generated score count:", normal_scores.size)
print("Resampled score count:", resampled_scores.size)
print("Resampled scores:", np.round(resampled_scores, 2))
```

Ten values are selected from eight generated scores, so at least
one input position is selected more than once. Every output value
comes from the existing `normal_scores` array.


Use a local generator and a seed for repeatable examples. Choose replacement according to whether the same observation may be selected again. [NumPy random generator](https://numpy.org/doc/2.3/reference/random/generator.html)
