```python
print(ord("b"))
```

```python
raw_text = "  Python,SQL,NumPy  "
clean_text = raw_text.strip()
topics = clean_text.split(",")
```

- `raw_text, clean_text, topics`: common string methods clean and divide text.



```python
print(clean_text)
```

```python
print(clean_text.lower())
```

```python
print(topics)
```

```python
print(" | ".join(topics))
```

```python
print(clean_text.replace("NumPy", "Pandas"))
```

```python
print("SQL" in clean_text)
```

- An f-string inserts variable values between braces.


```python
print(f"Topics: {topics}")
```
