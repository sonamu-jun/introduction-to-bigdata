# 012-010. Classes

*A class defines a new kind of object by keeping related state and behavior together.*

The conceptual foundation is developed in Functions and Object State.

## Method · Small Class Definition

`__init__` runs when an instance is created. `self` refers to that particular instance, and attributes store its state.

```python
class Measurement:
    def __init__(self, feature, value, unit):
        self.feature = feature
        self.value = value
        self.unit = unit

    def describe(self):
        return f"{self.feature}: {self.value} {self.unit}"
```

## Implementation · Instance Creation and Method Calls

The class is the definition; `bill_length` and `body_mass` are two independent instances. A method reads instance state through `self`.

```python
bill_length = Measurement("bill length", 39.1, "mm")
body_mass = Measurement("body mass", 3750, "g")

print(bill_length.describe())
print(body_mass.describe())
```

## Implementation · Independent Instance State

Each object keeps its own attribute values.

```python
bill_length.value = 40.3

print("Updated:", bill_length.describe())
print("Unchanged:", body_mass.describe())
```

### Interpretation · Independent Instance State

Use a class when a domain concept needs both state and reusable behavior. At this stage, understanding instances, attributes, and methods is more important than inheritance or advanced class syntax.
