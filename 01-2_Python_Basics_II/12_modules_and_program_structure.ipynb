# Modules and Program Structure

*Reuse library code through imports and give a small script one clear entry point.*

A module is a Python file that provides names such as functions, classes, and constants. Imports prevent every program from reimplementing common tools.

## Import Styles

`import module` keeps the module name visible. `from module import name` imports selected names directly, and `as` creates a short local alias.

## Helper Functions and `main()`

Helper functions handle separate jobs. A `main()` function states the program's top-level workflow instead of leaving many unrelated statements at file scope.

## The Main Guard

In a `.py` file, `__name__` equals `"__main__"` when the file is run directly. When another file imports it, `__name__` contains the module name instead.

Prefer standard-library tools over reimplementing common behavior, keep imports near the top of a script, divide work into focused functions, and use a main guard when one file should support both importing and direct execution.
