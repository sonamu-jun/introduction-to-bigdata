# JSON Files

*Learn to load, save, and reopen structured JSON data with Python's `json` module.*

JSON represents objects, arrays, strings, numbers, booleans, and null values in a portable text format.


## Setting Input and Output Directories

Set the JupyterLite base, locate the supplied JSON with `os.path`, and create `outputs/04_json_files` before writing.


```python
import json
import os

jupyterlite_base_directory = "/home/pyodide"
chapter_name = "01-3_Python_Basics_III"
launch_directory = os.getcwd()

if os.path.isdir(jupyterlite_base_directory):
    base_directory = jupyterlite_base_directory
elif os.path.basename(launch_directory) == chapter_name:
    base_directory = launch_directory
else:
    base_directory = os.path.join(launch_directory, chapter_name)

if not os.path.isdir(base_directory):
    raise FileNotFoundError(f"Base directory not found: {base_directory}")

os.chdir(base_directory)

input_directory = os.path.join(base_directory, "assets", "04_json_files")
input_path = os.path.join(input_directory, "study_profile.json")

assert os.path.isfile(input_path), (
    "Add study_profile.json and set Save folder to assets/04_json_files."
)

output_directory = os.path.join(base_directory, "outputs", "04_json_files")
os.makedirs(output_directory, exist_ok=True)
output_path = os.path.join(output_directory, "study_profile_copy.json")

print("JupyterLite base:", jupyterlite_base_directory)
print("Current directory name:", os.path.basename(os.getcwd()))
print("Input folder:", os.path.relpath(input_directory, base_directory))
print("Input file:", os.path.basename(input_path))
print("Output folder:", os.path.relpath(output_directory, base_directory))
print("Output directory exists:", os.path.isdir(output_directory))
```

The source file remains in the asset directory, while the new copy is written to the visible lesson output folder.


## Loading a JSON File

`json.load()` reads JSON text from an open file and converts it to Python values.


```python
with open(input_path, mode="r", encoding="utf-8") as source_file:
    loaded_profile = json.load(source_file)

print("Loaded profile:", loaded_profile)
print("Python type:", type(loaded_profile).__name__)
```

The JSON object becomes a dictionary, its array becomes a list, `true` becomes `True`, and `null` becomes `None`.


## Saving a JSON File

`json.dump()` writes a Python value to an open text file. Indentation keeps this teaching output readable.


```python
with open(output_path, mode="w", encoding="utf-8") as saved_file:
    json.dump(
        loaded_profile,
        saved_file,
        indent=2,
        ensure_ascii=False,
    )

print("Saved file:", os.path.basename(output_path))
print("Saved size in bytes:", os.path.getsize(output_path))
```

The dictionary is stored as a new UTF-8 JSON file, and the supplied source remains unchanged.


## Reopening the Saved JSON

Load the saved file again and compare the restored value with the in-memory source value.


```python
with open(output_path, mode="r", encoding="utf-8") as saved_file:
    reloaded_profile = json.load(saved_file)

print("Saved data matches:", reloaded_profile == loaded_profile)
print("Student:", reloaded_profile["student"])
```

The reopened dictionary matches the value passed to `json.dump()`, confirming the complete save-and-load round trip.


## Handling Invalid JSON

Malformed JSON raises `json.JSONDecodeError`. This program-owned invalid file demonstrates the specific exception without changing the supplied JSON.


```python
invalid_path = os.path.join(output_directory, "invalid.json")

with open(invalid_path, mode="w", encoding="utf-8") as invalid_file:
    invalid_file.write('{"student": "Amina",}')

try:
    with open(invalid_path, mode="r", encoding="utf-8") as invalid_file:
        json.load(invalid_file)
except json.JSONDecodeError as error:
    print("Invalid JSON at column:", error.colno)
```
