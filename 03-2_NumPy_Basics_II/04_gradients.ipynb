# Numerical Gradients

*Estimate change from sampled values and interpret spacing and axis directions.*

`np.gradient()` estimates a derivative from nearby samples. It works with numerical observations rather than a symbolic formula.


## Change Along One Axis

The values `[0, 1, 4, 9, 16]` sample `x ** 2` at positions 0 through 4.
With spacing 1, an interior estimate uses the next and previous
samples: `(next_value - previous_value) / 2`. By default, endpoints
use a one-sided difference because a neighbor is missing.

**Question.** What rate does `np.gradient()` estimate at position 2?


```python
import numpy as np

positions = np.arange(5, dtype=float)
measurements = np.array([0, 1, 4, 9, 16], dtype=float)
estimated_gradient = np.gradient(measurements, 1)

print("Positions:", positions)
print("Measurements:", measurements)
print("Estimated gradient:", estimated_gradient)
print("Exact derivative of x squared:", 2 * positions)
```

The answer is `(9 - 1) / 2 = 4`. The estimate is `[1, 2, 4, 6, 7]`.
The exact derivative is `[0, 2, 4, 6, 8]`; the default endpoint
estimates differ, showing that a numerical derivative is an approximation.


## The Distance Between Samples

The spacing argument supplies the distance between observations.
If the same measurement changes occur over twice the distance or
time, the rate per unit is halved. Always attach a meaning to that unit.


```python
gradient_per_two_minutes = np.gradient(measurements, 2)

print("Rate with a two-minute interval:", gradient_per_two_minutes)
print("Half the unit-spacing result:", np.allclose(gradient_per_two_minutes, estimated_gradient / 2))
```

The new rates are `[0.5, 1., 2., 3., 3.5]` measurement units per minute.
The check is `True`: the readings are unchanged, but their spacing
changes the interpretation of the rate.


## Change Along Two Axes

For a two-dimensional surface, `np.gradient(surface, row_spacing,
column_spacing)` returns two arrays. The first describes change
along increasing row positions (axis 0); the second describes change
along increasing column positions (axis 1).


```python
surface = np.array([[1, 2, 4],
                    [2, 3, 5],
                    [4, 6, 8]], dtype=float)
row_gradient, column_gradient = np.gradient(surface, 1, 1)

print("Surface:")
print(surface)
print("Gradient along rows:")
print(row_gradient)
print("Gradient along columns:")
print(column_gradient)
print("Gradient shapes:", row_gradient.shape, column_gradient.shape)
```

Both gradient arrays retain shape `(3, 3)`. At the center, the row
rate is `(6 - 2) / 2 = 2`, while the column rate is `(5 - 2) / 2 = 1.5`.
A gradient estimates local change; it does not reduce an axis to a total.
