## Method · Relative Path Reconstruction

This lesson performs the filesystem operations that the previous path lesson deliberately avoided. Each lesson uses its own current working folder, so this notebook rebuilds `output/path_demo/observations.txt` locally instead of depending on a file from the previous notebook. The relative path works in Jupyter, Colab, and Playground; the printed absolute path identifies its resolved location in the current environment.

```python
from pathlib import Path

lesson_dir = Path.cwd()
output_dir = lesson_dir / "output"
demo_dir = output_dir / "path_demo"
note_path = demo_dir / "observations.txt"
cleanup_after_demo = False

print("Working folder:", lesson_dir.name)
print("Folder:", demo_dir.relative_to(lesson_dir))
print("File:", note_path.relative_to(lesson_dir))
print("Resolved file:", note_path.resolve())
print("File exists before writing:", note_path.exists())
```

## Implementation · Directory and File Creation

`mkdir(parents=True, exist_ok=True)` safely creates the required folders. `write_text()` creates or replaces only the named UTF-8 file, so rerunning the cell produces the same content.

| Step | Filesystem effect |
|---|---|
| `note_path = demo_dir / "observations.txt"` | Describes an address only |
| `demo_dir.mkdir(...)` | Creates the named folder path |
| `note_path.write_text(...)` | Creates or replaces the named file |

Construction changes a Python value; only the explicit filesystem methods change stored objects.

```python
demo_dir.mkdir(parents=True, exist_ok=True)
characters_written = note_path.write_text(
    "08:00,12.4\n09:00,13.1\n10:00,13.8\n",
    encoding="utf-8",
)

print("Characters written:", characters_written)
print("File exists after writing:", note_path.is_file())
```

## Implementation · File Reading and Directory Structure

Reading the text and listing the exact demonstration folder displays the difference between a path description and an existing filesystem object. The compact table and tree provide two views of the same stored result.

```python
stored_text = note_path.read_text(encoding="utf-8")

print("kind   | relative path                     | exists")
print("-------|-----------------------------------|-------")
print("folder |", demo_dir.relative_to(lesson_dir), "|", demo_dir.is_dir())
print("file   |", note_path.relative_to(lesson_dir), "|", note_path.is_file())
print()
print("output/")
print("└── path_demo/")
for item in sorted(demo_dir.iterdir()):
    print("    └──", item.name, "-", item.stat().st_size, "bytes")
print()
print(stored_text)
```

## Implementation · Exact Optional Cleanup

With `cleanup_after_demo` set to `False`, the output remains available for inspection after class. Setting it to `True` removes only the exact file created here and its folder when that folder is empty; no other file in `output` is touched.

```python
if cleanup_after_demo:
    if note_path.exists():
        note_path.unlink()
    if demo_dir.exists():
        remaining_items = list(demo_dir.iterdir())
        if len(remaining_items) == 0:
            demo_dir.rmdir()
    print("Exact demonstration cleanup completed.")
else:
    print("Cleanup skipped; set cleanup_after_demo to True to remove the exact demo target.")

print("File remains:", note_path.exists())
print("Demo folder remains:", demo_dir.exists())
```

### Interpretation · Exact Optional Cleanup

`mkdir()`, `write_text()`, `read_text()`, `iterdir()`, `unlink()`, and `rmdir()` operate on the filesystem. The fixed target and guarded cleanup make the notebook repeatable without broad deletion.
