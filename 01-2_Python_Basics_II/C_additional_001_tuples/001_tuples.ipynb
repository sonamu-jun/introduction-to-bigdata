*Use a tuple for a short ordered record whose positions should not be added, removed, or reassigned.*

The conceptual foundation is developed in Collections and Iteration.

## Method · Ordered Tuple Records

Parentheses commonly display a tuple. A one-item tuple needs a trailing comma so Python can distinguish it from ordinary grouping.

```python
observation = ("Adelie", 39.1, 3750)
one_item = ("Adelie",)

print(observation, type(observation))
print("Species:", observation[0])
print("One-item tuple:", one_item)
print("Fields:", len(observation))
```

## Implementation · Positional Unpacking

The number of names must match the number of tuple values. Unpacking removes repeated index notation from later code.

```python
species, bill_length_mm, body_mass_g = observation

print("Species:", species)
print("Bill length (mm):", bill_length_mm)
print("Body mass (g):", body_mass_g)
```

### Interpretation · Positional Unpacking

A tuple can be indexed and sliced like a list, but its items cannot be reassigned. This immutability indicates that its positional structure is fixed.
