*Calculate income, cost, and balance from named inputs and print an itemized report.*

Read the problem statement and the following theory sections before running the notebook: Values, Names, and Numeric Types, String Structure and Transformation.

## Case Data

The problem fixes the event name, attendance estimate, per-person fee, fixed room cost, and per-person refreshment cost. The notebook prints these inputs before calculating derived amounts.

## Method Application · Named Budget Inputs

The first output preserves the planning assumptions beside explicit units.

```python
event_name = "Data Night"
attendees = 48
fee_per_attendee_krw = 18_000
room_cost_krw = 500_000
refreshment_per_person_krw = 6_500

print("Event:", event_name)
print("Expected attendees:", attendees)
print("Fee per attendee (KRW):", fee_per_attendee_krw)
```

## Method Application · Income and Cost Arithmetic

The calculation derives income and each cost component from the named assumptions, then shows every intermediate amount.

```python
ticket_income_krw = attendees * fee_per_attendee_krw
refreshment_cost_krw = attendees * refreshment_per_person_krw
total_cost_krw = room_cost_krw + refreshment_cost_krw
remaining_budget_krw = ticket_income_krw - total_cost_krw

print("Ticket income (KRW):", ticket_income_krw)
print("Refreshment cost (KRW):", refreshment_cost_krw)
print("Total cost (KRW):", total_cost_krw)
print("Remaining budget (KRW):", remaining_budget_krw)
```

## Results · Labeled Budget Balance

The final report includes the event identifier, calculated income, cost, and remaining balance.

```python
report = (
    f"{event_name} | attendees={attendees} | "
    f"income_krw={ticket_income_krw} | "
    f"cost_krw={total_cost_krw} | "
    f"balance_krw={remaining_budget_krw}"
)
print(report)
```

## Result Analysis

### Applied Theory

### Computed Results

The prepared inputs produce 864,000 KRW of ticket income and 312,000 KRW of refreshment cost. Adding the 500,000 KRW room cost gives 812,000 KRW of total cost, leaving a 52,000 KRW balance. The final report preserves the event name, attendance assumption, income, cost, and balance on one labeled line.

### Interpretation

Under the stated assumptions, expected income exceeds expected cost by 52,000 KRW. The labeled terms show that attendance affects both income and refreshment cost, while room cost remains fixed.

### Assumptions and Limits

The calculation treats attendance, fees, and costs as certain and excludes taxes, cancellations, and unplanned expenses. A positive estimate is therefore a planning result, not a guarantee.
