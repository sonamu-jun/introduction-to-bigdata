# Structuring Records

*Convert delimited text into typed records while preserving the observation represented by each row.*


## Records and a Schema

A schema names the fields, their types, and what one row represents. Here each
row describes one person: `name` and `city` are text, while `age` is an integer.
Names uniquely identify people only within this small prepared example.

Parsing splits a representation into fields. Type conversion gives a field
its intended computational meaning. These operations prepare data; they do
not compute a grouped summary.


## Parse the Supplied Text Records

`csv.reader()` reads comma-delimited fields and understands quoted commas,
unlike splitting every comma with `str.split(",")`. Each three-field row is
unpacked into named variables. `int(age_text)` converts the age explicitly.


```python
import csv
import pandas as pd

raw_lines = [
    "John,25,New York",
    "Alice,30,Los Angeles",
    "Bob,22,Chicago",
    "Grace,27,Houston",
]
structured_records = []
for name, age_text, city in csv.reader(raw_lines):
    structured_records.append({"name": name, "age": int(age_text), "city": city})

people_df = pd.DataFrame(structured_records)
print("Structured records:")
print(people_df.to_string(index=False))
print("Column types:")
print(people_df.dtypes)
```

Four input strings become four records. Age now has an integer type, so a
numeric comparison or mean operates on years rather than text characters.


## Sorting Reorders Records

`.sort_values()` orders rows by field values. It preserves the columns and
the meaning of each row. Use a secondary key to make ties predictable.
Sorting is different from pivoting, which changes which values appear as
rows and columns.


```python
ordered_df = people_df.sort_values(["age", "name"], ascending=[False, True])
print("People ordered by age:")
print(ordered_df.to_string(index=False))
print("Rows preserved:", len(ordered_df) == len(people_df))
```

Alice, Grace, John, and Bob appear in decreasing age order. No field has
moved from rows to columns, and no observations have been aggregated.


## Split a Field with a Known Separator

For a deliberately simple format in which fields never contain `:`,
`str.split(":")` is sufficient. Each record here represents a student–subject
score, so the same student may legitimately appear in another subject.


```python
score_lines = [
    "Jiyoung:100:Mathematics",
    "Hojun:90:English",
    "Hana:85:English",
    "Kihoon:95:Mathematics",
]
score_records = []
for line in score_lines:
    name, score_text, subject = line.split(":")
    score_records.append({"name": name, "subject": subject, "score": int(score_text)})

print("Student-subject records:")
print(pd.DataFrame(score_records).to_string(index=False))
```

The four scores retain their subject labels. The pair `(name, subject)`
describes the observation more precisely than a name alone.


Define one row and its fields before converting values. Parsing, numeric conversion, sorting, and reshaping solve different structural problems.
