# Strings

*Inspect characters, select slices, clean raw text, and format a labeled output string.*


## Text Creation and Inspection

Both single and double quotation marks create a string. Digits inside quotation marks are text, not a number.


```python
topic = 'Python'
text_number = "2026"

print(topic, type(topic))
print(text_number, type(text_number))
print("Characters:", len(topic))
```

## Character Indexing and Range Slicing

`text[start:stop]` includes `start` and excludes `stop`. A negative index counts backward from the end.


```python
language = "abcdef"

print("First character:", language[0])
print("Last character:", language[-1])
print("Indexes 1 through 3:", language[1:4])
print("First three:", language[:3])
```

## Raw-Text Cleaning and Splitting

Data often arrives with extra spaces or delimiters. `strip()` removes outer whitespace, `split()` divides one string into a list, and `join()` combines a list of strings.


```python
raw_text = " Python,SQL,NumPy "
clean_text = raw_text.strip()
topics = clean_text.split(",")

print("Original:", repr(raw_text))
print("Cleaned:", repr(clean_text))
print("Parts:", topics)
print("Joined:", " | ".join(topics))
```

`repr()` displays leading and trailing spaces that standard string output does not show. Notice that `raw_text` is unchanged: strings are immutable.


## Text Search, Replacement, and Formatting

Membership with `in` answers a yes-or-no question. An f-string places evaluated values inside braces.


```python
updated_text = clean_text.replace("NumPy", "pandas")

print("Contains SQL:", "SQL" in clean_text)
print("Replacement:", updated_text)
print(f"There are {len(topics)} topics: {topics}")
print("There are {} topics: {}".format(len(topics), topics))
```

Indexing returns one character, slicing returns a new substring, and cleaning methods leave the source string available for comparison.
