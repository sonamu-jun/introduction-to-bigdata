# Data Structure Concepts

*Explain how order, labels, and membership change the information a container preserves.*

A **data structure** organizes values and their relationships. The same numbers can form a sequence, values attached to labels, or unique members. These rules determine how we access and change the data.


## Organization and Operations

| Property | What to check |
| --- | --- |
| Order | Do positions matter? Ordered does not mean sorted. |
| Repetition | Must repeated observations remain separate? |
| Access | Do we select by position, key, or membership? |
| Mutability | Can the existing container change? |

Access retrieves an item; traversal visits items. Insertion, update, and deletion change a mutable container. `len()` counts items, or key-value pairs in a dictionary.


**Think before running or viewing the saved output.** For `[30, 45, 30]`, how many readings and distinct values are there? Would a set retain enough information to count repeated readings?

Read positions, key-value arrows, and membership as different relationships.


```python
readings = [30, 45, 30]
fixed_readings = tuple(readings)
minutes_by_day = {"Mon": 30, "Tue": 45}
unique_readings = set(readings)

print("Sequence:", readings)
print("Tuple:", fixed_readings)
print("Value at position 1:", readings[1])
print("Value for Tue:", minutes_by_day["Tue"])
print("Unique members:", sorted(unique_readings))
print("Readings and distinct values:", len(readings), len(unique_readings))
```

**Check:** There are three readings and two distinct values. The tuple retains order and repetitions; the set loses both occurrence counts and positions. The dictionary adds day labels that the numbers alone do not supply. `sorted()` only orders the displayed result.


Choose a structure by the relationships you need to preserve. Converting to another container can discard information.
