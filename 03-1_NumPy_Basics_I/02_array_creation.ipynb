# Array Creation and Reshaping

*Create arrays from prepared values or regular sequences and reshape them without changing their element count.*


## Arrays from Prepared Values

`np.array(values)` creates a one-dimensional array from one list or
a two-dimensional array from equally long nested lists. In a table,
each inner list supplies one row. Rows must have the same length for
an ordinary numeric array.


```python
import numpy as np

measurements = np.array([10, 20, 30, 40])
table = np.array([[1, 2, 3],
                  [4, 5, 6]])

print("Measurements:", measurements)
print("Table:")
print(table)
```

`measurements` holds four values along one axis. `table` has two rows
and three columns. The nested lists describe the table's structure.


## Regular Sequences

`np.arange(start, stop, step)` uses a step and excludes `stop`.
`np.linspace(start, stop, num)` uses a number of points and includes
both endpoints by default. Use `linspace()` when the number of
evenly spaced measurements is the important choice.


```python
index_values = np.arange(0, 10)
numbers = np.arange(1, 13)
time_points = np.linspace(0, 1, 5)

print("Indices:", index_values)
print("Numbers:", numbers)
print("Time points:", time_points)
```

The integer sequences end at 9 and 12 because their stop values are
excluded. The five time points include 0 and 1, with a spacing of 0.25.


## Arrays with an Initial Value

`np.zeros(shape)` and `np.ones(shape)` fill a requested shape with
zeros or ones. A shape such as `(2, 3)` is a tuple of axis lengths.
These functions produce floating-point values unless `dtype` is specified.


```python
empty_counts = np.zeros((2, 3), dtype=int)
initial_weights = np.ones((2, 3))

print("Integer counts:")
print(empty_counts)
print("Initial weights:")
print(initial_weights)
```

Both arrays contain six elements. The integer counts display as `0`,
while the floating-point weights display as `1.`.


## Changing Shape

`array.reshape(rows, columns)` arranges the same elements into a new
shape. The product of the requested axis lengths must equal the
original element count. The default order fills each row in turn.

**Question.** Can the twelve values in `numbers` form a `(3, 5)` array?


```python
print("Three rows and four columns:")
print(numbers.reshape(3, 4))
print("Two rows and six columns:")
print(numbers.reshape(2, 6))

try:
    numbers.reshape(3, 5)
except ValueError:
    print("Cannot reshape 12 values into 3 x 5 = 15 positions.")
```

Both valid shapes hold the values 1 through 12 in the same order.
The answer is no: `(3, 5)` needs fifteen elements. Reshaping does not
invent additional values, and the handled error lets execution continue.


Choose `array()` for prepared values, `arange()` for a step, and `linspace()` for a point count. A reshape must preserve the number of elements.
