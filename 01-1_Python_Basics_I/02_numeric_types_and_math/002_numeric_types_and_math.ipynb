# 011-002. Numeric Types and Math

The examples examine numeric types, arithmetic operators, type conversion, and rounding.


```python
# Integers and floats store whole numbers and decimal values.
whole_number = 10
decimal_number = 20.5
divisor = 3
```

- `whole_number`, `decimal_number`, and `divisor` store numbers that the calculations will use. Integer values are whole numbers, and float values have a decimal part.



```python
print(whole_number + decimal_number)
```

```python
print(whole_number - decimal_number)
```

```python
print(whole_number * decimal_number)
```

```python
print(whole_number / divisor)
```

```python
print(whole_number // divisor)  # quotient
```

```python
print(whole_number % divisor)   # remainder
```

```python
print(whole_number ** divisor)  # power
```

```python
# type() identifies a value's data type.
print(type(whole_number))
```

```python
print(type(decimal_number))
```

```python
print(2.1e3, type(2.1e3))
```

```python
# int() and float() convert between numeric types.
print(int(decimal_number), type(int(decimal_number)))
```

- `int(...)`: converts a compatible value to an integer.



```python
print(float(whole_number), type(float(whole_number)))
```

- `float(...)`: converts a compatible value to a float.



- `round(..., ...)`: rounds a number to the requested number of decimal places.



```python
value = 1.246
```

- `value`: `round()` changes the number of displayed decimal places.


```python
print(round(value, 2))
```

```python
# Python uses round-to-even when a value is exactly halfway.
print(round(8.5))
```

- `round(...)`: rounds a number to the requested number of decimal places.



```python
print(round(9.5))
```
