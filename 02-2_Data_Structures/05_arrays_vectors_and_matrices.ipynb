# Arrays, Vectors, and Matrices

*Read numeric array shape and select values by position or a Boolean condition.*

A **vector** is an ordered collection of numeric components. A **matrix** is rectangular, with rows and columns. A NumPy **array** represents values along axes with an explicit shape and a common element dtype.


## Shape and Selection

`np.array()` constructs an array; `shape` gives each axis length. Shape `(3, 2)` means three rows and two columns, while `(3,)` means one axis with three elements. `dtype="int64"` specifies integer storage.

Use `[row, column]` to select a value and `:` to select an entire axis. `[1, :]` selects row 1; `[:, 0]` selects column 0. Multiplying an array by 2 doubles its values; multiplying a list by 2 repeats its sequence.


<img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/02-2_Data_Structures/assets/05_arrays_vectors_and_matrices/matrix_coordinate_selection.webp" width="640" alt="Row 1 yields [45, 3], column 0 yields [30, 45, 40], and coordinate [2, 0] yields 40.">

Follow a row, a column, and one coordinate pair.


```python
import numpy as np

numeric_matrix = np.array([[30, 2], [45, 3], [40, 4]], dtype="int64")
minutes_vector = numeric_matrix[:, 0]

print("Matrix shape:", numeric_matrix.shape)
print("Row 1:", numeric_matrix[1, :])
print("Column 0:", minutes_vector, "shape:", minutes_vector.shape)
print("Element at row 2, column 0:", numeric_matrix[2, 0])
print("Doubled values:", minutes_vector * 2)
```

The matrix has two axes; selecting its minutes column gives one axis with values `[30, 45, 40]`. The program supplies the meaning of each row and column.


## Boolean Masks

A comparison creates one Boolean per array position. Indexing with this **mask** retains the `True` positions. The short animation changes the threshold while keeping the source values fixed.

Change `threshold_values`, run the next two cells, and use **Play** or the slider. Amber bars pass the condition. The printed trace shows the same selections.


**Think before running or viewing the saved output.** For the minutes `[30, 45, 40]`, which values pass `>= 40`? Which pass `>= 50`? Does filtering change the source vector?


```python
# Change these thresholds, then rerun this cell and the visualization.
filter_readings = minutes_vector.copy()
threshold_values = [30, 40, 50]
mask_frames = [filter_readings >= threshold for threshold in threshold_values]

for threshold, mask in zip(threshold_values, mask_frames):
    print("Threshold:", threshold, "mask:", mask, "selected:", filter_readings[mask])
```

**Provided visualization.** Run the next cell to display the saved results. Its drawing code is collapsed where supported; edit the data cell above.


```python
# Provided visualization: run this cell; edit the data cell above.
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

mask_figure, mask_axis = plt.subplots(figsize=(6, 2.8), dpi=96)


def draw_mask_frame(frame_number):
    threshold = threshold_values[frame_number]
    mask = mask_frames[frame_number]
    selected_values = filter_readings[mask].tolist()
    colors = ["#dca52b" if keep else "#d3d8dd" for keep in mask]
    mask_axis.clear()
    bars = mask_axis.bar(range(filter_readings.size), filter_readings, color=colors)
    mask_axis.bar_label(bars, label_type="center", fontsize=10.5)
    mask_axis.axhline(threshold, color="#9b5f00", linestyle="--", linewidth=1)
    mask_axis.set_xticks(range(filter_readings.size))
    mask_axis.set_ylim(0, max(filter_readings.max(), max(threshold_values)) + 20)
    mask_axis.set_xlabel("Array position", fontsize=10.5)
    mask_axis.set_ylabel("Minutes", fontsize=10.5)
    mask_axis.set_title(f"At least {threshold} minutes: {selected_values}", fontsize=10.5)
    mask_axis.tick_params(labelsize=10.5)
    mask_figure.tight_layout()


mask_animation = FuncAnimation(
    mask_figure, draw_mask_frame, frames=len(mask_frames), interval=1000, repeat=False,
)
animation_html = mask_animation.to_jshtml(default_mode="once")
# Use text symbols instead of downloading an icon font for the controls.
animation_html = animation_html[animation_html.index("<script"):]
control_icons = [
    "minus", "fast-backward", "step-backward", "play fa-flip-horizontal",
    "pause", "play", "step-forward", "fast-forward", "plus",
]
control_symbols = ["−", "|◀", "‹", "◀", "Ⅱ", "▶", "›", "▶|", "+"]
for icon, symbol in zip(control_icons, control_symbols):
    animation_html = animation_html.replace(f'<i class="fa fa-{icon}"></i>', symbol)
display(HTML(animation_html))
plt.close(mask_figure)
```

**Check:** At `40`, the selected values are `[45, 40]`; at `50`, the result is empty. `>=` includes equality. The source remains `[30, 45, 40]`.


Read shape before indexing. Arrays calculate by position; a Boolean mask selects values without changing the source.
