## Method · Initial Text Output

The file contains one comma-separated observation per line, but it is still ordinary text.

`with log_path.open(...) as file:` opens the file and gives it the temporary name `file`. Only the indented block uses the open file; Python closes it automatically when the block ends. Here, `mode="w"` replaces the named file.

```python
from pathlib import Path

output_dir = Path.cwd() / "output"
output_dir.mkdir(exist_ok=True)
log_path = output_dir / "text_file_notes.txt"

initial_text = "08:00,12.4\n09:00,13.1\n"
with log_path.open(mode="w", encoding="utf-8") as file:
    file.write(initial_text)

print("Saved:", log_path.relative_to(Path.cwd()))
```

## Implementation · Line Appending

`mode="a"` selects append mode: it preserves the existing two lines and adds the new observation at the end.

> **Before you run**
>
> Predict how many lines the file will contain.

```python
with log_path.open(mode="a", encoding="utf-8") as file:
    file.write("10:00,13.8\n")
```

## Implementation · Complete-File Reading

`mode="r"` selects read mode. A whole-string read suits a small file, while line-by-line reading suits records that should be processed separately.

```python
with log_path.open(mode="r", encoding="utf-8") as file:
    complete_text = file.read()

observations = []
with log_path.open(mode="r", encoding="utf-8") as file:
    for line in file:
        observations.append(line.strip())

print(complete_text)
print("Parsed lines:", observations)
print("Line count:", len(observations))
```

### Interpretation · Complete-File Reading

The final file has three lines because append mode retained the first two. `strip()` removes each line ending after the file has already been divided into records.
