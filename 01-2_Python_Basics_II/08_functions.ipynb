# Functions

*Give a repeated calculation a name, pass inputs through parameters, and return a result to the caller.*

## Function Definition, Calls, and Reuse

A parameter is a local name for an input. The function does not run until it is called.


```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = celsius * 9 / 5 + 32
    return fahrenheit

print(celsius_to_fahrenheit(0))
print(celsius_to_fahrenheit(25))
```

## Returned Results

A returned value can be stored, compared, or passed into another function. Printing only displays it.


```python
room_temperature_f = celsius_to_fahrenheit(22)
print("Room temperature (°F):", room_temperature_f)
print("Above 70°F:", room_temperature_f > 70)
```

## Numeric Sequence Summary

One function may call another. A default argument supplies a value only when the caller omits that argument.


```python
def mean(values):
    return sum(values) / len(values)

def summarize(values, label="values"):
    return {
        "label": label,
        "count": len(values),
        "mean": round(mean(values), 2),
        "minimum": min(values),
        "maximum": max(values),
    }

flipper_lengths_mm = [181, 186, 195, 193, 190]
summary = summarize(flipper_lengths_mm, "flipper_length_mm")
print(summary)
```

## Positional and Keyword Arguments

Positional arguments are matched by order. Keyword arguments are matched by parameter name and make calls with several inputs easier to read.

```python
def calculate_total(unit_price, quantity, tax_rate=0.1):
    return round(unit_price * quantity * (1 + tax_rate))

positional_total = calculate_total(12000, 3)
keyword_total = calculate_total(quantity=3, unit_price=12000, tax_rate=0.05)

print("Default-tax total:", positional_total)
print("Keyword-call total:", keyword_total)
```

The first call uses the default tax rate. The second names every argument, so their written order may differ from the parameter order.

## Early Returns

An early `return` handles an exceptional or completed case immediately and prevents deeply nested branches.

```python
def score_status(score):
    if score < 0 or score > 100:
        return "invalid"
    if score >= 60:
        return "pass"
    return "review"

print("105:", score_status(105))
print("82:", score_status(82))
print("55:", score_status(55))
```

Each call stops at the first executed `return`. The final return handles the only remaining valid case.

## Returning and Unpacking Multiple Values

Separating returned values with commas creates a tuple. Tuple unpacking assigns the results to descriptive names.

```python
def minimum_and_maximum(values):
    return min(values), max(values)

minimum_score, maximum_score = minimum_and_maximum([72, 85, 91, 68])

print("Minimum:", minimum_score)
print("Maximum:", maximum_score)
```

The function returns one two-item tuple, and unpacking gives each position a clear name.

## Docstrings and Local Scope

A docstring states the function's purpose. Parameters and names created inside the function are local and do not replace unrelated names outside it.

```python
course_name = "Introduction to Big Data"

def make_topic_label(topic):
    """Return a course label for one topic."""
    course_name = "Python Basics"
    local_label = f"{course_name}: {topic}"
    return local_label

print(make_topic_label("Functions"))
print("Outside course name:", course_name)
print("Function purpose:", make_topic_label.__doc__)
```

The local `course_name` is used only during the call. The outside value remains unchanged, and `local_label` is not available outside the function.

## Type Hints

Type hints document expected input and output types. Python does not automatically enforce them at runtime, but editors and analysis tools can use them.

```python
def completion_rate(completed: int, total: int) -> float:
    if total == 0:
        return 0.0
    return round(completed / total * 100, 1)

print("Completion rate:", completion_rate(completed=4, total=5))
```

The annotations describe integer inputs and a floating-point result. The zero-total guard returns before division is attempted.

## Functions as Sorting Keys

`sorted()` may receive a function through its `key` argument. A short `lambda` is useful when the key needs only one simple expression.

```python
student_scores = [
    {"name": "Minji", "score": 85},
    {"name": "Joon", "score": 91},
    {"name": "Sora", "score": 78},
]

ranked_students = sorted(
    student_scores,
    key=lambda student: student["score"],
    reverse=True,
)

print("Ranked students:", ranked_students)
```

The lambda returns one score for each record, and `reverse=True` places the largest score first. Use a named function when the key logic needs more than one short expression.

## Use Cases

Functions commonly preserve one reusable rule, standardize repeated cleaning, and build a consistent result structure.

### Reuse a Validation Rule

A named validation function keeps the same score boundary everywhere it is called.

```python
def is_valid_score(score):
    return 0 <= score <= 100

print("85 is valid:", is_valid_score(85))
print("125 is valid:", is_valid_score(125))
```

The function returns a Boolean result instead of printing inside the function. The caller can later store, compare, or combine that result.

### Standardize Text Cleaning

Repeated text-cleaning steps belong in one function so every input receives the same treatment.

```python
def clean_subject_name(raw_name):
    cleaned_name = raw_name.strip()
    return cleaned_name.title()

print(clean_subject_name("  data science "))
print(clean_subject_name("PYTHON PROGRAMMING"))
```

Both inputs lose unwanted outside spaces and receive consistent capitalization. If the cleaning rule changes, only the function body needs revision.

### Return a Consistent Record

A function can package related outputs in a dictionary. A default argument handles the common unit while still allowing another unit.

```python
def make_measurement(feature, value, unit="mm"):
    return {
        "feature": feature,
        "value": value,
        "unit": unit,
    }

bill_record = make_measurement("bill length", 39.1)
mass_record = make_measurement("body mass", 3750, "g")

print(bill_record)
print(mass_record)
```

Every returned record has the same three keys. The default `"mm"` is used only when the caller omits the unit argument.

### Reusable Calculation Rules

A function keeps a fee formula in one place and exposes the values that are allowed to vary.

```python
def calculate_registration_fee(unit_fee, participants, discount_rate=0):
    subtotal = unit_fee * participants
    return round(subtotal * (1 - discount_rate))

print("Regular fee:", calculate_registration_fee(10000, 4))
print("Discounted fee:", calculate_registration_fee(10000, 4, 0.15))
```

Both calls use the same formula. The default discount keeps the common call short, while the third argument handles a discounted case.

### Repeated Text Parsing

A parsing function turns every input line with the same format into a predictable dictionary.

```python
def parse_measurement(line):
    feature, raw_value, unit = line.split(",")
    return {
        "feature": feature.strip(),
        "value": float(raw_value),
        "unit": unit.strip(),
    }

print(parse_measurement("bill length, 39.1, mm"))
print(parse_measurement("body mass, 3750, g"))
```

The function centralizes splitting, whitespace removal, numeric conversion, and field naming. Every successful call returns the same record structure.

Define one purpose per function, name each parameter by its role, and return a predictable data type. The summary function returns a dictionary that labels each value.
