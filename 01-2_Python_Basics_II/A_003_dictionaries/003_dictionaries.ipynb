# 012-004. Dictionaries

*Represent one labeled record as key–value pairs, then retrieve and update values by name.*

The conceptual foundation is developed in Collections and Iteration.

## Method · Record Creation and Retrieval

A colon separates each key from its mapped value. Descriptive keys make a record self-documenting.

```python
penguin = {
    "species": "Adelie",
    "bill_length_mm": 39.1,
    "body_mass_g": 3750,
}

print("Species:", penguin["species"])
print("Body mass:", penguin["body_mass_g"])
```

## Implementation · Dictionary Mutation

Assigning a new key adds a field. Assigning an existing key replaces its previous value.

```python
penguin["island"] = "Torgersen"
penguin["body_mass_g"] = 3800
removed_value = penguin.pop("bill_length_mm")

print("Updated record:", penguin)
print("Removed bill length:", removed_value)
```

## Implementation · Key Inspection and Missing-Key Handling

Direct lookup of an absent key raises `KeyError`. `get()` is appropriate when a missing key has an expected fallback.

```python
print("Keys:", list(penguin.keys()))
print("Key-value pairs:", list(penguin.items()))
print("Unknown sex:", penguin.get("sex", "not recorded"))
```

### Interpretation · Key Inspection and Missing-Key Handling

Use a list for an ordered collection of records and a dictionary for the labeled fields inside one record. A later DataFrame will combine both ideas into rows and columns.
