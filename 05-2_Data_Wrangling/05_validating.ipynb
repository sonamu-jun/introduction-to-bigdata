# Validating

<span style="color: #777777; font-style: italic;">검증</span>


## Keys and Missing Matches

<span style="color: #777777; font-style: italic;">키와 연결 누락</span>

- `Validating` checks whether the result follows the rules.  
  <span style="color: #777777; font-style: italic;">`검증`(validating)은 결과가 규칙에 맞는지 확인하는 작업입니다.</span>

- A `lookup table` holds information to match by key; here it maps each subject to a department.  
  <span style="color: #777777; font-style: italic;">`조회 표`(lookup table)는 키로 연결할 정보를 담으며, 여기서는 과목별 소속 분야를 담습니다.</span>

- `.duplicated().any()` checks for repeated subjects in the lookup table.  
  <span style="color: #777777; font-style: italic;">`.duplicated().any()`로 조회 표에 중복된 과목이 있는지 확인합니다.</span>

- `.notna()` checks for missing departments after the join; `.all()` checks whether every row has a match.  
  <span style="color: #777777; font-style: italic;">조인 후 `.notna()`로 소속 분야의 누락을 확인하고, `.all()`로 모든 행이 연결되었는지 확인합니다.</span>


```python
import pandas as pd

scores_df = pd.DataFrame({
    "name": ["Jiyoung", "Hojun", "Hana"],
    "subject": ["Mathematics", "English", "Science"],
    "score": [98, 90, 85],
    "maximum_score": [100, 100, 100],
})
lookup_df = pd.DataFrame({
    "subject": ["Mathematics", "English"],
    "department": ["STEM", "Languages"],
})

unique_subjects = not lookup_df["subject"].duplicated().any()
enriched_df = scores_df.merge(lookup_df, on="subject", how="left")

matched = enriched_df["department"].notna()

print("Unique subjects:", unique_subjects)
print(enriched_df)
print("Complete lookup:", matched.all())
```

- `Unique subjects` is `True`; all three score rows remain.  
  <span style="color: #777777; font-style: italic;">`Unique subjects`는 `True`이며 점수 행 3개가 모두 남습니다.</span>

- Science has no department, so Hana’s department is `NaN`.  
  <span style="color: #777777; font-style: italic;">Science의 소속 분야가 없어 Hana의 소속 분야가 `NaN`으로 표시됩니다.</span>

- `Complete lookup` is `False`: unique lookup keys do not guarantee every needed subject is present.  
  <span style="color: #777777; font-style: italic;">`Complete lookup`은 `False`이며, 조회 키가 중복되지 않아도 필요한 과목이 빠질 수 있습니다.</span>


## Score Rules

<span style="color: #777777; font-style: italic;">점수 규칙</span>

- Each maximum score must be positive; each score must be from zero to its maximum.  
  <span style="color: #777777; font-style: italic;">만점은 양수여야 하며, 점수는 0 이상이고 해당 만점 이하여야 합니다.</span>

- Use `&` to combine the lower and upper limits, and `.all()` to check all rows.  
  <span style="color: #777777; font-style: italic;">`&`로 하한과 상한 조건을 묶고 `.all()`로 모든 행이 통과하는지 확인합니다.</span>


```python
scores = enriched_df["score"]
maximums = enriched_df["maximum_score"]
valid_scores = (scores >= 0) & (scores <= maximums)

print("Positive maximums:", (maximums > 0).all())
print("Valid scores:", valid_scores.all())
```

- The maximums are positive and all scores are in `range`.  
  <span style="color: #777777; font-style: italic;">만점은 양수이고 점수는 모두 허용 `범위` 안에 있습니다.</span>


## Validation Results

<span style="color: #777777; font-style: italic;">검증 결과</span>

- The scores pass their checks, but the Science lookup is missing.  
  <span style="color: #777777; font-style: italic;">점수 검사는 통과하지만 Science의 조회 정보는 누락되어 있습니다.</span>
