# Measures of Center

## Describing the Observed Data

<span style="color: #777777;"><em>관측 데이터 설명</em></span>

- `Descriptive statistics` summarize the data we have: center, `spread`, shape, and relationships.  
  <span style="color: #777777;"><em>`기술통계`(descriptive statistics)는 관측 자료의 중심, `산포`(spread), 모양, 관계를 요약합니다.</em></span>

- A `measure of center` gives a typical value: mean, median, or mode.  
  <span style="color: #777777;"><em>`대푯값`(measure of center)은 자료를 대표하는 값이며, 평균·중앙값·최빈값이 있습니다.</em></span>

- The five study times are 20, 25, 30, 30, and 45 minutes.  
  <span style="color: #777777;"><em>학습 시간 다섯 개는 20분, 25분, 30분, 30분, 45분입니다.</em></span>


- Mean, median, and mode are all 30 here, but each uses a different rule.  
  <span style="color: #777777;"><em>평균·중앙값·최빈값은 모두 30이지만 계산 기준은 서로 다릅니다.</em></span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/06-1_Descriptive_Statistics/assets/01_measures_of_center/concept.webp" width="422" alt="Mean, median, and mode are all 30 here, but each uses a different rule.">


## Mean

<span style="color: #777777;"><em>평균</em></span>

- The `mean` is the total divided by the count: $\bar{x}=\frac{1}{n}\sum_i x_i$.  
  <span style="color: #777777;"><em>`평균`(mean)은 합을 개수로 나눈 값입니다: </em></span>${\color{#777777}\bar{x}=\frac{1}{n}\sum_i x_i}$<span style="color: #777777;"><em>.</em></span>

- Here $n$ is the count and $\sum$ means add the values; `mean()` calculates the result.  
  <span style="color: #777777;"><em>여기서 </em></span>${\color{#777777}n}$<span style="color: #777777;"><em>은 개수, </em></span>${\color{#777777}\sum}$<span style="color: #777777;"><em>는 값의 합을 뜻하며, `mean()`으로 계산합니다.</em></span>


```python
from statistics import mean, median, multimode

minutes = [20, 25, 30, 30, 45]
print("Mean:", mean(minutes))
```

- The total is 150 minutes; dividing by 5 gives a mean of 30 minutes.  
  <span style="color: #777777;"><em>합계 150분을 5로 나누면 평균은 30분입니다.</em></span>


## Median

<span style="color: #777777;"><em>중앙값</em></span>

- The `median` is the middle value after sorting; for an even count, average the two middle values.  
  <span style="color: #777777;"><em>`중앙값`(median)은 정렬했을 때 가운데 값이며, 개수가 짝수이면 가운데 두 값의 평균입니다.</em></span>


```python
print("Median:", median(minutes))
```

- The middle of the five sorted values is 30.  
  <span style="color: #777777;"><em>정렬된 값 5개의 가운데 값은 30입니다.</em></span>


## Mode

<span style="color: #777777;"><em>최빈값</em></span>

- The `mode` is the most common value; `multimode()` returns all values tied for the highest count.  
  <span style="color: #777777;"><em>`최빈값`(mode)은 가장 자주 나오는 값이며, `multimode()`는 최고 빈도의 값을 모두 반환합니다.</em></span>


```python
print("Mode:", multimode(minutes))
```

- 30 appears twice, so the result is `[30]`.  
  <span style="color: #777777;"><em>30이 두 번 나오므로 결과는 `[30]`입니다.</em></span>


## Effect of a Large Value

<span style="color: #777777;"><em>큰 값의 영향</em></span>

- Change the longest time from 45 to 180 and compare the mean and median.  
  <span style="color: #777777;"><em>가장 긴 시간을 45에서 180으로 바꾸고 평균과 중앙값을 비교합니다.</em></span>


```python
long_minutes = [20, 25, 30, 30, 180]
print("Mean:", mean(long_minutes))
print("Median:", median(long_minutes))
```

- The mean rises from 30 to 57, while the median stays 30.  
  <span style="color: #777777;"><em>평균은 30에서 57로 올라가지만 중앙값은 30입니다.</em></span>
