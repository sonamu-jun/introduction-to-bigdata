# Aggregation

*Summarize numeric columns and distinguish totals across rows from totals across columns.*


## Summarize a Numeric Column

`.sum()`, `.mean()`, `.min()`, and `.max()` reduce many values to one
result. Select the numeric column explicitly so category labels do not
become part of a calculation.


```python
import pandas as pd

sales_df = pd.DataFrame({
    "month": ["Jan", "Feb", "Mar", "Apr"],
    "online": [120, 135, 150, 160],
    "store": [80, 90, 85, 95],
})

print("Monthly sales (sales units):")
print(sales_df)
print("Total online sales:", sales_df["online"].sum())
print("Average store sales:", sales_df["store"].mean())
print("Minimum online sales:", sales_df["online"].min())
print("Maximum online sales:", sales_df["online"].max())
```

Online sales total 565, with a minimum of 120 and maximum of 160. Average store sales are 87.5 per month.


## Choose the Reduction Axis

For a numeric DataFrame, `axis=0` reduces down rows and returns one total
per column. `axis=1` reduces across columns and returns one total per row.
Here rows are months and columns are sales channels.


```python
channel_sales = sales_df[["online", "store"]]
sales_df["total"] = channel_sales.sum(axis=1)

print("Totals by channel:")
print(channel_sales.sum(axis=0))
print("Totals by month:")
print(sales_df[["month", "total"]])
print("Grand total:", sales_df["total"].sum())
```

`axis=1` produces monthly totals 200, 225, 235, and 255. The grand total is 915, equal to online 565 plus store 350.


## Request Several Statistics Together

`.agg(["sum", "mean", "max"])` applies several named reductions to
each selected column. Its output labels identify both the statistic and
the channel, so the summary remains interpretable.


```python
summary_df = channel_sales.agg(["sum", "mean", "max"])

print("Channel summaries:")
print(summary_df)
print("Grand totals agree:", channel_sales.sum(axis=0).sum() == sales_df["total"].sum())
```

Online mean sales are 141.25 and store mean sales are 87.5. The equality check confirms that either direction of summation accounts for the same eight values.


Select numeric fields deliberately and state the meaning of the axis. Use `agg` when a report needs several statistics for the same columns.
