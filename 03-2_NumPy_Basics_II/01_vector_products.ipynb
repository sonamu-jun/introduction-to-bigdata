# Vector Products

*Distinguish elementwise, dot, outer, and cross products by their calculation and output shape.*

A one-dimensional array can represent a mathematical vector. The product to choose depends on whether we need matching products, a combined score, all pairwise products, or a direction in three-dimensional space.


## Elementwise and Dot Products

`left * right` keeps each matching product. For one-dimensional arrays,
`left @ right` and `np.dot(left, right)` add those products into one
number. `np.inner()` gives the same result for these one-dimensional inputs.

**Question.** Will `left * right` and `left @ right` return the same
number of values?


```python
import numpy as np

left = np.array([1, 2, 3])
right = np.array([4, 5, 6])

print("Elementwise product:", left * right)
print("Dot product with @:", left @ right)
print("Dot product with np.dot:", np.dot(left, right))
print("Inner product:", np.inner(left, right))
```

No: the elementwise result has three values, `[4, 10, 18]`, while the
dot product is one scalar, `4 + 10 + 18 = 32`. A weighted total is
one common application of the dot product.


## All Pairs with an Outer Product

`np.outer(left, right)` multiplies every value on the left by every
value on the right. If their lengths are `m` and `n`, the result
has shape `(m, n)`.


```python
pairwise_products = np.outer(left, right)

print("Outer product:")
print(pairwise_products)
print("Outer product shape:", pairwise_products.shape)
```

The result has shape `(3, 3)`. Its rows are `[4, 5, 6]`, `[8, 10, 12]`,
and `[12, 15, 18]`. Every left-right pair has its own position.


## A Cross Product in Three Dimensions

For two vectors with three components, `np.cross()` calculates their
cross product. For nonparallel input vectors, its direction is
perpendicular to both inputs. A dot product of zero verifies
perpendicularity for the exact integer values in this example.


```python
perpendicular = np.cross(left, right)

print("Cross product:", perpendicular)
print("Dot product with left:", perpendicular @ left)
print("Dot product with right:", perpendicular @ right)
```

The cross product is `[-3, 6, -3]`. Both checks are zero, so this
vector is perpendicular to the two inputs. Reversing the input order
would reverse its direction.


Use `*` for matching products, `@` for a dot product of vectors, and `np.outer()` for all pairs. The three-dimensional cross product expresses a geometric direction. [NumPy cross product](https://numpy.org/doc/2.3/reference/generated/numpy.cross.html)
