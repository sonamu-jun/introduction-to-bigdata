## Method · Two-Way Branching

The condition after `if` is evaluated first. `else` handles every remaining case.

```python
score = 82

if score >= 60:
    result = "pass"
else:
    result = "fail"

print(result)
```

## Implementation · Ordered Multiway Branching

The most specific high threshold belongs first. Once a true branch runs, Python skips the rest of the chain.

> **Before you run**
>
> Predict the letter grade for a score of 82.

```python
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D or F"

print("Grade:", grade)
```

## Implementation · Validation Before Classification

Validate the input range before applying the classification rule.

```python
if not 0 <= score <= 100:
    print("Invalid score")
elif score >= 60:
    print("Valid passing score")
else:
    print("Valid non-passing score")
```

### Interpretation · Validation Before Classification

Branch order is part of the rule. Validate the allowed range before interpreting a value, and use `==` only when testing equality.
