# Why pandas?

*Represent table data with named columns and compare a manual calculation with pandas.*


## Table Data in Plain Python

A list of dictionaries can store one record per student. To find the
Python mean with familiar Python syntax, visit each record, add its score,
and divide by the number of records.


```python
student_rows = [
    {"name": "Minji", "python": 78, "sql": 85},
    {"name": "Joon", "python": 92, "sql": 88},
    {"name": "Hana", "python": 85, "sql": 94},
]

python_total = 0
for student in student_rows:
    python_total += student["python"]

manual_average = python_total / len(student_rows)
print("Student records:", student_rows)
print("Manual Python average:", manual_average)
```

The Python scores sum to 255. Dividing by three students gives a mean of 85.0.


## Named Columns in a DataFrame

`import pandas as pd` gives the library its usual short name.
`pd.DataFrame(records)` turns the records into a table with row labels
and column labels. `table["column"]` selects a column; `.mean()` summarizes it.

NumPy arrays are useful for calculations on numeric axes. A DataFrame
adds meaningful labels and allows different data types in different columns,
such as student names beside numeric scores.


```python
import pandas as pd

student_df = pd.DataFrame(student_rows)

print("Student table:")
print(student_df)
print("Python average with pandas:", student_df["python"].mean())
print("Means agree:", manual_average == student_df["python"].mean())
```

Both calculations return 85.0. Each row keeps a name together with that student's two scores.


## A Condition for Each Row

Comparing a column with a number produces a Boolean Series: one `True`
or `False` per row. Put this mask inside `table[mask]` to keep matching rows.


```python
high_sql_mask = student_df["sql"] > 90

print("SQL score above 90:")
print(high_sql_mask)
print("Matching students:")
print(student_df[high_sql_mask])
```

Only Hana remains, with SQL score 94. The mask is `False, False, True`, and the selected row keeps its original label 2.


A DataFrame keeps records in a labeled table. Select a named column for a calculation and use a Boolean mask to select records.
