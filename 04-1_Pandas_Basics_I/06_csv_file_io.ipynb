## Locate and Load the Supplied File

`pd.read_csv()` reads the header as column names and parses the rows into
a DataFrame. Use separate input and output directories. `os.makedirs()`
creates the program's output folder before any write.


```python
import os

import pandas as pd

input_directory = os.path.join("assets", "06_csv_file_io")
output_directory = os.path.join("outputs", "06_csv_file_io")
input_path = os.path.join(input_directory, "sample_scores.csv")
output_path = os.path.join(output_directory, "passed_students.csv")
os.makedirs(output_directory, exist_ok=True)

score_df = pd.read_csv(input_path, encoding="utf-8")

print("Input file:", input_path)
print("Loaded scores:")
print(score_df)
print("Shape:", score_df.shape)
print("Average score:", score_df["score"].mean())
```

The supplied file contains four students and four columns. The score mean is 85.5, computed from 88, 76, 93, and 85.


## Select the Rows to Save

This file example uses only the score threshold, at least 85.
Keep the full selected records so the saved file still includes department
and attendance information.


```python
passed_df = score_df.loc[score_df["score"] >= 85].copy()

print("Students reaching the score threshold:")
print(passed_df)
```

`>=` includes equality, so Minji, Hana, and Taeho remain. The selected index labels are 0, 2, and 3.


## Save a CSV without an Extra Index Column

`.to_csv(..., index=False)` writes the data columns and omits the pandas
row index. The default write mode replaces this program-owned output on
repeated runs. The supplied input is never overwritten.


```python
passed_df.to_csv(output_path, index=False, encoding="utf-8")
print("Saved file:", output_path)
```

The saved file is `outputs/06_csv_file_io/passed_students.csv`. In the course site this path is under `/home/pyodide`.


## Reopen and Compare

Reading the output assigns a new default index. `.reset_index(drop=True)`
gives the expected table that same index without adding the old labels
as a data column. `.equals()` compares the resulting tables, including types.


```python
reloaded_df = pd.read_csv(output_path, encoding="utf-8")

print("Reopened table:")
print(reloaded_df)
print("Round trip matches:", reloaded_df.equals(passed_df.reset_index(drop=True)))
print("Original input row count:", len(pd.read_csv(input_path, encoding="utf-8")))
```

The comparison is `True`: all three saved rows and four columns survive the round trip. The input still contains four rows.


Keep supplied inputs and generated outputs separate. Save with `index=False` when row labels are not data, then reopen the file to verify the result.
