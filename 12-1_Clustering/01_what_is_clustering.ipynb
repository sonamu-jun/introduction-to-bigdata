# Clustering

## 1) Concept

- Grouping observations by similarity to identify structure in data
- Unsupervised learning from observation features, without target labels
- Cluster membership determined by the selected features, distance measure, and algorithm
- Classification predicts predefined classes from labeled examples; clustering forms groups without predefined target classes

## 2) Terminology

- Observation: one object or record to be grouped
- Feature: a measured attribute used to describe an observation
- Feature vector: the feature values representing one observation
- Feature space: a space with one axis per feature; two features represent observations as points on a plane
- Distance metric: a rule for calculating separation between observations
- Euclidean distance: straight-line distance, calculated as the square root of the sum of squared feature differences
- Cluster: a group of observations formed under the selected similarity criterion
- Cluster label: an identifier for a group, without numerical order or predefined meaning

## 3) Methodology

1. Selection of features representing the characteristics to compare
2. Comparison of feature units and ranges; standardization when needed by subtracting the mean and dividing by the standard deviation
3. Selection of a distance metric and a grouping criterion, such as proximity to centers or connectivity in dense regions
4. Assignment of observations using the selected algorithm and settings; noise labels where supported
5. Interpretation of group characteristics and comparison of cluster separation and stability

- Larger numerical ranges can give some features greater influence on distance; standardization changes these relative contributions
- K-means requires a cluster count and assigns every observation to a center; DBSCAN uses neighborhood density and can leave observations as noise
- Different features and settings can produce different groupings; no single correct grouping guaranteed
- Classification accuracy unavailable without target labels; cohesion, separation, and stability provide other evaluation criteria

## 4) Example Interpretation

- Same two-feature observations before and after K-means clustering
- Colors indicating assigned clusters, rather than target classes
- Group meaning determined by member characteristics, rather than numerical labels
