# Aggregation and Boolean Filtering

*Summarize the intended axis and use Boolean masks to select values or complete records.*


## Summarizing All Values

A reduction combines multiple values into a summary. With no `axis`,
`sum()`, `mean()`, `min()`, and `max()` summarize the whole array.
In this sales table, rows are branches and columns are products.


```python
import numpy as np

sales = np.array([[120, 80, 95, 110],
                  [100, 90, 105, 115],
                  [130, 70, 100, 120]])

print("Total units:", sales.sum())
print("Mean units per cell:", round(sales.mean(), 2))
print("Smallest cell:", sales.min())
print("Largest cell:", sales.max())
```

The twelve cells total 1,235 units, with a mean of about 102.92 units
per branch-product pair. The smallest and largest cells are 70 and 130.


## Choosing the Reduced Axis

The `axis` argument identifies the dimension to combine. For shape
`(3, 4)`, reducing axis 0 combines the three rows and leaves four
column results. Reducing axis 1 combines the four columns and leaves
three row results.

| Expression | Remaining meaning | Result shape |
|---|---|---|
| `sales.sum(axis=0)` | One total per product | `(4,)` |
| `sales.sum(axis=1)` | One total per branch | `(3,)` |

**Question.** Which axis gives each branch's total, and which branch
position has the largest total?


```python
product_totals = sales.sum(axis=0)
branch_totals = sales.sum(axis=1)

print("Product totals:", product_totals)
print("Branch totals:", branch_totals)
print("Branch means:", sales.mean(axis=1))
```

Use `axis=1` for the branch totals `[405, 410, 420]`. Branch position
2 has the largest total, 420. Product totals are `[350, 240, 300, 345]`;
both sets of totals sum to 1,235.


## Selecting Individual Values

A comparison produces a Boolean array of `True` and `False` values.
Using a mask with the same shape as the data selects matching cells.
The selected values form a one-dimensional array, so their original
row and column positions are no longer represented in the shape.


```python
high_sales_mask = sales >= 110
high_sales_values = sales[high_sales_mask]

print("Cells with at least 110 units:")
print(high_sales_mask)
print("Selected values:", high_sales_values)
print("Selected shape:", high_sales_values.shape)
```

The selected values are `[120, 110, 115, 130, 120]` with shape `(5,)`.
This selection answers which cells meet the condition, not which
complete branches should be retained.


## Selecting Complete Rows

A Boolean mask with one value per row selects complete rows. Build
the mask from branch totals, then apply it to the row axis.
Use `&` for elementwise AND and parentheses around each comparison
when combining array conditions.


```python
selected_branches = sales[branch_totals >= 410, :]
middle_values = sales[(sales >= 100) & (sales < 120)]

print("Branches with at least 410 units:")
print(selected_branches)
print("Selected branch shape:", selected_branches.shape)
print("Cell values from 100 through 119:", middle_values)
```

The second and third branches remain as a `(2, 4)` table. The combined
cell condition selects `[110, 100, 105, 115, 100]`. Choosing a row mask
preserves all product values for each selected branch.


Choose an axis according to the summary you want to retain. A cell mask selects values; a row mask preserves complete records.
