# Lists

*Use a list when several values belong in an ordered, changeable sequence.*

A list is like a row of labeled storage slots. Each item has a position, the first position is index `0`, and the same value may appear more than once. A list is **mutable**, so its items and length can change after creation.

## Creation, Indexing, and Slicing

Square brackets create a list. An index selects one item, while a slice selects a new list containing a range of items. The slice stop index is excluded.

## Adding Items

The methods `append()`, `insert()`, and `extend()` all make the existing list longer.

- `append(value)` adds one item at the end.
- `insert(index, value)` adds one item at a selected position.
- `extend(iterable)` adds every item from another sequence.

## Removing Items

`pop()` removes an item by position and returns it. With no index, it removes the last item. `remove()` finds and deletes the first matching value.

## Finding and Counting Values

`count()` reports how often a value occurs, and `index()` reports the first matching position. The `in` operator answers a membership question with `True` or `False`.

## Copying and Reordering

`copy()` creates an independent list. `sort()` rearranges a list in ascending order, and `reverse()` reverses its current order. These methods change the list on which they are called.

## Nested Lists

A list may contain other lists. The first index selects an inner list, and the second index selects an item inside that list.

## Common List Expressions

Lists also support operators and slice assignment. These expressions are useful when a small program needs to combine, initialize, or replace several items.

### Concatenation and Repetition

`+` creates a new list by joining two lists. `*` creates a new list by repeating the source items.

### Slice Assignment and `del`

Slice assignment replaces a range and may change the list length. The `del` statement removes an item or slice without returning it.

## List Method Coverage

The public list methods can be grouped by purpose.

| Purpose | Methods |
| --- | --- |
| Add items | `append()`, `extend()`, `insert()` |
| Remove items | `remove()`, `pop()`, `clear()` |
| Find values | `index()`, `count()` |
| Reorder items | `sort()`, `reverse()` |
| Make an independent shallow copy | `copy()` |


### Clearing a List

`clear()` removes every item from an existing list without creating a new list. Any other name that refers to the same list observes that change.


A list is the natural choice when order, repeated values, and position-based access matter. Its mutability is convenient, but a copy is needed when later changes must not affect the source list.
