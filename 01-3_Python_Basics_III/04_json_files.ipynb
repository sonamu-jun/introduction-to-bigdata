# JSON Files

*Learn to load, save, and verify structured data with the `json` module.*


## Before You Run

In **Add files**, select `study_profile.json` and enter `assets/04_json_files` for **Save folder**. The file will be available at `/home/pyodide/assets/04_json_files/study_profile.json`.


## Locate and Load

`json.load()` converts JSON from an open text file into Python values.


```python
import json
import os


jupyterlite_base_directory = "/home/pyodide"
chapter_name = "01-3_Python_Basics_III"

if os.path.isdir(jupyterlite_base_directory):
    os.chdir(jupyterlite_base_directory)
elif os.path.basename(os.getcwd()) != chapter_name:
    os.chdir(chapter_name)

base_directory = os.getcwd()

input_directory = os.path.join(base_directory, "assets", "04_json_files")
output_directory = os.path.join(base_directory, "outputs", "04_json_files")
input_path = os.path.join(input_directory, "study_profile.json")
output_path = os.path.join(output_directory, "study_profile_copy.json")

os.makedirs(output_directory, exist_ok=True)

with open(input_path, mode="r", encoding="utf-8") as source_file:
    loaded_profile = json.load(source_file)

print(loaded_profile)
```

The JSON object is now a Python dictionary.


## Save and Verify

`json.dump()` writes the dictionary. Reopen the output with `json.load()` to verify it.


```python
with open(output_path, mode="w", encoding="utf-8") as saved_file:
    json.dump(loaded_profile, saved_file, indent=2)

with open(output_path, mode="r", encoding="utf-8") as saved_file:
    reloaded_profile = json.load(saved_file)

print("Student:", reloaded_profile["student"])
print("Saved correctly:", reloaded_profile == loaded_profile)
```

## Takeaway

Use `json.load()` to read JSON and `json.dump()` to write it.
