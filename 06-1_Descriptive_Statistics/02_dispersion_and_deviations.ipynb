# Dispersion and Deviations

<span style="color: #777777;"><em>산포와 편차</em></span>


- `Dispersion` describes how widely values are spread.  
  <span style="color: #777777;"><em>`산포`(dispersion)는 값들이 얼마나 넓게 퍼져 있는지를 나타냅니다.</em></span>


## Range

<span style="color: #777777;"><em>범위</em></span>

- The `range` is the largest value minus the smallest: $\max(x)-\min(x)$.  
  <span style="color: #777777;"><em>`범위`(range)는 최댓값에서 최솟값을 뺀 값입니다: </em></span>${\color{#777777}\max(x)-\min(x)}$<span style="color: #777777;"><em>.</em></span>


```python
import numpy as np

values = np.array([1, 3, 5, 7, 9])
print("Range:", values.max() - values.min())
```

- The range is 9 − 1 = 8; it uses only the two end values.  
  <span style="color: #777777;"><em>범위는 9 − 1 = 8이며, 양 끝값 두 개만 사용합니다.</em></span>


## Deviation

<span style="color: #777777;"><em>편차</em></span>

- A `deviation` is a value minus the mean: $d_i=x_i-\bar{x}$.  
  <span style="color: #777777;"><em>`편차`(deviation)는 각 값에서 평균을 뺀 값입니다: </em></span>${\color{#777777}d_i=x_i-\bar{x}}$<span style="color: #777777;"><em>.</em></span>

- Positive and negative deviations add to zero.  
  <span style="color: #777777;"><em>양수·음수 편차를 모두 더하면 0입니다.</em></span>


```python
deviations = values - values.mean()
print("Deviations:", deviations)
```

- The mean is 5, so the deviations are -4, -2, 0, 2, and 4.  
  <span style="color: #777777;"><em>평균은 5이므로 편차는 -4, -2, 0, 2, 4입니다.</em></span>


## Variance and Standard Deviation

<span style="color: #777777;"><em>분산과 표준편차</em></span>

- For these values as a full population, `variance` is the mean of squared deviations.  
  <span style="color: #777777;"><em>이 값들을 모집단 전체로 보면, `분산`(variance)은 편차를 제곱한 값의 평균입니다.</em></span>

- `Standard deviation` is the square root of variance, in the original units.  
  <span style="color: #777777;"><em>`표준편차`(standard deviation)는 분산의 제곱근이며 원래 단위로 해석합니다.</em></span>


```python
variance = (deviations**2).mean()
print("Variance:", variance)
print("Std:", round(np.sqrt(variance), 4))
```

- The squared deviations total 40; variance is 40 / 5 = 8 and standard deviation is about 2.8284.  
  <span style="color: #777777;"><em>편차 제곱합은 40이며, 분산은 40 / 5 = 8이고 표준편차는 약 2.8284입니다.</em></span>


## Same Mean, Different Spread

<span style="color: #777777;"><em>같은 평균과 다른 산포</em></span>

- The groups [4, 5, 6] and [1, 5, 9] have the same mean; compare their standard deviations.  
  <span style="color: #777777;"><em>[4, 5, 6]과 [1, 5, 9]는 평균이 같으며, 표준편차로 흩어진 정도를 비교합니다.</em></span>


- Both groups have mean 5; the values 1, 5, and 9 have more spread.  
  <span style="color: #777777;"><em>두 집단의 평균은 5이며, 값이 1·5·9인 집단이 더 넓게 퍼져 있습니다.</em></span>

  <img src="https://raw.githubusercontent.com/sonamu-jun/introduction-to-bigdata/main/06-1_Descriptive_Statistics/assets/02_dispersion_and_deviations/concept.webp" width="439" alt="Both groups have mean 5; the values 1, 5, and 9 have more spread.">


```python
compact = np.array([4, 5, 6])
spread = np.array([1, 5, 9])
print("Compact std:", round(compact.std(), 4))
print("Spread std:", round(spread.std(), 4))
```

- Both means are 5, but the standard deviations are about 0.8165 and 3.2660.  
  <span style="color: #777777;"><em>평균은 모두 5이지만 표준편차는 약 0.8165와 3.2660입니다.</em></span>

- The second group has a larger standard deviation because its values are farther from 5.  
  <span style="color: #777777;"><em>두 번째 집단의 값이 5에서 더 멀리 떨어져 있어 표준편차가 큽니다.</em></span>
