## Method · Literal and Named Values

Text inside quotation marks is a **string literal**. In `course_name = "Introduction to Big Data"`, the equals sign performs assignment; it does not ask whether two values are equal.

```python
print("Hello, world!")

course_name = "Introduction to Big Data"
print(course_name)
```

The first line prints literal text. The second `print()` reads the value currently stored in `course_name`. Quotation marks identify text in the code, but they are not part of the printed result.

## Implementation · Value Combination

`print()` can receive several values separated by commas. This is a simple way to display text and numbers together before learning string formatting.

```python
language_name = "Python"
week_number = 1

print("Course:", course_name)
print("Language:", language_name)
print("Week:", week_number)
```
